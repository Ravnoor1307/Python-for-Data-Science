````python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
````

````python
tips = sns.load_dataset('tips')
tips
````

## 🔬 Deep Dive : Seaborn — figure-level vs axes-level

```python
# axes-level: plot on current ax
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex', style='smoker', size='size')

# figure-level: owns the figure (sns.displot, relplot, catplot, lmplot)
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', hue='sex', style='smoker')
```

- sidebar style: no seaborn minimal, high-level — layer atop matplotlib
- axes-level (scatterplot, lineplot, histplot...) — reuse matplotlib figure/axes grid
- figure-level (relplot/displot/...) — auto-creates multi-panel fig
- hue=col — color encode variable; style= col dashes/markers; size= col point sizes
- Semantic mapping → 4 variables one scatter
- figure-level supports col/row faceting; axes-level not
- seaborn handles stats (KDE smoothing, summary) out of box
- choose: quick expressive (figure-level) vs precise control (axes-level)

````python
# scatter plot -> axes level function
sns.scatterplot(data=tips, x='total_bill', y='tip',hue='sex',style='time',size='size')
````

````python
# relplot -> figure level -> square shape
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter',hue='sex',style='time',size='size')
````

## 🔬 Deep Dive : Line plots + faceting

```python
sns.lineplot(data=temp_df, x='year', y='lifeExp')
sns.relplot(kind='line', data=temp_df, x='year', y='lifeExp', hue='country', style='country')
```

- lineplot — trend over continuous x (year)

- row= — second facet dimension
- Gapminder India/Brazil trend — multi-line compare
- Largest: facet plot = one shortcut for many subplots
- semantics map col vars → visual channels — readability
- .inunching & random samples explore grids

## 🔬 Deep Dive : Distribution plots — hist/kde/rug

```python
sns.histplot(data=tips, x='total_bill')            # axes-level hist
sns.displot(data=tips, x='total_bill', kind='hist', bins=2, hue='sex', element='step')
sns.displot(data=tips, x='total_bill', kind='kde', hue='sex', fill=True)
# rugplot: ticks along axis showing every obs
# bivariate hist / kde: 2D density
```

- Distribution plots — univariate analysis (one col)
- figure-level displot wraps hist/kde/rug/ecdf
- bins — bucket count/edges
- hue — separate distributions per class
- element='step' — outline (overlap readable)
- kde — smooth continuous density (sum of kernels)
- fill, bw_adjust, cut params
- rugplot — every observation tick (all raw positions)
- bivariate (x+y) — 2D hist heat & kde contour
- Compare DAILY numeric spread, age by sex
- Categorical dist can use x=cat with hue num

````python
# plotting univariate histogram
sns.histplot(data=tips, x='total_bill')
````

````python
sns.displot(data=tips, x='total_bill', kind='hist')
````

````python
# bins parameter
sns.displot(data=tips, x='total_bill', kind='hist',bins=2)
````

````python
# It’s also possible to visualize the distribution of a categorical variable using the logic of a histogram. 
# Discrete bins are automatically set for categorical variables

# countplot
sns.displot(data=tips, x='day', kind='hist') 
````

````python
# hue parameter
sns.displot(data=tips, x='tip', kind='hist',hue='sex')
````

````python
# element -> step
sns.displot(data=tips, x='tip', kind='hist',hue='sex',element='step')
````

````python
titanic = sns.load_dataset('titanic')
titanic
````

````python
sns.displot(data=titanic, x='age', kind='hist',element='step',hue='sex')
````

````python
# faceting using col and row -> not work on histplot function
sns.displot(data=tips, x='tip', kind='hist',col='sex',element='step')
````

````python
# kdeplot
# Rather than using discrete bins, a KDE plot smooths the observations with a Gaussian kernel, producing a continuous density estimate
sns.kdeplot(data=tips,x='total_bill')
````

````python
sns.displot(data=tips,x='total_bill',kind='kde')
````

````python
# hue -> fill
sns.displot(data=tips,x='total_bill',kind='kde',hue='sex',fill=True,height=10,aspect=2)
````

````python
# Rugplot

# Plot marginal distributions by drawing ticks along the x and y axes.

# This function is intended to complement other plots by showing the location of individual observations in an unobtrusive way.
sns.kdeplot(data=tips,x='total_bill')
sns.rugplot(data=tips,x='total_bill')
````

````python
# Bivariate histogram
# A bivariate histogram bins the data within rectangles that tile the plot 
# and then shows the count of observations within each rectangle with the fill color

# sns.histplot(data=tips, x='total_bill', y='tip')
sns.displot(data=tips, x='total_bill', y='tip',kind='hist')
````

````python
# Bivariate Kdeplot
# a bivariate KDE plot smoothes the (x, y) observations with a 2D Gaussian
sns.kdeplot(data=tips, x='total_bill', y='tip')
````

## 🔬 Deep Dive : Matrices — heatmap & clustermap

```python
temp_df.pivot(index='region', columns='year', values='lifeExp')
sns.heatmap(matrix, annot=True, linewidths=0.5, cmap='coolwarm')

sns.clustermap(iris.iloc[:, :4])      # rows/cols dendrogram clustering
```

- heatmap — 2D matrix color-encoded (annot adds values)
- pivot first — long→wide rows×cols matrix
- linewidth — cell spacing; cmap palette
- best: correlation matrices, pivot summaries
- clustermap — heatmap + hierarchical clustering dendrograms
- reorders rows/cols to reveal patterns/groups
- rows similar → clusters block diagonal visual
- Useful: find related variables/regions
- annot large matrices overwhelming — subset
- seaborn = publication-ready matrix visuals

````python
# Heatmap

# Plot rectangular data as a color-encoded matrix
temp_df = gap.pivot(index='country',columns='year',values='lifeExp')

# axes level function
plt.figure(figsize=(15,15))
sns.heatmap(temp_df)
````

````python
# annot
temp_df = gap[gap['continent'] == 'Europe'].pivot(index='country',columns='year',values='lifeExp')

plt.figure(figsize=(15,15))
sns.heatmap(temp_df,annot=True,linewidth=0.5, cmap='summer')
````
