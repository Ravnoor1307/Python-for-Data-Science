### Colored Scatterplots

## 📖 Colored scatterplots

```python
plt.scatter(x, y, c=species_codes, cmap='jet', alpha=0.7)
```

- `c=` — numeric category codes (colors)
- `cmap` — colormap (`'jet'`, `'viridis'`, ...)
- `alpha` — transparency 0-1
- Categorical data → numeric codes pehle

````python
iris = pd.read_csv('iris.csv')
iris.sample(5)
````

````python
iris['Species'] = iris['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
iris.sample(5)
````

````python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet',alpha=0.7)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
````

## 🔬 Deep Dive : Colored scatter — 3rd dimension via c

```python
iris['Species'] = iris['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'], c=iris['Species'], cmap='viridis')
```

- c — color by a numeric column (kategorikal labels → numbers first)
- cmap — colormap ('viridis','plasma','RdYlGn','Set1')
- encode category in color — 3rd dimension on 2D plot
- alpha — transparency (overlap visualization)
- colorbar: plt.colorbar() legend for numeric scale
- mapping rule: min→first cmap color, max→last
- categorical dtype color mapping manually
- Use: species/class separation visually
- With hue real dots grouped clusters visible instantly
- Colored scatter = cheap clustering peek

### Plot size

## 📖 Figure size

```python
plt.figure(figsize=(15, 7))
plt.scatter(...)
```

- `figsize=(width, height)` in inches
- Plot render se pehle figure chaho
- Bigger figures — readability

## 🔬 Deep Dive : Figure size & annotations

```python
plt.figure(figsize=(15,7))      # wider canvas
plt.scatter(x, y)
plt.text(1, 5, 'Point 1')       # raw label at coords
plt.annotate('minimum', xy=(min_x,min_y),
             xytext=(10,10), arrowprops=dict(arrowstyle='->'))
```

- plt.figure(figsize=(w,h)) — set canvas inches BEFORE axes
- Larger figsize — readable dense scatter
- text(x,y,label) — free floating label at data point
- annotate — label + arrow to point (xy target, xytext label offset)
- arrowprops — arrow style config
- annotate outliers, extremes, insights
- Contextual labels — inform the analyst
- aspect policy: subplots param for equal axes
- guides for knee points / maxima
- callback patterns sample_df → annotate few

````python
plt.figure(figsize=(15,7))

plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet',alpha=0.7)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
````

### Subplots

## 📖 Horizontal / Vertical lines

```python
plt.axhline(y)   # horizontal reference line
plt.axvline(x)   # vertical reference line
```

- Threshold / baseline dikhane ke liye
- `axhline/axvline` — full-size line, `hlines/vlines` — segments

## 🔬 Deep Dive : Horizontal/vertical lines

```python
plt.axhline(y_value, color='red', linestyle='--')   # horizontal reference
plt.axvline(x_value, color='g', label='threshold')  # vertical reference
plt.axhline(df['avg'].mean(), color='k', linewidth=2)
```

- axhline — infinite horizontal reference line (y)
- axvline — vertical corresponding
- Highlight thresholds — avg, target, cutoff
- cv lines full-width across axes
- dash style to not confuse with data
- Multiple thresholds — multiple calls
- legend labels → line meaning
- Used in anomaly flagging & decision boundaries
- Pair with annotation for reasoned labels
- easier than plotting short arrays as vertical segs

## 📖 Subplots

```python
fig, ax = plt.subplots(nrows=2, ncols=2, sharex=True, figsize=(10,10))
ax[0,0].scatter(...)
ax[0,1].plot(...)

fig = plt.figure()
ax1 = fig.add_subplot(2,2,1)
```

- `subplots(nrows, ncols)` — grid of axes
- `ax[row, col]` indexing (or `ax[0]` for 1-D)
- `sharex/sharey` — axis shared
- `fig.add_subplot(r,c,i)` — positional alternative

````python
fig,ax = plt.subplots(figsize=(15,6))

ax.scatter(batters['avg'],batters['strike_rate'],color='red',marker='+')
ax.set_title('Something')
ax.set_xlabel('Avg')
ax.set_ylabel('Strike Rate')

fig.show()
````

## 🔬 Deep Dive : Subplots — multiple figures one window

```python
fig, ax = plt.subplots(nrows=2, ncols=1, sharex=True, figsize=(10,6))
ax[0].scatter(batters['avg'], batters['strike_rate'])
ax[1].scatter(batters['matches'], batters['strike_rate'])

fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(10,10))
ax[0,0].plot(x1,y1); ax[0,1].plot(x2,y2); ax[1,0].plot(x3,y3); ax[1,1].plot(x4,y4)
```

- subplots(nrows, ncols) → fig + array of Axes
- Plot on each ax explicitly: ax[i].plot/scatter/set_xlabel
- sharex/sharey — same scale across for compare
- indexing: 1D row-major (ax[0], ax[1]); 2D (ax[0,0]...)
- fig.add_subplot(r,c,index) — manual incremental
- Titles per axis via ax[i].set_title
- Layout: aligned comparison & storytelling rows
- tweak: fig.tight_layout() spacing
- Multi-panel reports standard
- AXES-LEVEL control core of matplotlib

### 3D Scatter Plots

## 📖 3D plots

```python
ax = plt.subplot(projection='3d')
ax.scatter(x, y, z)
```

- `projection='3d'` — 3D axes
- scatter / line / surface all possible
- 3 cols data → 3D space

## 🔬 Deep Dive : 3D plots & surfaces

```python
fig = plt.figure()
ax = plt.subplot(projection='3d')
ax.scatter(x, y, z)          # 3D scatter
ax.plot(x, y, z)             # 3D line

xx, yy = np.meshgrid(x, y)   # 2D grids for surface
z = xx**2 + yy**2
ax.plot_surface(xx, yy, z, cmap='viridis')     # surface
ax.contour(xx, yy, z)        # contour lines
```

- projection='3d' — new axes, third dim
- meshgrid — all (x,y) coordinate pairs grid
- plot_surface — height map surface
- contour, contourf — level curves top-down
- 3D reveals 3-variable relationships
- Contour = 2D projection of height
- functions of two variables visualized
- Rotatable interactive (notebook)
- depth perception — helpful overview
- Extra dims beyond 3 → animation/color code

````python
batters

fig = plt.figure()

ax = plt.subplot(projection='3d')

ax.scatter3D(batters['runs'],batters['avg'],batters['strike_rate'],marker='+')
ax.set_title('IPL batsman analysis')

ax.set_xlabel('Runs')
ax.set_ylabel('Avg')
ax.set_zlabel('SR')
````

## 📖 Heatmaps (imshow)

```python
plt.imshow(grid)
plt.colorbar()
plt.yticks(...); plt.xticks(...)
```

- `imshow` — 2D matrix → color-encoded grid
- `grid` pivot_table / matrix shape me chahiye
- `colorbar` — value scale
- Categorical axes labels ticks set karo

## 🔬 Deep Dive : Heatmap — imshow matrix

```python
grid = temp_df.pivot_table(index='overs', columns='ballnumber', values='batsman_runs', aggfunc='mean')
plt.imshow(grid)
plt.yticks(delivery['overs'].unique(), fontsize=8)   # label rows
plt.colorbar()                                        # value scale
```

- imshow(matrix) — 2D grid as color cells
- pivot_table → matrix (rows×cols) first
- color = value magnitude immediate
- yticks/xticks — map row/col indexes to labels
- colorbar — legend scale
- annotate numeric (seaborn.heatmap annot=True better)
- Use: run-scoring patterns, correlation matrices
- Dark = low, bright = high (or reverse cmap)
- density overview compact — heat map thousands values
- seaborn heatmap has annot + nicer defaults

### Pandas Plot()

## 📖 Pandas plot()

```python
s.plot(kind='pie')
df.plot(kind='scatter', x='col1', y='col2')
df.plot(kind='bar', stacked=True)
df.plot(kind='line', x='Date')
df[['m1','m2']].plot(kind='hist', bins=40)
df[['m1','m2','m3']].plot(kind='pie', subplots=True)
```

- DataFrame/Series ka `.plot()` — matplotlib wrapper
- `kind` — pie/bar/line/hist/scatter...
- Dataframe pe index hi default x-axis
- `subplots=True` — har col ka separate plot
- Stacked bars, grouped plots — readable analysis

## 🔬 Deep Dive : pandas plot() — quick charts

```python
s.plot(kind='pie')                    # single series
tips.plot(kind='scatter', x='total_bill', y='tip', figsize=(10,6), legend=True)
stocks.plot(kind='line', x='Date')
temp.plot(kind='bar', stacked=True)
stocks.plot(kind='hist', bins=30)
df[['match1','match2']].plot(kind='pie', subplots=True, autopct='%0.1f%%')
```

- .plot(kind=...) — DataFrame/Series native quick chart
- kinds: line, bar, barh, pie, hist, box, scatter, hexbin, kde, area
- subplots=True — each col separate panel
- legend, figsize, stacked params pass-through
- Time x values — line plots fine
- dataset m-mehdi pandas tutorial online (dataset var)
- wide/multiindex pivots also plot-able
- fast EDA defaults over control fidelity
- choose when prototyping visuals vs custom matplotlib
- input dims: numeric cols

````python
# on a series

s = pd.Series([1,2,3,4,5,6,7])
s.plot(kind='pie')
````
