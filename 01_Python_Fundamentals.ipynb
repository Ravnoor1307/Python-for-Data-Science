## Python Data Types

Every value in Python has a **data type**. Understanding data types is fundamental because:
- Different types support different operations
- Data is stored differently in memory
- Mixing incompatible types causes errors

### Python's Built-in Data Types:

| Category | Type | Example | Mutable? |

#### Integer (`int`)

Whole numbers without any decimal point. Python supports **arbitrarily large** integers - there is no upper limit, unlike C/C++ or Java.

````python
# Integer
print(8)
# 1*10^308
print(1e309)
````


`1e309` means `1 x 10^309`. When an integer exceeds Python's float range, it becomes `inf` (infinity).

#### Float (Decimal Numbers)

Numbers with decimal points. Stored using IEEE 754 double-precision format.

````python
# Decimal/Float
print(8.55)
print(1.7e309)
````


`1.7e309` exceeds the maximum float value, so Python returns `inf`.

#### Boolean (`bool`)

Booleans represent one of two values: `True` or `False`. In Python, `bool` is actually a **subclass of `int`**:
- `True` has the value `1`
- `False` has the value `0`

````python
# Boolean
print(True)
print(False)
````

````python
# Booleans are subclass of int!
print(True + 1)   # 2 (True = 1)
print(False + 1)  # 1 (False = 0)
print(True * 5)   # 5
print(False * 5)  # 0
````

## Variables

Variables are **named containers** that store data. Think of them as labeled boxes where you put values.

###  Static vs Dynamic Typing

| Feature | Static Typing (C/Java) | Dynamic Typing (Python) |
|---------|----------------------|------------------------|
| Type declaration | Required: `int a = 5` | Not needed: `a = 5` |
| Type flexibility | Fixed once declared | Can change anytime |
| Type errors | Caught at compile time | Caught at runtime |

````python
# Static Vs Dynamic Typing
# Static Vs Dynamic Binding
# stylish declaration techniques
````

````python
# Basic variable assignment
name = 'noor'
print(name)

a = 5
b = 6
print(a + b)
````

````python
# Dynamic Typing - Python infers the type automatically
a = 5
# Static Typing - In C/Java you would write: int a = 5
# Python doesn't need this!
````

###  Dynamic Binding

In Python, a variable can **change its type** during execution. This is called **dynamic binding**.

````python
# Dynamic Binding - variable changes type at runtime
a = 5          # a is an integer
print(a)       # 5
a = 'nitish'   # now a is a string!
print(a)       # nitish

# In C/Java (Static Binding), this would cause a compile-time error
# because 'a' was declared as int and can't become a string
````

### Interview Question: What is Dynamic Typing in Python?

**Answer:** Dynamic typing means:
1. You don't need to declare variable types before using them
2. A variable can hold any type of data
3. The type can change during program execution
4. Type checking happens at runtime, not compile time

```python
# Python (Dynamic Typing)
x = 5        # x is int
x = 'hello'  # x is now str - perfectly valid!

// Java (Static Typing)
int x = 5;
x = 'hello'; // ERROR! Can't assign string to int
```

## Type Conversion (Type Casting)

Type conversion means changing a value from one data type to another.

###  Implicit Type Conversion
Python **automatically** converts one type to another when it's safe. This is done by Python itself.

###  Explicit Type Conversion
You **manually** convert types using built-in functions like `int()`, `float()`, `str()`, etc.

````python
# Implicit Vs Explicit
print(5+5.6)
print(type(5),type(5.6))
````

````python
# This will cause an error - can't implicitly convert string to int
# print(4 + '4')  # TypeError!
print('Use explicit conversion: 4 + int("4") =', 4 + int('4'))
````

###  Explicit Type Conversion

````python
# Explicit
# str -> int
#int(4+5j)  # TypeError! Can't convert complex to int

# int to str
print(str(5))          # '5'
print(type(str(5)))    # <class 'str'>

# float
print(float(4))        # 4.0
print(type(float(4)))  # <class 'float'>
````

````python
# More explicit conversion examples

# str to int
print(int('10'))       # 10

# float to int (truncates, does NOT round!)
print(int(9.99))       # 9
print(int(9.5))        # 9
print(int(-9.99))      # -9

# str to float
print(float('3.14'))   # 3.14

# int/float to bool
print(bool(0))         # False
print(bool(1))         # True
print(bool(''))        # False (empty string)
print(bool('hello'))   # True (non-empty string)
````



| Conversion | Function | Example | Result |
|-----------|----------|---------|--------|
| int to float | `float()` | `float(5)` | `5.0` |
| float to int | `int()` | `int(9.99)` | `9` (truncates!) |
| int to str | `str()` | `str(5)` | `'5'` |
| str to int | `int()` | `int('10')` | `10` |
| str to float | `float()` | `float('3.14')` | `3.14` |

**Common Errors:**
- `int('hello')` -> `ValueError` (not a valid number)
- `int('3.14')` -> `ValueError` (use `float('3.14')` instead)
- `int(5+6j)` -> `TypeError` (can't convert complex to int)

###  The `None` Literal

`None` represents the **absence of a value**. It is Python's way of saying "nothing here".

- It is of type `NoneType`
- Always use `is None` (not `== None`) to check for it
- `None` is different from `0`, `''` (empty string), or `False`

````python
k = None
a = 5
b = 6
print('Program exe')
````

````python
# None is different from other falsy values
print(None is None)      # True (preferred way to check)
print(type(None))        # <class 'NoneType'>

print(None == 0)         # False
print(None == '')        # False
print(None == False)     # False

print(0 is None)         # False
print('' is None)        # False
print(False is None)     # False
````

### Q6: Dogs and Chickens (Heads and Legs Problem)

Write a program that will tell the number of dogs and chickens when the user provides the value of total heads and legs.

**Logic:**
- Each animal has 1 head
- Dog has 4 legs, Chicken has 2 legs
- If heads = `h` and legs = `l`:
  - `chickens = (4*h - l) / 2`
  - `dogs = h - chickens`

**Example:** Input: heads = 4, legs = 12 -> Output: dogs = 2, chicken = 2

**Concept:** User Input + Mathematical Logic

````python
# Write your code here
heads = int(input('Enter total heads: '))
legs = int(input('Enter total legs: '))

# Chicken has 2 legs, Dog has 4 legs
# chickens + dogs = heads
# 2*chickens + 4*dogs = legs

chickens = (4 * heads - legs) / 2
dogs = heads - chickens

print(f'dogs -> {int(dogs)} chicken -> {int(chickens)}')
````

**Explanation:**
- `chickens = (4*4 - 12) / 2 = (16-12)/2 = 4/2 = 2`
- `dogs = 4 - 2 = 2`
- Verify: 2 dogs (8 legs) + 2 chickens (4 legs) = 12 legs, 4 heads

Given the height, width and breadth of a milk tank, find out how many glasses of milk can be obtained. Assume all the inputs are provided by the user.

**Tank Volume:** `V_tank = H * L * B`

**Glass Volume (cylinder):** `V_glass = pi * r^2 * h`

**Number of glasses:** `floor(V_tank / V_glass)`

**Concept:** User Input + Math Module + Volume Calculations

````python
# Write your code here
import math

h_t = float(input('height'))
b_t = float(input('breadth'))
l_t = float(input('length'))

h_g = float(input('height of glass'))
r_g = float(input('radius of the glass'))

vol_tank = h_t*b_t*l_t
vol_glass = 3.14*r_g*r_g*h_g

print('no of glasses',math.floor(vol_tank/vol_glass))
````

**Explanation:**
- Tank volume: `10 * 19 * 10 = 1900`
- Glass volume: `3.14 * 2^2 * 5 = 3.14 * 4 * 5 = 62.8`
- Number of glasses: `floor(1900 / 62.8)` = `floor(30.25)` = `30`

**Note:** `math.floor()` rounds down to the nearest integer (we can't have partial glasses!)
