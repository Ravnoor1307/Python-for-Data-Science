### Theory : All Operator Families

**Arithmetic** — normal math. Watch the tricky three :
```python
5 / 2   # true division  -> 2.5
5 // 2  # floor division -> 2   (rounds down)
5 % 2   # modulus (remainder) -> 1
5 ** 2  # power -> 25
```

**Relational (Comparison)** — answer is always a bool :
`4 > 5` F · `4 < 5` T · `4 >= 4` T · `4 <= 4` T · `4 == 4` T · `4 != 4` F

**Logical** — and / or / not on truth-things :
```python
1 and 0    # -> 0   (Python returns the actual value, not True/False!)
1 or 0     # -> 1
not 1      # -> False
```
- `x and y` : if x is False → x ; else → y
- `x or y`  : if x is True  → x ; else → y
- `not x`   : always a bool.

**Bitwise** — operate on the BINARY digits :
```python
2 & 3   # 010 & 011 = 010 -> 2
2 | 3   # 010 | 011 = 011 -> 3
2 ^ 3   # 010 ^ 011 = 001 -> 1
~3      # NOT -> -(3+1) = -4   (two's complement trick!)
4 >> 2  # 100 >> 2 = 001 -> 1  (right shift = divide by 2^n)
5 << 2  # 101 << 2 = 10100 -> 20 (left shift = multiply by 2^n)
```

**Assignment** — `=` stores a value. Compound forms repeat the operation:
```python
a = 2
a %= 2     # a = a % 2  -> 0
```
Note → `++` / `--` do NOT exist in Python.

**Membership** — check if something exists in a container :
```python
'D' in 'Delhi'        # True
'D' not in 'Delhi'    # False
1 in [2,3,4,5,6]      # False
```

### Theory : Loops - while & for

**Why loops?** Repetition. Without loops you would copy-paste code 1000 times for 1000 items.

**while** — repeat AS LONG AS a condition stays True.
```python
i = 1
while i < 11:                # runs 10 times
    print(i)
    i += 1                   # IMPORTANT: update the counter or infinite loop!
```

**while … else** — the else runs once when the loop finishes NORMALLY (no break, no crash).
```python
x = 1
while x < 3:
    print(x)
    x += 1
else:
    print('limit crossed')
```

**for** — iterate over a sequence element by element.
```python
for i in {1,2,3,4,5}:   # works on str, list, tuple, set, dict, range...
    print(i)
```

**for + range(start, stop, step)** — the classic counted loop :
```python
for i in range(10, 0, -1):    # 10,9,8,...,1
    print(i)
```

**while vs for — which one?**
- **for** when you know the collection / count (iterate over items).
- **while** when you keep going until some condition flips (games, menus, "keep asking till 0").


````python
# While loop example -> program to print the table
# Program -> Sum of all digits of a given number
# Program -> keep accepting numbers from users till he/she enters a 0 and then find the avg
````

````python
number = int(input('enter the number'))

i = 1

while i<11:
  print(number,'*',i,'=',number * i)
  i += 1
````

### Program - The current population of a town is 10000. The population of the town is increasing at the rate of 10% per year. You have to write a program to find out the population at the end of each of the last 10 years.

### Theory : The Population Trick

If population grows 10% every year :
> Next year's population = current × 1.1

Going BACKWARDS in time (finding the past) must UNDO that ×1.1 → **divide by 1.1** :

```python
previous = current / 1.1
```

**Why not subtract 10%?**
```python
wrong = current - 0.1 * current    # off by a little every year - errors accumulate!
correct = current / 1.1
```
Subtracting 10% is not the inverse of adding 10% (percentages multiply, they don't add).
Start from current = 10000 and loop 10 years : print year · population, then update `curr_pop /= 1.1`.


````python
curr_pop = 10000

for i in range(10,0,-1):
  print(i,curr_pop)
  curr_pop = curr_pop - 0.1*curr_pop
````

# Nested Loops

### Theory : Nested Loops

A loop INSIDE a loop. The **inner loop completes fully for every single pass of the outer loop**.

```python
for i in range(1, 5):      # outer
    for j in range(1, 5):  # inner
        print(i, j)        # all 16 pairs - runs 4 × 4 = 16 times
```

**Use cases :** pairs/combinatmons, matrices, and printing PATTERNS.

**Patterns rule :** outer loop = number of ROWS, inner loop = what happens in each row.
Use `end=''` to build a row without line breaks, then one plain `print()` to break the line.

```python
# right triangle of stars : row i prints i stars
for i in range(1, rows + 1):
    for j in range(1, i + 1):
        print('*', end='')
    print()
```

Number pyramid 1 / 121 / 12321 : two inner loops - one counting up 1..i, one counting down i-1..1.

# Loop Control Statement

- Break
- Continue
- Pass

### Theory : break, continue, pass

**break** — immediately EXIT the whole loop.
```python
for i in range(1, 10):
    if i == 5:
        break        # loop stops - prints 1 2 3 4
    print(i)
```

**continue** — SKIP this iteration, jump to the next one.
```python
for i in range(1, 10):
    if i == 5:
        continue     # 5 is skipped - prints 1 2 3 4 6 7 8 9
    print(i)
```

**pass** — do NOTHING. A placeholder for code you will write later. (An empty block is a syntax error; pass fills it.)

**for … else + break — the prime-finder pattern :**
```python
for i in range(lower, upper + 1):
    for j in range(2, i):
        if i % j == 0:
            break          # i has a divisor → i is NOT prime
    else:
        print(i)           # no break → no divisor → prime!
```
The inner `else` executes ONLY when the inner loop ended without a break. This is the clean pythonic prime test.


## 🔬 Deep Dive : break


- `break` — poori loop terminate, next code line pe

- Inner loop me use → sirf inner exit hota
- Usage: early stop (prime search, found item)
- `break` ke baad loop `else` nahi chalta (agar else block ho)
- vs continue: break=exit, continue=skip one
