## Exercise 1 (Easy)

Sum every multiple of 3 or 5 below 1000 and assert the answer is 233168.

```python
total = sum(n for n in range(1000) if n % 3 == 0 or n % 5 == 0)
assert total == 233168
print(total)
```

## Exercise 2 (Easy)

Write `count_vowels(text)` using a loop; assert it returns 5 for 'data science'.

```python
def count_vowels(text):
    return sum(1 for ch in text.lower() if ch in 'aeiou')

assert count_vowels('data science') == 5
print(count_vowels('data science'))
```

## Exercise 3 (Easy)

Print the first 10 Fibonacci numbers into a list `fib` and assert fib[-1] == 34.

```python
fib = [0, 1]
while len(fib) < 10:
    fib.append(fib[-1] + fib[-2])
assert fib[-1] == 34
print(fib)
```

## Exercise 4 (Medium)

Write `is_prime(n)` and assert it flags 97 as prime and 91 as not prime.

```python
def is_prime(n):
    if n < 2:
        return False
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            return False
    return True

assert is_prime(97)
assert not is_prime(91)
print('prime check ok')
```

## Exercise 5 (Medium)

Reverse the digits of 12345 using a while loop (no slicing). Assert 54321.

```python
n, rev = 12345, 0
while n:
    n, digit = divmod(n, 10)
    rev = rev * 10 + digit
assert rev == 54321
print(rev)
```

## Exercise 6 (Medium)

Use for-else to search for 'gold' in a list and report 'not found' when missing.

```python
items = ['iron', 'copper', 'silver']
for item in items:
    if item == 'gold':
        result = 'found'
        break
else:
    result = 'not found'
assert result == 'not found'
print(result)
```

## Exercise 7 (Hard)

Build a pyramid of stars of height 5 as a list of strings; assert the last row has 9 stars.

```python
rows = [('*' * (2 * i - 1)).center(9) for i in range(1, 6)]
assert rows[-1].strip().count('*') == 9
print('\n'.join(rows))
```
