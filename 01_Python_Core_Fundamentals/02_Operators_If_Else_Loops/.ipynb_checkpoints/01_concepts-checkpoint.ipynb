## 1. Comparison chaining and the is/== distinction

Related idea: **Operators**. Arithmetic, comparison, logical, membership (`in`), identity (`is`) and bitwise. `is` compares object identity and must never be used for value comparison except against `None`.

```python
# Comparison chaining and the is/== distinction
x = 15
print('chained:', 10 < x < 20)
a = [1, 2]; b = [1, 2]
print('a == b:', a == b, '| a is b:', a is b)
print('correct None check:', a is not None)
```

## 2. Short-circuit evaluation prevents errors

Related idea: **Conditional statements**. `if/elif/else` evaluates top to bottom and stops at the first true branch. Order your conditions from most specific to most general, and prefer early returns over deep nesting.

```python
# Short-circuit evaluation prevents errors
values = []
if values and values[0] > 0:      # values[0] never runs when empty
    print('positive first value')
else:
    print('safely skipped the index lookup')
print('or-default:', (values or ['fallback'])[0])
```

## 3. if/elif/else: order matters

Related idea: **Loops and iteration**. `for` iterates over an iterable; `while` repeats until a condition fails. In Python you almost always want `for`, often with `enumerate`, `zip` or `range`.

```python
# if/elif/else: order matters
def grade(score):
    if score >= 90:
        return 'A'
    elif score >= 75:
        return 'B'
    elif score >= 60:
        return 'C'
    return 'F'

for s in [95, 80, 65, 30]:
    print(s, '->', grade(s))
```

## 4. The ternary expression and match statement

Related idea: **Loop control**. `break` exits, `continue` skips to the next iteration, and the rarely used `for ... else` block runs only when the loop was not broken - perfect for search.

```python
# The ternary expression and match statement
n = 7
print('even' if n % 2 == 0 else 'odd')

def describe(value):
    match value:
        case 0:
            return 'zero'
        case int() if value < 0:
            return 'negative integer'
        case int():
            return 'positive integer'
        case _:
            return 'not an integer'

print([describe(v) for v in [0, -3, 9, 'x']])
```

## 5. for with range, enumerate and zip

Related idea: **Operators**. Arithmetic, comparison, logical, membership (`in`), identity (`is`) and bitwise. `is` compares object identity and must never be used for value comparison except against `None`.

```python
# for with range, enumerate and zip
names = ['asha', 'ben', 'chen']
scores = [88, 92, 79]
for i, (name, score) in enumerate(zip(names, scores), start=1):
    print(f'{i}. {name.title():<6} {score:>3}')
print('squares:', [i ** 2 for i in range(1, 6)])
```

## 6. while loop with a guaranteed exit condition

Related idea: **Conditional statements**. `if/elif/else` evaluates top to bottom and stops at the first true branch. Order your conditions from most specific to most general, and prefer early returns over deep nesting.

```python
# while loop with a guaranteed exit condition
balance, target, month = 10000, 20000, 0
while balance < target and month < 200:
    balance *= 1.01
    month += 1
print(f'reached {balance:,.0f} after {month} months at 1% per month')
```

## 7. break, continue and the for-else search idiom

Related idea: **Loops and iteration**. `for` iterates over an iterable; `while` repeats until a condition fails. In Python you almost always want `for`, often with `enumerate`, `zip` or `range`.

```python
# break, continue and the for-else search idiom
primes = []
for n in range(2, 30):
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            break
    else:
        primes.append(n)
print('primes under 30:', primes)
```

## 8. Nested loops: a multiplication table

Related idea: **Loop control**. `break` exits, `continue` skips to the next iteration, and the rarely used `for ... else` block runs only when the loop was not broken - perfect for search.

```python
# Nested loops: a multiplication table
for row in range(1, 6):
    line = ' '.join(f'{row * col:>3}' for col in range(1, 6))
    print(line)
```
