## Example 1: FizzBuzz, the classic screening question

Goal: fizzbuzz, the classic screening question. Watch the shapes and types printed by the cell - they are the cheapest correctness check you have.

```python
# FizzBuzz, the classic screening question
def fizzbuzz(n):
    out = []
    for i in range(1, n + 1):
        if i % 15 == 0:
            out.append('FizzBuzz')
        elif i % 3 == 0:
            out.append('Fizz')
        elif i % 5 == 0:
            out.append('Buzz')
        else:
            out.append(str(i))
    return out

print(fizzbuzz(20))
```

## Example 2: Classify transactions into spending buckets

Goal: classify transactions into spending buckets. Watch the shapes and types printed by the cell - they are the cheapest correctness check you have.

```python
# Classify transactions into spending buckets
transactions = [('rent', 18000), ('coffee', 180), ('laptop', 62000), ('bus', 40)]
for label, amount in transactions:
    if amount >= 50000:
        bucket = 'major purchase'
    elif amount >= 5000:
        bucket = 'large'
    elif amount >= 500:
        bucket = 'medium'
    else:
        bucket = 'small'
    print(f'{label:<8} {amount:>7,} -> {bucket}')
```

## Example 3: Find the first duplicate in a stream of ids

Goal: find the first duplicate in a stream of ids. Watch the shapes and types printed by the cell - they are the cheapest correctness check you have.

```python
# Find the first duplicate in a stream of ids
def first_duplicate(ids):
    seen = set()
    for value in ids:
        if value in seen:
            return value
        seen.add(value)
    return None

print(first_duplicate([4, 9, 2, 9, 7]))
print(first_duplicate([1, 2, 3]))
```

## Example 4: Collatz sequence length - a loop with unknown bounds

Goal: collatz sequence length - a loop with unknown bounds. Watch the shapes and types printed by the cell - they are the cheapest correctness check you have.

```python
# Collatz sequence length - a loop with unknown bounds
def collatz_steps(n):
    steps = 0
    while n != 1:
        n = n // 2 if n % 2 == 0 else 3 * n + 1
        steps += 1
    return steps

for start in [6, 7, 27]:
    print(f'{start:>3} reaches 1 in {collatz_steps(start)} steps')
```

## Example 5: Running totals and a simple moving average

Goal: running totals and a simple moving average. Watch the shapes and types printed by the cell - they are the cheapest correctness check you have.

```python
# Running totals and a simple moving average
sales = [120, 135, 98, 160, 145, 170, 155]
window = 3
for i in range(window - 1, len(sales)):
    chunk = sales[i - window + 1:i + 1]
    print(f'days {i - window + 2}-{i + 1}: avg {sum(chunk) / window:.1f}')
```

## Example 6: Validate a password against several rules at once

Goal: validate a password against several rules at once. Watch the shapes and types printed by the cell - they are the cheapest correctness check you have.

```python
# Validate a password against several rules at once
def check_password(pw):
    problems = []
    if len(pw) < 8:
        problems.append('too short')
    if not any(ch.isdigit() for ch in pw):
        problems.append('needs a digit')
    if not any(ch.isupper() for ch in pw):
        problems.append('needs an uppercase letter')
    if not any(ch in '!@#$%^&*' for ch in pw):
        problems.append('needs a symbol')
    return problems or ['ok']

for pw in ['abc', 'Password1', 'Str0ng!pass']:
    print(f'{pw:<14} {check_password(pw)}')
```
