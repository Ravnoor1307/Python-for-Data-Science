# 1. Python Output

### Theory : The print() function

Python has a built-in function called print() that writes text to the screen.

```python
print(value1, value2, ... , sep=" ", end="\n")
```

**Rules to remember :**
1. You can pass many values in one print() - they print one after another.
2. **sep** = the string placed *between* two values. Default is a space.
3. **end** = the string placed *after* the last value. Default is a newline (\n).
4. Every value is automatically converted to a string before printing.
5. Strings MUST be quoted. `print(Ravnoor Singh)` is a **SyntaxError** because Python reads Ravnoor and Singh as variable names.

**Demos :**
- `print('Hello',1,4.5,True)` → `Hello 1 4.5 True`
- `print('Hello',1,4.5,True,sep='/')` → `Hello/1/4.5/True`
- `print('hello',end='-')` + `print('world')` → `hello-world`


## 🔬 Deep Dive : print() 

**`print()` ka kaam** — values ko console/output me dikhana.

```python
print('Hello World')            # ek string
print('Hello', 1, 4.5, True)    # multiple values, space se judte
print('Hello', 1, 4.5, True, sep='/')   # sep = separator char
print('hello', end='-')         # end = final char (default '\n')
print('hello'); print('world')  # agar end nahi diya → nayi line
```

Key points:
- **sep** — values ke beech ka character (default space)
- **end** — last ke baad kya aaye (default newline)
- Python **case sensitive** hai — `Print` chalega nahi
- Quotes ke bina string error: `print(Ravnoor Singh)` → NameError
- Multiple types ek saath allowed — type checking nahi
- Kabhi internally har value ka `str()` bula kar concatenate karta hai

**Yaad rakhne wali trick**: `print('a', 'b')` → `a b`, `print('a', 'b', sep='')` → `ab`.

# 2. Data Types

### Theory : Core Data Types

Every value in Python has a **type** and that type decides what you can do with it.

**1. Numeric types**
- `int` : whole numbers → 8, -3, 0, 10**100 (no size limit!)
- `float` : decimal numbers → 8.55, 1.5e2 (scientific). Overflow prints `inf`.
- `complex` : real + imaginary part → 5+6j. Access `.real` and `.imag`.

**2. Collection types**
- `str` (string) : text → 'Hello World'
- `list` : ordered, CHANGEABLE → [1,2,3,4,5]  (like a flexible C array)
- `tuple` : ordered, IMMUTABLE → (1,2,3,4,5)
- `set` : unordered, all values UNIQUE → {1,2,3,4,5}
- `dict` : key:value pairs → {'name':'Nitish','gender':'Male','weight':70}

**3. Boolean type**
- `bool` : only True and False.

Use `type(x)` anytime to ask "what is the type of this value?".

# 3. Variables

### Theory : Variables, Dynamic Typing & Binding

A **variable** is a name that refers to a value kept in memory.

**Static vs Dynamic typing**
- Static (C/C++/Java) : you must declare the type → `int a = 5;`
- Dynamic (Python) : `a = 5` - Python decides the type at run time. Writing `int a = 5` gives a **SyntaxError**.

**Static vs Dynamic binding**
- The same name can point to different types while the program runs:
```python
a = 5          # int
a = 'nitish'   # now binds to a string - allowed
```
C/C++ would reject this (static binding). Python allows it = **dynamic binding**.

**Stylish assignment techniques**
```python
a = 1; b = 2; c = 3      # one per line
a, b, c = 1, 2, 3        # tuple unpacking / parallel assignment
a = b = c = 5            # chained assignment - all get 5
```


## 🔬 Deep Dive : Variables

```python
name = 'nitish'     # no type declaration needed
a = 5               # DYN TYPING — type set karte waqt
a = 'nitish'        # DYN BINDING — ab a string ban gaya
a, b, c = 1, 2, 3   # multiple in one line
a = b = c = 5       # chain assignment
```

**Static vs Dynamic Typing**:
- Static (C/Java): `int a = 5` — type fixed, change allowed nahi
- Dynamic (Python): type runtime pe decide hota hai

**Static vs Dynamic Binding**:
- `a = 5`; `a = 'nitish'` — Python me name ko naye type se rebind kar sakte ho
- C me aise type change → error

**Memory model**: variable ≠ box. Variable = **name tag** jo kisi object ko refer karta hai. `a = 5` matlab object `5` ko tag `a` lagaya.

# 4. Keywords & Identifiers

### Theory : Keywords & Identifiers

**Keywords** = words reserved by the language. You cannot use them as variable names.
Python 3 has 35 keywords → get the live list:
```python
import keyword
print(keyword.kwlist)
```

**Identifiers** = names you create. Rules :
1. Only letters, digits, underscore `_` allowed.
2. Cannot start with a digit → `1name` invalid.
3. Cannot be a keyword.
4. Case-sensitive → `name`, `Name`, `NAME` = three different things.
5. `_` is a valid identifier (used for "garbage" values).

**Conventions (PEP 8) :** variables → `snake_case`; classes → `PascalCase`.


## 🔬 Deep Dive : Keywords & Identifiers

```python
import keyword
keyword.kwlist          # list of all reserved words

name1 = 'Nitish'        # identifier
second-name = 5         # ❌ hyphen not allowed
2ndname = 5             # ❌ digit se start nahi
```

- **Keywords** — language reserved (`if`, `for`, `class`, `return`...). Variable naam nahi bana sakte.
- **Identifiers** — custom names (variables, functions)
- Identifier rules:
  - letters, digits, `_` se bante
  - digit se **start nahi** kar sakte
  - keywords use nahi kar sakte
  - case sensitive (`name` ≠ `Name`)
  - `_private`, `__mangled`, `__dunder__` conventions aage

````python
# Keywords
import keyword
keyword.kwlist
````

# 6. Type Conversion

### Theory : Type Conversion (Casting)

**Implicit** - Python converts automatically when safe :
```python
print(5 + 5.6)   # int + float -> float (10.6). Smaller type widens.
```

**Explicit** - you force conversion with constructors :
- `int(x)` : float→int (truncates decimals), str→int (digits only)
- `float(x)` : int→float, str→float
- `str(x)` : ANY value → string
- `bool(x)` : 0, '', None, [] → False ; everything else → True

**Errors you will meet :**
- `int('4abc')` → ValueError
- `int(4+5j)` → TypeError
- `int('4.5')` → ValueError (use float('4.5') first)
- `4 + '4'` → TypeError (int + str illegal)

**Golden rule :** an int and a string can NEVER be added unless you cast one of them first.


## 🔬 Deep Dive : Type Conversion

```python
print(5 + 5.6)        # 10.6  Implicit — int→float promote
print(4 + 'x')        # ❌ TypeError (found, required)

str(5)                # '5'    Explicit
int('1234')           # 1234
float('12.5')         # 12.5
list((1,2))           # [1,2]
```

**Implicit** — Python khud convert (int+float → float). Narrow→wide.
**Explicit (casting)** — hum bata ke convert karte:
- `int()` — str/float → int (decimal ko truncate)
- `float()` → float
- `str()` → string (har value ko)
- `bool()` — 0/''/None/[] → False, baaki True

Edge cases: `int(4+5j)` → error; `int('12.5')` → error (decimal string int() nahi leta).

````python
# Implicit Vs Explicit
print(5+5.6)
print(type(5),type(5.6))

print(4 + '4')
````

````python
# Explicit
# str -> int
#int(4+5j)

# int to str
str(5)

# float
float(4)
````

# 7. Literals

### Theory : Literals

A **literal** is a value written directly in the code.

**Numeric literals**
- Binary `0b1010` (=10) · Decimal `100` · Octal `0o310` (=200) · Hexa `0x12c` (=300)
- Float `10.5` · Scientific `1.5e2` (=150.0) · `1.5e-3` (=0.0015)
- Complex `3.14j` → `x.real` = 0.0, `x.imag` = 3.14

**String literals**
- `'single'` `"double"` `'''triple'''` `"""triple"""` (triple = multi-line)
- Unicode → `u"\U0001f600"` (emoji everywhere)
- Raw → `r"raw \\n string"` — backslashes stay literal (paths, regex)

**Boolean & None**
- `True` behaves as 1, `False` as 0 → `True + 4` = 5, `False + 10` = 10
- `None` = "nothing / not yet assigned".


````python
a = 0b1010 #Binary Literals
b = 100 #Decimal Literal 
c = 0o310 #Octal Literal
d = 0x12c #Hexadecimal Literal

#Float Literal
float_1 = 10.5 
float_2 = 1.5e2 # 1.5 * 10^2
float_3 = 1.5e-3 # 1.5 * 10^-3

#Complex Literal 
x = 3.14j

print(a, b, c, d)
print(float_1, float_2,float_3)
print(x, x.imag, x.real)
````
