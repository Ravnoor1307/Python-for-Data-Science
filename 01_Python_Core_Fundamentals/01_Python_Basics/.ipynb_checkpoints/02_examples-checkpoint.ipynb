## Example 1: Compute a restaurant bill with tax and tip

Goal: compute a restaurant bill with tax and tip. Watch the shapes and types printed by the cell - they are the cheapest correctness check you have.

```python
# Compute a restaurant bill with tax and tip
subtotal = 1250.00
gst_rate = 0.05
tip_rate = 0.10
gst = subtotal * gst_rate
tip = subtotal * tip_rate
total = subtotal + gst + tip
print(f'Subtotal: {subtotal:>10,.2f}')
print(f'GST 5%  : {gst:>10,.2f}')
print(f'Tip 10% : {tip:>10,.2f}')
print(f'Total   : {total:>10,.2f}')
```

## Example 2: Convert temperature between scales

Goal: convert temperature between scales. Watch the shapes and types printed by the cell - they are the cheapest correctness check you have.

```python
# Convert temperature between scales
def c_to_f(celsius):
    return celsius * 9 / 5 + 32

def f_to_c(fahrenheit):
    return (fahrenheit - 32) * 5 / 9

for c in [-40, 0, 37, 100]:
    print(f'{c:>5}C = {c_to_f(c):>6.1f}F  (round trip {f_to_c(c_to_f(c)):.1f}C)')
```

## Example 3: Split a bill between friends, handling the remainder in paise

Goal: split a bill between friends, handling the remainder in paise. Watch the shapes and types printed by the cell - they are the cheapest correctness check you have.

```python
# Split a bill between friends, handling the remainder in paise
total_paise = 100000  # rupees 1000.00
people = 3
share, remainder = divmod(total_paise, people)
shares = [share] * people
for i in range(remainder):
    shares[i] += 1
print('shares (rupees):', [s / 100 for s in shares])
assert sum(shares) == total_paise
```

## Example 4: Simple interest vs compound interest over 10 years

Goal: simple interest vs compound interest over 10 years. Watch the shapes and types printed by the cell - they are the cheapest correctness check you have.

```python
# Simple interest vs compound interest over 10 years
principal, rate, years = 100000, 0.08, 10
simple = principal * (1 + rate * years)
compound = principal * (1 + rate) ** years
print(f'Simple  : {simple:,.0f}')
print(f'Compound: {compound:,.0f}')
print(f'Gain from compounding: {compound - simple:,.0f}')
```

## Example 5: Validate and normalise a phone number typed by a user

Goal: validate and normalise a phone number typed by a user. Watch the shapes and types printed by the cell - they are the cheapest correctness check you have.

```python
# Validate and normalise a phone number typed by a user
def normalise_phone(raw):
    digits = ''.join(ch for ch in str(raw) if ch.isdigit())
    if len(digits) == 12 and digits.startswith('91'):
        digits = digits[2:]
    return digits if len(digits) == 10 else None

for raw in ['98765-43210', '+91 98765 43210', '12345', '(987) 654 3210']:
    print(f'{raw:>20} -> {normalise_phone(raw)}')
```

## Example 6: Rounding: banker's rounding surprises people

Goal: rounding: banker's rounding surprises people. Watch the shapes and types printed by the cell - they are the cheapest correctness check you have.

```python
# Rounding: banker's rounding surprises people
for value in [0.5, 1.5, 2.5, 3.5]:
    print(value, '-> round():', round(value))
print('round(2.675, 2) =', round(2.675, 2), '(float representation, not a bug)')
from decimal import Decimal, ROUND_HALF_UP
print('half-up:', Decimal('2.675').quantize(Decimal('0.01'), rounding=ROUND_HALF_UP))
```
