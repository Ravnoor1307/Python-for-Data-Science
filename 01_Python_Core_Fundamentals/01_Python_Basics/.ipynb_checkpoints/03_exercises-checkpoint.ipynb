## Setup

## Exercise 1 (Easy)

Bind `radius = 7` and compute the area of a circle into `area`, rounded to two decimals. Assert it equals 153.94.

```python
radius = 7
area = round(math.pi * radius ** 2, 2)
print(area)
assert area == 153.94
```

## Exercise 2 (Easy)

Write `to_int(value, default=0)` that converts a string to int and returns the default on failure. Test it on '10', 'x' and None.

```python
def to_int(value, default=0):
    try:
        return int(str(value).strip())
    except (ValueError, TypeError):
        return default

assert to_int('10') == 10
assert to_int('x') == 0
assert to_int(None, -1) == -1
print('to_int works')
```

## Exercise 3 (Easy)

Swap two variables without a temporary variable, then assert the swap happened.

```python
a, b = 'left', 'right'
a, b = b, a
assert (a, b) == ('right', 'left')
print(a, b)
```

## Exercise 4 (Medium)

Format 1234567.891 as an Indian-style currency string with two decimals and commas: '1,234,567.89'.

```python
value = 1234567.891
text = f'{value:,.2f}'
assert text == '1,234,567.89'
print(text)
```

## Exercise 5 (Medium)

Given `seconds = 9875`, convert to h/m/s using divmod and assert the tuple is (2, 44, 35).

```python
seconds = 9875
minutes, s = divmod(seconds, 60)
h, m = divmod(minutes, 60)
assert (h, m, s) == (2, 44, 35)
print(f'{h}h {m}m {s}s')
```

## Exercise 6 (Medium)

Demonstrate that floats are approximate: assert 0.1 + 0.2 != 0.3 but that math.isclose says they match.

```python
assert 0.1 + 0.2 != 0.3
assert math.isclose(0.1 + 0.2, 0.3)
print('difference:', abs(0.1 + 0.2 - 0.3))
```

## Exercise 7 (Hard)

Write `safe_divide(a, b)` returning None when b is zero, otherwise the quotient.

```python
def safe_divide(a, b):
    return None if b == 0 else a / b

assert safe_divide(10, 4) == 2.5
assert safe_divide(1, 0) is None
print('ok')
```

## Exercise 9 (Hard)

Write a function `is_ready()` that returns True only when you can explain Python Basics without notes. Flip the flag honestly.

```python
def is_ready(explained_without_notes: bool = True) -> bool:
    return bool(explained_without_notes)

assert is_ready(True) is True
assert is_ready(False) is False
print('self assessment recorded')
```
