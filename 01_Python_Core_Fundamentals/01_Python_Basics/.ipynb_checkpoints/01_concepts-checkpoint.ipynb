## 1. Names are bindings, not boxes

Related idea: **Variables and dynamic typing**. A variable is a name bound to an object, not a labelled box. Rebinding a name never changes the old object; it points the name somewhere else. Types live on objects, which is why `x = 5` then `x = 'five'` is legal.

```python
# Names are bindings, not boxes
a = 5
b = a          # both names point at the same int object
a = 'five'     # rebinding a leaves b untouched
print(a, b, type(a).__name__, type(b).__name__)
```

## 2. int is arbitrary precision, float is not

Related idea: **Numeric types**. `int` has arbitrary precision, `float` is IEEE-754 double precision and therefore approximate, `complex` exists for signal work. Money should never be a float; use `decimal.Decimal`.

```python
# int is arbitrary precision, float is not
big = 2 ** 200
print('2**200 has', len(str(big)), 'digits')
print('0.1 + 0.2 ==', 0.1 + 0.2)
print('exactly equal to 0.3?', 0.1 + 0.2 == 0.3)
print('safe comparison:', math.isclose(0.1 + 0.2, 0.3))
```

## 3. Money must not be a float

Related idea: **Type conversion**. Explicit conversion with `int()`, `float()`, `str()` and `bool()` is safe and readable. Implicit coercion in Python is deliberately minimal, which prevents a whole class of silent bugs.

```python
# Money must not be a float
from decimal import Decimal, getcontext
getcontext().prec = 10
price = Decimal('19.99')
qty = 3
print('float  :', 19.99 * 3)
print('decimal:', price * qty)
```

## 4. Explicit conversion and its failure mode

Related idea: **Input and output**. `input()` always returns a string, so numeric input needs conversion. f-strings are the modern formatting tool and support alignment, precision and expressions.

```python
# Explicit conversion and its failure mode
raw = '42'
print(int(raw) + 8)
for bad in ['4.5', 'abc', '']:
    try:
        int(bad)
    except ValueError as err:
        print(f'int({bad!r}) failed: {err}')
```

## 5. f-strings: alignment, precision, expressions

Related idea: **Variables and dynamic typing**. A variable is a name bound to an object, not a labelled box. Rebinding a name never changes the old object; it points the name somewhere else. Types live on objects, which is why `x = 5` then `x = 'five'` is legal.

```python
# f-strings: alignment, precision, expressions
name, score = 'Asha', 0.87345
print(f'{name:<10}|{score:>8.2%}')
print(f'{name.upper()} scored {score * 100:.1f} points')
print(f'{1234567:,}')
print(f'{255:#x} {255:#o} {255:#b}')
```

## 6. input() returns a string - simulate it safely for a notebook

Related idea: **Numeric types**. `int` has arbitrary precision, `float` is IEEE-754 double precision and therefore approximate, `complex` exists for signal work. Money should never be a float; use `decimal.Decimal`.

```python
# input() returns a string - simulate it safely for a notebook
def read_int(raw, default=0):
    """Parse an integer from user input, falling back to a default."""
    try:
        return int(raw.strip())
    except (ValueError, AttributeError):
        return default

for raw in ['7', ' 12 ', 'oops', None]:
    print(repr(raw), '->', read_int(raw))
```

## 7. Truthiness: what counts as False

Related idea: **Type conversion**. Explicit conversion with `int()`, `float()`, `str()` and `bool()` is safe and readable. Implicit coercion in Python is deliberately minimal, which prevents a whole class of silent bugs.

```python
# Truthiness: what counts as False
falsy = [0, 0.0, '', [], {}, set(), None, False]
for value in falsy:
    assert not value
print('all of these are falsy:', falsy)
print('non-empty string is truthy:', bool('0'))
```

## 8. Operator precedence surprises

Related idea: **Input and output**. `input()` always returns a string, so numeric input needs conversion. f-strings are the modern formatting tool and support alignment, precision and expressions.

```python
# Operator precedence surprises
print(2 ** 3 ** 2)        # right-associative: 2**9
print(-3 ** 2)            # unary minus binds looser than **
print(10 - 3 - 2)         # left-associative
print(True + True)        # bool is a subclass of int
```
