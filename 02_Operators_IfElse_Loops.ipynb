# Operators, If-Else, Modules & Loops


## Operators in Python

Operators are **special symbols** that perform operations on values (operands).



### Types of Operators

| **Category**   | **Operators**                          | **Description**            |
|----------------|----------------------------------------|----------------------------|
| **Arithmetic** | `+`, `-`, `*`, `/`, `//`, `%`, `**`    | Mathematical operations    |
| **Relational** | `==`, `!=`, `>`, `<`, `>=`, `<=`       | Comparison between values  |
| **Logical**    | `and`, `or`, `not`                     | Boolean logic              |
| **Bitwise**    | `&`, `^`, `~`, `<<`, `>>`         | Binary-level operations    |
| **Assignment** | `=`, `+=`, `-=`, `*=`, `/=`, `%=`, `**=`, `//=` | Assign/update values |
| **Membership** | `in`, `not in`                         | Check existence in sequence|




#### Arithmetic Operators

`````python
# Arithmetic Operators

print(5+6)    # Addition:        11

print(5-6)    # Subtraction:     -1

print(5*6)    # Multiplication:  30

print(5/2)    # True Division:   2.5

print(5//2)   # Floor Division:  2

print(5%2)    # Modulus:         1

print(5**2)   # Exponentiation:  25
`````



| Operator | Name | Example | Result |
|----------|------|---------|--------|
| `+` | Addition | `5+6` | `11` |
| `-` | Subtraction | `5-6` | `-1` |
| `*` | Multiplication | `5*6` | `30` |
| `/` | True Division | `5/2` | `2.5` (always float) |
| `//` | Floor Division | `5//2` | `2` (rounds down) |
| `%` | Modulus (remainder) | `5%2` | `1` |
| `**` | Exponentiation | `5**2` | `25` |

#### Relational (Comparison) Operators

`````python
# Relational Operators

print(4>5)    # Greater than:          False

print(4<5)    # Less than:             True

print(4>=4)   # Greater than or equal: True

print(4<=4)   # Less than or equal:    True

print(4==4)   # Equal to:              True

print(4!=4)   # Not equal to:          False
`````


All relational operators return a **Boolean** (`True` or `False`).

#### Logical Operators

`````python
# Logical Operators

print(1 and 0)    # AND: both must be True -> 0 (False)

print(1 or 0)     # OR: at least one True -> 1 (True)

print(not 1)      # NOT: reverses the value -> False
`````


| Operator | Description | Example | Result |
|----------|-------------|---------|--------|
| `and` | True if both are True | `1 and 0` | `0` (False) |
| `or` | True if at least one is True | `1 or 0` | `1` (True) |
| `not` | Reverses the value | `not 1` | `False` |

**Note:** In Python, `0` is False and any non-zero number is True.

#### Bitwise Operators

`````python
# Bitwise Operators

# bitwise and
print(2 & 3)    # AND: 10 & 11 = 10 = 2

# bitwise or
print(2 | 3)    # OR:  10 | 11 = 11 = 3

# bitwise xor
print(2 ^ 3)    # XOR: 10 ^ 11 = 01 = 1

print(~3)       # NOT: -(3+1) = -4

print(4 >> 2)   # Right Shift: 4 / 2^2 = 1

print(5 << 2)   # Left Shift:  5 * 2^2 = 20
`````


| Operator | Description | Example | Binary | Result |
|----------|-------------|---------|--------|--------|
| `&` | AND | `2 & 3` | `10 & 11 = 10` | `2` |
| `|` | OR | `2 | 3` | `10 | 11 = 11` | `3` |
| `^` | XOR | `2 ^ 3` | `10 ^ 11 = 01` | `1` |
| `~` | NOT | `~3` | `-(3+1)` | `-4` |
| `>>` | Right Shift | `4 >> 2` | `4 / 2^2` | `1` |
| `<<` | Left Shift | `5 << 2` | `5 * 2^2` | `20` |

#### Assignment Operators

`````python
# Assignment Operators

a = 2

a %= 2    # a = a % 2 -> 0

print(a)
`````


| Operator | Equivalent |
|----------|------------|
| `=` | `a = b` |
| `+=` | `a = a + b` |
| `-=` | `a = a - b` |
| `*=` | `a = a * b` |
| `/=` | `a = a / b` |
| `//=` | `a = a // b` |
| `%=` | `a = a % b` |
| `**=` | `a = a ** b` |


#### Membership Operators

`````python
# Membership Operators (in/not in)

print('D' not in 'Delhi')    # False ('D' IS in 'Delhi')

print(1 in [2,3,4,5,6])      # False (1 is NOT in the list)
`````


| Operator | Description | Example | Result |
|----------|-------------|---------|--------|
| `in` | True if value is in sequence | `'D' in 'Delhi'` | `True` |
| `not in` | True if value is not in sequence | `'D' not in 'Delhi'` | `False` |

# Loops in Python

Loops let you **repeat code** multiple times.

### Types:

### While Loop

`````python
# While loop - Print multiplication table
number = int(input('enter the number'))
i = 1

while i < 11:
    print(number,'*',i,'=',number * i)
    i += 1
`````

### While Loop with Else

`````python
# while loop with else
x = 1
while x < 3:
    print(x)
    x += 1
else:
    print('limit crossed')
`````


The `else` block runs when the while condition becomes False (without `break`).

### Guessing Game

`````python
# Guessing game
import random
jackpot = random.randint(1,100)
guess = int(input('guess karo'))
counter = 1

while guess != jackpot:
    if guess < jackpot:
        print('galat! guess higher')
    else:
        print('galat! guess lower')
    guess = int(input('guess karo'))
    counter += 1
else:
    print('correct guess')
    print('attempts',counter)
`````

# Nested Loops

`````python
# Pattern 1
for i in range(1,5):
    for j in range(i):
        print('*',end=' ')
    print()
`````

#### Population Growth Calculation

To calculate the population for each previous year when there is a **10% annual increase**, we can use a simple equation based on the population of the previous year.

Let the population of the previous year be represented by **x**.

##### 1. Calculate the Current Year's Population

If the population increases by **10%** every year:

```text
Current Year Population = x × 1.1
````

Here:

* `x` = Population of the previous year
* `1.1` = 100% + 10% increase

For example, if the previous year's population is `10,000`:

```text
Current Year Population = 10,000 × 1.1
                        = 11,000
```

##### 2. Calculate the Previous Year's Population

If we already know the current year's population and want to find the population of the previous year, rearrange the equation:

```text
Current Year Population = x × 1.1
```

Divide both sides by `1.1`:

```text
x = Current Year Population / 1.1
```

Therefore:

```text
Previous Year Population = Current Year Population / 1.1
```

##### Example

Suppose the current year's population is:

```text
10,000
```

Then the previous year's population would be:

```text
Previous Year Population = 10,000 / 1.1
                         ≈ 9,090.91
```

##### 3. Finding Population for Multiple Previous Years

The same calculation can be repeated for every previous year:

```text
Previous Year 1 = Current Population / 1.1

Previous Year 2 = Previous Year 1 / 1.1

Previous Year 3 = Previous Year 2 / 1.1
```

In general:

```text
Previous Year Population = Current Year Population / 1.1
```

This method allows us to calculate the population for each previous year when the population is assumed to have increased by **10% every year**.

### Problem 13: Armstrong Numbers in Range

Print all Armstrong numbers in a given range.

**Armstrong number:** Sum of digits each raised to power n equals the number itself.

Example: `153 = 1^3 + 5^3 + 3^3 = 153`

`````python
# Write code here
start = int(input("Enter the start of the range: "))
end = int(input("Enter the end of the range: "))

for num in range(start, end+1):
    temp = num
    su_m = 0
    order = len(str(num))
    while temp > 0:
        digit = temp % 10
        su_m += digit ** order
        temp //= 10
    if num == su_m:
        print(num)
`````


**Explanation:**
- `order = len(str(num))` gets the number of digits
- Extract each digit with `% 10`, raise to power `order`, sum them
- If sum equals original number, it's an Armstrong number

### Problem 14: Clock Angle Between Hands

Calculate the minimum angle between hour and minute hand.

**Key facts:**
- 1 hour = 30 degrees (360/12)
- 1 minute = 6 degrees (360/60)
- Hour hand moves 0.5 degrees per minute

`````python
# Write code here
h = int(input('Enter hours hand-'))
m = int(input('Enter minute hand-'))

# validate the input
if (h < 0 or m < 0 or h > 12 or m > 60):
    print('Wrong input')

# minute angle (1 min = 6 degrees)
m_angle = m*6

# hour angle (1 hour = 30 degrees + 0.5 per minute)
h_angle = h*30 + m*0.5

# Take abs difference
angle = abs(h_angle - m_angle)

if angle>180:
    print(360-angle)
else:
    print(angle)
`````


**Explanation:**
- Hour hand at 9: `9 * 30 = 270` degrees
- Minute hand at 0: `0 * 6 = 0` degrees
- Difference: `270` degrees
- Since 270 > 180: `360 - 270 = 90` degrees
