## 1. What is Data Wrangling?

**Data Wrangling** (also called Data Munging) is the process of transforming raw data into a clean, usable format suitable for analysis.

```
Raw Data  ->  Assess  ->  Clean  ->  Analyze  ->  Visualize
```

It typically involves:
- **Discovering** - Understanding what's in the data
- **Structuring** - Reshaping data into the right format
- **Cleaning** - Fixing quality issues
- **Enriching** - Adding more data or features
- **Validating** - Checking that the cleaned data meets quality standards
- **Publishing** - Making the cleaned dataset available

Data scientists spend roughly **60-80%** of their time on wrangling - it's the most time-consuming part of the pipeline!

## 2. Dirty Data vs Messy Data

These two terms are often confused. Let's clarify:

### 2.1 Dirty Data
Dirty data contains **errors or inaccuracies** - the data values themselves are wrong.

Examples:
- Misspelled names
- Duplicate rows
- Invalid entries (e.g. age = -5)
- Missing values (NaN, empty cells)
- Outliers that don't belong
- Inconsistent formatting (e.g. NY, N.Y., New York)
- Data entry errors (e.g. M in an age column)

**Dirty data is about content** - the values are incorrect.

### 2.2 Messy Data
Messy data has **structural problems** - the layout or format makes it hard to work with.

Examples:
- Column headers used as values instead of variable names
- Multiple variables stored in one column
- Variables stored in both rows and columns
- Observation units stored across multiple tables
- Metadata stored as data

**Messy data is about structure** - the data is organized poorly.

### 2.3 Tidy Data
**Tidy Data** (Hadley Wickham, 2014) is the opposite of messy data:

1. Each variable forms a **column**
2. Each observation forms a **row**
3. Each type of observational unit forms a **table**

```python
# Messy (wide format)
# | Person | Jan | Feb | Mar |
# | Alice  |  10 |  15 |  12 |

# Tidy (long format)
# | Person | Month | Value |
# | Alice  | Jan   |    10 |
# | Alice  | Feb   |    15 |
# | Alice  | Mar   |    12 |
```

## 3. Data Quality Dimensions

These are the key dimensions used to evaluate data quality:

| Dimension | Description |
|-----------|-------------|
| **Accuracy** | Do values correctly represent reality? |
| **Completeness** | Are all required values present? |
| **Consistency** | Is data the same across different sources/tables? |
| **Timeliness** | Is data up-to-date and available when needed? |
| **Validity** | Does data conform to defined formats/rules? |
| **Uniqueness** | Are there duplicate records? |

When assessing data, always check each dimension systematically.

## 4. Data Cleaning Order of Operations

The recommended order for cleaning:

```
1. Back up your original data first!
2. Remove duplicates
3. Fix structural errors (typos, case, wrong column names)
4. Handle missing data (drop, impute, or flag)
5. Handle outliers (decide: remove, cap, or keep)
6. Validate (check against known constraints)
```

**Always work on a copy** - never modify the original data without a backup.

## 5. The Define, Code, Test Framework

For every cleaning task, follow this three-step process:

### Step 1: Define
- Clearly state the cleaning issue
- Document what needs to be fixed and why

### Step 2: Code
- Write the code to fix the issue
- Always use a copy of the DataFrame

### Step 3: Test
- Verify that the cleaning worked as expected
- Use assertions, visual checks, or statistical summaries

```python
# Define: Missing values in 'age' column
# Code: Impute with median
df['age'] = df['age'].fillna(df['age'].median())
# Test: Confirm no NaN remains
assert df['age'].isna().sum() == 0
```

### 6.1 Assessment Summary

Based on our exploration, here are the issues we found:

**patients.csv issues:**
- `zip_code` has non-standard format (should be 5-digit, some have decimals)
- Some columns have inconsistent formatting
- `address` column contains personal identifiable information

**treatments.csv issues:**
- Column names are unclear (a, b, c, d) - these represent: start_date, end_date, hba1c_start, hba1c_end, hba1c_change
- Missing values in `hba1c_change`
- `hba1c_change` should be calculated as `hba1c_start - hba1c_end`
- Data split across two files (`treatments.csv` and `treatments_cut.csv`)

**adverse_reactions.csv issues:**
- Relatively clean but needs to be verified against treatments data

### 6.2 Cleaning Steps

Let's follow the **Define, Code, Test** framework for each issue.

#### Issue 1: Merge the two treatment tables

````python
# Define: treatments data is split across two files
# Code: Concatenate them
treatments_combined = pd.concat([treatments, treatments_cut], ignore_index=True)
print(f'treatments: {treatments.shape}, treatments_cut: {treatments_cut.shape}')
print(f'combined: {treatments_combined.shape}')
````

````python
# Test
assert treatments_combined.shape[0] == treatments.shape[0] + treatments_cut.shape[0]
print('Merge verified successfully')
````

#### Issue 2: Rename columns

````python
# Define: Column names a, b, c, d are unclear
# Code: Rename them
treatments_combined = treatments_combined.rename(columns={
    'a': 'start_date',
    'b': 'end_date',
    'c': 'hba1c_start',
    'd': 'hba1c_end'
})
treatments_combined.head()
````

````python
# Test
assert 'start_date' in treatments_combined.columns
assert 'hba1c_start' in treatments_combined.columns
print('Column rename verified')
````

#### Issue 3: Fix hba1c_change

````python
# Define: hba1c_change should be calculated from start and end values
# Code: Calculate it
treatments_combined['hba1c_change'] = treatments_combined['hba1c_start'] - treatments_combined['hba1c_end']
````

````python
# Test
print(f"Missing hba1c_change: {treatments_combined['hba1c_change'].isna().sum()}")
treatments_combined[['hba1c_start', 'hba1c_end', 'hba1c_change']].head()
````

#### Issue 4: Fix zip_code format

````python
# Define: zip_code has decimals and inconsistent format
# Code: Convert to string and remove decimals
patients['zip_code'] = patients['zip_code'].astype(str).str.replace('.0', '', regex=False)
````

````python
# Test
print(patients['zip_code'].head(10))
````

#### Issue 5: Verify data consistency across tables

````python
# Define: Treatment records should match patients
# Code: Check for orphan records
treatment_patient_ids = set(treatments_combined['patient_id'])
patient_ids = set(patients['patient_id'])

orphan_treatments = treatment_patient_ids - patient_ids
orphan_patients = patient_ids - treatment_patient_ids

print(f'Treatment records without patient: {len(orphan_treatments)}')
print(f'Patients without treatment: {len(orphan_patients)}')
````

````python
# Check adverse reactions are linked to valid treatments
adverse_patient_ids = set(adverse_reactions['patient_id'])
adverse_in_treatments = adverse_patient_ids - treatment_patient_ids
print(f'Adverse reactions from patients not in treatments: {len(adverse_in_treatments)}')
````

### 6.3 Final Cleaned Data

````python
print('=== Patients ===')
print(f'Shape: {patients.shape}')
print(f'Columns: {list(patients.columns)}')
print()
print('=== Treatments (Combined) ===')
print(f'Shape: {treatments_combined.shape}')
print(f'Columns: {list(treatments_combined.columns)}')
print()
print('=== Adverse Reactions ===')
print(f'Shape: {adverse_reactions.shape}')
print(f'Columns: {list(adverse_reactions.columns)}')
````

### 7.1 Assessment - Audible

**Issues found:**
- `Time` column is in string format (e.g., '2 hrs 30 mins') - needs conversion to numeric
- `Releasedate` may need datetime conversion
- `Rating` and `Price` may need cleaning
- `Author` and `Narrator` may have inconsistent formatting

### 7.2 Cleaning Steps - Audible

````python
# Define: Time column is string format
# Code: Extract hours and minutes, convert to total minutes
def parse_time(time_str):
    if pd.isna(time_str):
        return np.nan
    total_minutes = 0
    time_str = str(time_str).lower()
    if 'hr' in time_str:
        hours = int(time_str.split('hr')[0].strip())
        total_minutes += hours * 60
    if 'min' in time_str:
        mins_part = time_str.split('min')[0].strip()
        if mins_part:
            mins = mins_part.split()[-1]
            total_minutes += int(mins)
    return total_minutes if total_minutes > 0 else np.nan

audible['Time_minutes'] = audible['Time'].apply(parse_time)
````

````python
# Test
print(audible[['Time', 'Time_minutes']].head(10))
````

````python
# Define: Releasedate should be datetime
# Code: Convert
audible['Releasedate'] = pd.to_datetime(audible['Releasedate'], errors='coerce')
````

````python
# Test
print(audible['Releasedate'].dtype)
print(audible['Releasedate'].head())
````

````python
# Define: Price may have non-numeric values
# Code: Clean price column
non_numeric = audible[~audible['Price'].str.replace('.','',regex=False).str.isnumeric()]['Price'].unique()
print(f'Non-numeric prices: {non_numeric}')
````

````python
audible['Price_clean'] = pd.to_numeric(
    audible['Price'].str.replace('Free','0',regex=False).str.replace(',','',regex=False),
    errors='coerce'
)
````

````python
audible.head()
````

### 7.3 Key Cleaning Functions Reference

| Function | Purpose |
|----------|---------|
| `df.drop_duplicates()` | Remove duplicate rows |
| `df.dropna()` | Drop rows with missing values |
| `df.fillna()` | Fill missing values |
| `df.replace()` | Replace specific values |
| `df.rename()` | Rename columns |
| `df.astype()` | Convert data types |
| `pd.to_datetime()` | Convert to datetime |
| `pd.to_numeric()` | Convert to numeric |
| `df.apply()` | Apply function to each row/column |
| `df.str.contains()` | String pattern matching |
| `df.str.replace()` | String replacement |
| `df.info()` | Check dtypes and missing values |
| `df.describe()` | Statistical summary |
| `df.value_counts()` | Count unique values |
| `df.duplicated()` | Identify duplicates |
| `df.isna()` | Identify missing values |

---
## Key Takeaways

1. **Always assess before cleaning** - use `info()`, `describe()`, `isna()`, `duplicated()` to understand the data
2. **Define, Code, Test** - follow this framework for every cleaning task
3. **Back up your data** - work on copies, not originals
4. **Dirty vs Messy** - know the difference (content errors vs structural problems)
5. **Tidy Data** - each variable is a column, each observation is a row
6. **Clean systematically** - duplicates, structural errors, missing values, outliers, validate
7. **Document everything** - keep track of what you changed and why
