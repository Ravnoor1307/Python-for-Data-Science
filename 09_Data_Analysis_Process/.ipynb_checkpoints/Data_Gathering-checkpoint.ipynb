## What is Data Analysis?

**Data Analysis** is the systematic process of
**inspecting, cleaning, transforming, and modelling** data with the goal of
**discovering useful information, drawing conclusions, and supporting decision-making**.

It is not just "making charts". A real analysis answers a **business or scientific question**
and produces an **actionable recommendation**.

### Why does it matter?
| Reason | Explanation |
|--------|-------------|
| Evidence over opinion | Decisions based on data instead of guesswork |
| Find hidden patterns | Relationships that are invisible in raw data |
| Predict the future | Forecasting sales, churn, risk, demand |
| Measure performance | KPIs like revenue, profit, conversion, retention |
| Tell a story | Communicate findings so others act on them |

### Data → Information → Insight → Decision
```text
Raw Data   (unstructured, messy numbers)
   ↓  (clean, organise)
Information (structured, meaningful values)
   ↓  (analyze, compare)
Insight    (what it means, why it matters)
   ↓  (act)
Decision   (business action / recommendation)
```




##  The Complete Data Analysis Process — Overview

Every data project, no matter how big or small, follows the same loop.
The steps are **iterative** — you often jump back a step when you learn something new.

```text
┌───────────────────────────────────────────────────────────────┐
│                THE DATA ANALYSIS PROCESS                      │
├───────────────────────────────────────────────────────────────┤
│                                                               │
│    0. DEFINE THE PROBLEM  (ask a clear question)              │
│              ↓                                                │
│    1. DATA GATHERING      (collect from sources)   ← THIS FILE│
│              ↓                                                │
│    2. DATA CLEANING       (fix errors, missing, dup)          │
│              ↓                                                │
│    3. DATA TRANSFORMATION (engineer features, encode, scale)  │
│              ↓                                                │
│    4. EXPLORE (EDA)       (summaries, patterns, correlations) │
│              ↓                                                │
│    5. DATA MODELING       (ML / statistics / predictions)     │

│              ↓                                                │

└───────────────────────────────────────────────────────────────┘
```

> ⚠️ This is a **guide**, not a prison. In practice the steps overlap and loop
> constantly. But every project goes through these stages at least once.



##  Step 0 — Define the Problem (before touching any data)

The single most important step. **You cannot gather the right data if you
do not know the question you are trying to answer.**

### SMART questions
- ❌ *"Analyze the sales data."* (too vague)
- ✅ *"Which product categories are losing money, and why?"* (actionable)

### What to define
1. **Objective** — what decision will this support?
2. **Key questions** — the specific questions to answer (e.g. "who are our most valuable customers?")
3. **Metrics / KPIs** — how success is measured (revenue, profit, churn...)
4. **Data needed** — which variables you will have to collect
5. **Stakeholders** — who will act on the result

### Deliverable
A short **problem statement** with the questions you will answer.
Everything downstream (what data to gather, what to clean, what to plot) is driven by this statement.





##  Step 1 — Data Gathering

**Definition:** The process of **collecting and importing** the raw data you need
from their original sources into your analysis environment (Python / Pandas).

### Why it is the foundation
- No data → no analysis.
- The **quality of your analysis can never exceed the quality of your data** (garbage in, garbage out).
- Different sources demand different tools, which is exactly what this notebook teaches.

### The five main sources
| Source | Typical format | Main Python tool |
|--------|----------------|------------------|
| Local / cloud files | CSV, Excel, JSON, TSV, text | `pd.read_csv()`, `pd.read_excel()`, `pd.read_json()` |
| Databases | SQL, NoSQL | `pd.read_sql()`, `mysql.connector`, `sqlalchemy` |
| Web APIs | JSON, XML over HTTP | `requests` + `pd.read_json()` |
| Web scraping (static) | HTML | `requests` + `BeautifulSoup` |
| Web scraping (dynamic / JS) | rendered HTML | `selenium` |
| Streaming | real-time events | Kafka, custom consumers |

### Key considerations before gathering
- **Where** is the data? (file, DB, API, website)
- **Format** is it in? (CSV, JSON, SQL...)
- **Size** — does it fit in memory, or must it be read in chunks?
- **Cleanliness** — what problems will you have to handle later?
- **Access** — authentication, rate limits, encoding?





##  Step 2 — Data Cleaning

**Definition:** Finding and fixing data-quality problems so your analysis is trustworthy.

### Common problems you will fix
| Problem | Example | Tool |
|---------|---------|------|
| Missing values | `NaN`, blank cells, "N/A" | `dropna()`, `fillna()` |
| Duplicates | same customer twice | `drop_duplicates()` |
| Outliers | age = 300, salary = 1 crore | quantiles, IQR, domain rules |
| Wrong data types | numbers stored as text | `to_numeric()`, `astype()` |
| Inconsistent formatting | "new york", "New York", "NY" | string methods `.str.title()` |
| Impossible values | negative price, future birth date | validation rules |

### The cleaning mantra
> **Inspect first → decide a rule → apply it → verify.**  
> Always *look* at the data with `.info()`, `.isna().sum()`, `.duplicated().sum()` and
> `.describe()` **before** you change anything.





##  Step 3 — Data Transformation

**Definition:** Converting the clean data into a shape that is ready for analysis and modelling.

### Typical transformations
1. **Feature engineering** — create new, useful columns
   ```python
   df["Revenue"] = df["Price"] * df["Quantity"]
   df["AOV"]    = df["Revenue"] / df["Orders"]
   ```
2. **Encoding categoricals** — text → numbers
   - Label encoding (`pd.factorize`), One-hot encoding (`pd.get_dummies`)
3. **Scaling / normalising** — bring features to a similar range
   - Min-Max, Z-score (Standardisation)
4. **Aggregation** — summarize groups
   ```python
   df.groupby("Category")["Revenue"].sum()
   ```
5. **Reshaping** — pivot / melt / stack
   ```python
   df.pivot_table(index="month", columns="region", values="revenue")
   ```
6. **Joining / merging** datasets together (`merge`, `join`, `concat`)

**Deliverable:** a tidy, analysis-ready table.





##  Step 4 — Data Exploration (EDA)

**Exploratory Data Analysis** is the process of **understanding** the data through
summary statistics and visualisations **before** formal modelling.

### What EDA answers
- What does the data *look* like? (shape, dtypes, head)
- What are the main statistics? (`.describe()` — mean, median, std, min, max)
- How are values distributed? (histograms, boxplots)
- What correlates with what? (correlation matrix, heatmap)
- Which groups behave differently? (groupby comparisons)
- Are there patterns or anomalies?

### Typical EDA commands
```python
df.shape            # (rows, columns)
df.info()           # dtypes + non-null counts
df.describe()       # numeric stats
df.isna().sum()     # missing values
df.groupby("...").mean()
df.corr()           # correlation
```

> EDA is where you *discover* the story; modelling is where you *formalise* it.






## Step 5 — Data Modeling

**Definition:** Using statistics or **Machine Learning** to find relationships and make predictions.

### Two broad flavours
- **Supervised** — learn from labelled examples
  - *Classification* (spam / not spam, churn / not churn) → `LogisticRegression`, `RandomForest`
  - *Regression* (predict price, sales) → `LinearRegression`, `XGBoost`
- **Unsupervised** — find structure without labels
  - *Clustering* (customer segments) → `KMeans`
  - *Dimensionality reduction* → `PCA`

### A typical modelling loop
1. Split data → train / validation / test
2. Train a model on the training set
3. Evaluate on validation (tune hyperparameters)
4. Evaluate once on the **test** set (unseen data)
5. Choose the best model

> This notebook focuses on **gathering** the data. Modelling is covered in later,
> ML-focused notebooks — but the whole pipeline leads here.




##  Step 6 — Data Visualisation & Communication

**Definition:** Turning the results into charts, tables and narratives that
**other people understand and act upon**.

### Visualisation libraries
| Library | Best for |
|---------|----------|
| `matplotlib` | low-level control, static plots |
| `seaborn` | statistical plots, beautiful defaults |
| `plotly` | interactive charts |
| `pandas` `.plot()` | quick charts straight from DataFrames |

### Communication tools
- Reports & presentations
- Interactive dashboards (e.g. **Streamlit**, Power BI, Tableau)
- Charts embedded in notebooks

### Good communication rules
- Choose the right chart for the message (line = trends, bar = compare, pie = share...)
- Label axes, add titles, keep it simple
- Lead with the **insight**, not the method
- Always end with a **recommendation**

### Common BeautifulSoup methods
| Method | What it does |
|--------|--------------|
| `soup.find("tag")` | find the **first** matching element |
| `soup.find_all("tag")` | find **all** matching elements |
| `soup.select("css")` | find with a CSS selector (`"div.item h2"`) |
| `.text` / `.get_text()` | extract the visible text |
| `["attr"]` / `.get("attr")` | read an attribute (e.g. `href`) |
| `.find(class_="price")` | find by class |



````python
# Extracting text and attributes
html = '''<a class="product" href="/p/1">Laptop</a>
<a class="product" href="/p/2">Mouse</a>'''
soup = BeautifulSoup(html, "html.parser")

names = [a.text for a in soup.find_all("a", class_="product")]
links = [a["href"] for a in soup.find_all("a", class_="product")]
print("names:", names)
print("links:", links)
````

### Building a complete (small) scraper
The standard loop:
1. **Fetch** the HTML
2. **Parse** with BeautifulSoup
3. **Find** the repeated container elements
4. **Extract** fields into lists
5. **Assemble** lists into a DataFrame
6. **Save** to CSV/JSON

**Always wrap in try/except** — real pages change and break scrapers.



````python
import pandas as pd
from bs4 import BeautifulSoup

# Simulated page with several products
html = '''<div class="card">
  <h2 class="name">Phone X</h2><p class="price">₹15000</p>
</div>
<div class="card">
  <h2 class="name">Phone Y</h2><p class="price">₹20000</p>
</div>'''
soup = BeautifulSoup(html, "html.parser")

names  = [n.text for n in soup.find_all("h2", class_="name")]
prices = [p.text for p in soup.find_all("p", class_="price")]

df_scrape = pd.DataFrame({"name": names, "price": prices})
df_scrape
````

### Handling errors with try / except
```python
try:
    r = requests.get(url, timeout=10)
    r.raise_for_status()
    soup = BeautifulSoup(r.text, "html.parser")
    ...extract...
except requests.exceptions.RequestException as e:
    print("Skipping page:", e)
```
If one page fails, you **skip it and continue** instead of crashing the whole job.



````python
import requests
from bs4 import BeautifulSoup

def scrape_page(url):
    try:
        r = requests.get(url, timeout=10)
        r.raise_for_status()
        return BeautifulSoup(r.text, "html.parser")
    except requests.exceptions.RequestException as e:
        print("Failed:", url, "->", e)
        return None

# soup = scrape_page("https://example.com")    # replace with a real URL
print("Robust scraper function defined.")
````

## B24 · Web Scraping Dynamic Sites with Selenium

Some sites render content with **JavaScript** — the HTML you download with `requests`
is empty. `Selenium` controls a **real browser** (Chrome/Firefox), executes the JS,
and lets you read the *rendered* page.

```bash
pip install selenium
# also need a matching chromedriver (or use Selenium Manager)
```

### Basic workflow
1. Create a `WebDriver` (Chrome options: headless mode, no sandbox)
2. `driver.get(url)` — load the page (JS runs)
3. Wait for elements (explicit waits, **not** `time.sleep`)
4. Find elements (`By.ID`, `.CLASS_NAME`, `.CSS_SELECTOR`...)
5. Read / click / scroll
6. `driver.quit()` to close it



### Selenium locators (`By`)
| Locator | Example |
|---------|---------|
| `By.ID` | `driver.find_element(By.ID, "email")` |
| `By.NAME` | `driver.find_element(By.NAME, "q")` |
| `By.CLASS_NAME` | `driver.find_element(By.CLASS_NAME, "price")` |
| `By.XPATH` | `driver.find_element(By.XPATH, "//div[@class='card']")` |
| `By.CSS_SELECTOR` | `driver.find_element(By.CSS_SELECTOR, "div.card h2")` |
| `By.TAG_NAME` | `driver.find_element(By.TAG_NAME, "h2")` |



### Waiting strategies — use explicit waits
```python
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

wait = WebDriverWait(driver, 10)
el = wait.until(EC.presence_of_element_located((By.CLASS_NAME, "price")))
```
Common expected conditions: `presence_of_element_located`, `visibility_of_element_located`,
`element_to_be_clickable`, `text_to_be_present_in_element`.

> ⚠️ Avoid `time.sleep(...)` — it slows scraping and is fragile.

### Selenium vs BeautifulSoup — when to use which
| Situation | Use |
|-----------|-----|
| Static HTML, simple selectors | `BeautifulSoup` (fast, light) |
| Data behind JS / login / interaction | `Selenium` (real browser) |
| Huge volume, polite & fast | `BeautifulSoup` + `requests` |
| Complex clicks, scrolling, sessions | `Selenium` |

**Best practice:** always try `requests` + `BeautifulSoup` first; only bring
`Selenium` in when the page truly requires a rendered browser.

## B26 · Summary — A Complete Data-Gathering Cheatsheet

| Source | Method | Notes |
|--------|--------|-------|
| Local CSV | `pd.read_csv()` | many params (`sep`, `dtype`, `parse_dates`...) |
| CSV from URL | `pd.read_csv(url)` or `requests` | handle headers/errors |
| JSON | `pd.read_json()`, `json_normalize` | flatten nested data |
| Excel | `pd.read_excel()` | `sheet_name` for multi-sheet |
| Text/TSV | `pd.read_csv(sep="\t")` | set the delimiter |
| SQL | `pd.read_sql_query()` / `read_sql_table()` | `mysql.connector` / `sqlalchemy` |
| API | `requests` + `pd.DataFrame()` | pagination + rate limits |
| Scrape static | `requests` + `BeautifulSoup` | parse HTML |
| Scrape dynamic | `selenium` | JS-rendered pages |
| Export | `to_csv/excel/json/sql/html` | `index=False` recommended |

### The universal "load then inspect" habit
```python
df = <gather the data>
df.shape        # rows, columns
df.info()       # dtypes + nulls
df.head()       # a peek
df.isna().sum() # missing values
```
Get this habit down and the rest of the Data Analysis Process flows naturally.
