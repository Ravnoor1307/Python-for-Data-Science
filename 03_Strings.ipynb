### Program 0.0.1 - Population Growth

The current population of a town is 10000. The population of the town is increasing at the rate of 10% per year. Find out the population at the end of each of the last 10 years.

````python
curr_pop = 10000
for i in range(10,0,-1):
    print(i, curr_pop)
    curr_pop = curr_pop / 1.1
````

**Explanation:**
- We start with current population = 10000
- Using `range(10, 0, -1)` we count backwards from 10 to 1
- Each iteration we divide by 1.1 (to go back one year with 10% growth)
- `curr_pop / 1.1` reverses the 10% increase

### Program 0.0.2 - Sequence Sum

Calculate the sum of the series: 1/1! + 2/2! + 3/3! + ... + n/n!

````python
n = int(input('enter n'))
result = 0
fact = 1
for i in range(1, n+1):
    fact = fact * i
    result = result + i/fact
print(result)
````

**Explanation:**
- We use a running factorial: instead of recalculating `i!` each time, we multiply `fact` by `i` in each iteration
- For n=2: result = 1/1! + 2/2! = 1 + 1 = 2.0
- Each term `i/fact` is added to the running sum


## Nested Loops

A nested loop is a loop inside another loop. The inner loop completes all its iterations for each iteration of the outer loop.

### Example - Unique Pairs

````python
for i in range(1, 5):
    for j in range(1, 5):
        print(i, j)
````

**Explanation:**
- Outer loop runs 4 times (i = 1, 2, 3, 4)
- For each `i`, inner loop runs 4 times (j = 1, 2, 3, 4)
- Total iterations = 4 x 4 = 16
- Output shows all unique pairs: (1,1), (1,2), ... (4,4)

### Pattern 1 - Right Triangle of Stars

```
*
**
***
****
*****
******
*******
********
*********
**********
```

````python
rows = int(input('enter number of rows'))
for i in range(1, rows+1):
    for j in range(1, i+1):
        print('*', end='')
    print()
````

**Explanation:**
- Outer loop controls the row number (1 to rows)
- Inner loop prints `i` stars for row `i`
- `end=''` prevents newline after each star
- `print()` moves to the next line after each row

### Pattern 2 - Number Pyramid

```
1
121
12321
1234321
```

````python
rows = int(input('enter number of rows'))
for i in range(1, rows+1):
    for j in range(1, i+1):
        print(j, end='')
    for k in range(i-1, 0, -1):
        print(k, end='')
    print()
````

**Explanation:**
- First inner loop prints ascending numbers: 1 to i
- Second inner loop prints descending numbers: i-1 down to 1
- Combined they create the pyramid effect

## Loop Control Statements

Python provides three statements to control loop execution:

| Statement | Description |
|-----------|-------------|
| `break` | Terminates the loop immediately |
| `continue` | Skips the current iteration, moves to next |
| `pass` | Does nothing, acts as a placeholder |

### Break Statement

````python
for i in range(1, 10):
    if i == 5:
        break
    print(i)
````

**Output:** 1, 2, 3, 4

When `i == 5`, `break` exits the loop immediately.

### Prime Number Check Using Break with For-Else

````python
lower = int(input('enter lower range'))
upper = int(input('enter upper range'))

for i in range(lower, upper+1):
    for j in range(2, i):
        if i%j == 0:
            break
    else:
        print(i)
````

**Explanation:**
- For each number `i`, check if any `j` from 2 to i-1 divides it
- If yes, `break` exits inner loop (not prime)
- The `else` block only runs if the loop completed WITHOUT `break` (prime number)
- This is Python's unique for-else construct!

### Continue Statement

````python
for i in range(1, 10):
    if i == 5:
        continue
    print(i)
````

**Output:** 1, 2, 3, 4, 6, 7, 8, 9

`continue` skips the current iteration (i=5) and moves to the next.

### Pass Statement

````python
for i in range(1, 10):
    pass
````

`pass` does nothing. It's used as a placeholder when syntactically a statement is required but you want no code to execute.

###  Editing and Deleting in Strings

> **Important:** Python strings are **immutable** - they cannot be changed after creation.

````python
s = 'hello world'
s[0] = 'H'
````

**TypeError:** `str` object does not support item assignment. Strings are immutable!

````python
s = 'hello world'
del s
print(s)
````

**NameError:** `s` is not defined after deletion.

````python
s = 'hello world'
del s[-1:-5:2]
print(s)
````

**TypeError:** `str` object does not support item deletion. You can only delete the entire string variable.

### Problem 18: Find Uncommon Words

**Input:**
- A = `"apple banana mango"`
- B = `"banana fruits mango"`

**Output:** `['apple', 'fruits']`

````python
# Code here
A = "apple banana mango" 
B = "banana fruits mango"

L = []

for i in A.split():
  if i not in B and i not in L:
    L.append(i)

for j in B.split():
  if j not in A and j not in L:
    L.append(j)

print(L)
````
