## 1. What are Arrays in Python?

An **array** is a collection of items stored at contiguous memory locations. In Python, arrays are used to store multiple values in a single variable.

### Key Characteristics:

- **Homogeneous**: All elements are of the same type
- **Contiguous Memory**: Elements are stored in adjacent memory locations
- **Index-Based**: Access elements using their index (starting from 0)
- **Mutable**: Can be modified after creation

### Memory Layout:

```
Index:    0     1     2     3     4
Value:  [10]  [20]  [30]  [40]  [50]
Memory: 100   104   108   112   116  (contiguous)
```

### Why Use Arrays?

1. **Efficient Storage**: Store multiple values efficiently
2. **Quick Access**: O(1) time to access any element
3. **Iteration**: Easy to loop through all elements
4. **Memory Efficient**: Better than individual variables


## 2. Array vs List

In Python, **lists** are the built-in array implementation. Here's how they compare:

| Feature | Python List | Array Module |
|---------|-------------|--------------|
| Type | Dynamic | Fixed-type |
| Memory | More (object overhead) | Less (compact) |
| Flexibility | Any data type | Same data type |
| Performance | Slightly slower | Slightly faster |

### For Most Use Cases:

- Use **Python lists** - they're more Pythonic and flexible
- Use **array module** for memory-critical applications
- Use **NumPy arrays** for numerical computations

### Example:

```python
# Python List (recommended)
arr_list = [1, 2, 3, 4, 5]

# Array Module
from array import array
arr_module = array('i', [1, 2, 3, 4, 5])
```


## 8. Array Searching

### 8.1 Linear Search

````python
# Linear Search - O(n)
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

arr = [10, 23, 45, 70, 11, 15]
target = 70

result = linear_search(arr, target)
if result != -1:
    print(f"Element found at index {result}")
else:
    print("Element not found")
````

### 8.2 Binary Search (Sorted Array Only)

````python
# Binary Search - O(log n)
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1
    
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1
````

````python
arr = [10, 23, 45, 70, 11, 15]
target = 70
````

````python
# Must sort first for binary search
arr.sort()
print("Sorted array:", arr)

result = binary_search(arr, target)
if result != -1:
    print(f"Element found at index {result}")
else:
    print("Element not found")
````


## 12. Time Complexity

| Operation | Time Complexity | Description |
|-----------|-----------------|-------------|
| Access | O(1) | Direct index access |
| Search | O(n) | Linear search |
| Insert at end | O(1) | Append |
| Insert at beginning | O(n) | Shift all elements |
| Insert at index | O(n) | Shift elements after index |
| Delete at end | O(1) | Pop last element |
| Delete at beginning | O(n) | Shift all elements |
| Delete at index | O(n) | Shift elements after index |
| Search in sorted | O(log n) | Binary search |
| Sort | O(n log n) | Python's Timsort |

### Tips for Optimized Code:

- **Use `append()` instead of `insert(0, x)`** for adding at the end
- **Use `pop()` instead of `pop(0)`** for removing from the end
- **Use list comprehension** for creating new lists
- **Use `in` operator** for membership testing
- **Use `sorted()`** when you need a sorted copy without modifying original

### Q9. Find element with left side smaller/right side greater in an array

````python
L = [3,1,2,5,8,7,9]

max_arr = []
min_arr = []

max_val = L[0]
min_val = L[-1]

for i in L:
  if i>max_val:
    max_val = i
  max_arr.append(max_val)

for i in range(len(L)-1,-1,-1):
  if L[i] < min_val:
    min_val = L[i]
  min_arr.insert(0,min_val)

for i in range(1,len(L)-1):
  if max_arr[i-1] < L[i] < min_arr[i+1]:
    print(L[i])
````

### Q10. Maximum sum subarray (Kadane's Algorithm)

````python
L = [-2,4,7,-1,6,-11,14,3,-1,-6]

curr_sum = 0
curr_seq = []
best_sum = L[0]
best_seq = []

for i in L:
  if i + curr_sum > i:
    curr_sum = curr_sum + i
    curr_seq.append(i)
  else:
    curr_sum = i
    curr_seq.clear()
    curr_seq.append(i)

  if curr_sum > best_sum:
    best_sum = curr_sum
    best_seq = curr_seq

print(best_sum,best_seq)
````

### Q11. Sort arrays with items 1 and 0

````python
L = [0,1,0,1,0,1,1,0]

i = 0
j = len(L) - 1

while i < j:
  if L[i] == 0:
    i += 1
  elif L[j] == 1:
    j -= 1
  else:
    L[i], L[j] = L[j], L[i]
    i += 1
    j -= 1

print(L)
````

### Q13. Maximum Product Subarray

````python
L = [2,3,-2,4]

max_prod = L[0]
min_prod = L[0]
result = L[0]

for i in range(1, len(L)):
  temp = max_prod
  max_prod = max(L[i], L[i]*max_prod, L[i]*min_prod)
  min_prod = min(L[i], L[i]*temp, L[i]*min_prod)
  result = max(result, max_prod)

print(result)
````

### Q14. Find union of 2 arrays

````python
a = [1,2,3,4,5]
b = [3,4,5,6,7]

# Using set
union = list(set(a) | set(b))
print(union)

# Without using set
union = []
for i in a:
  if i not in union:
    union.append(i)
for i in b:
  if i not in union:
    union.append(i)
print(union)
````

### Q15. Find Pythagorean triplets in an array

````python
L = [3,4,5,6,8,10]

for i in range(len(L)):
  for j in range(i+1, len(L)):
    for k in range(j+1, len(L)):
      a, b, c = sorted([L[i], L[j], L[k]])
      if a*a + b*b == c*c:
        print(f"({a}, {b}, {c})")
````
