### Namespaces

A namespace is a space that holds names(identifiers).Programmatically speaking, namespaces are dictionary of identifiers(keys) and their objects(values)

There are 4 types of namespaces:
- Builtin Namespace
- Global Namespace
- Enclosing Namespace
- Local Namespace

### Scope and LEGB Rule

A scope is a textual region of a Python program where a namespace is directly accessible.

The interpreter searches for a name from the inside out, looking in the local, enclosing, global, and finally the built-in scope. If the interpreter doesn’t find the name in any of these locations, then Python raises a NameError exception.

## 📖 Local vs Global scope

```python
a = 2              # global

def temp():
    b = 3          # local — sirf is function ke andar
    print(b)

temp()
print(a)           # global accessible
# print(b)         # NameError! `b` local hi tha
```

- **Global** — file top-level, sab jagah dikhta hai
- **Local** — function ke andar bana, wahi me access
- `temp()` ke andar bana `b` bahar nahi milega
- Namespace = names ka dictionary, scope = region

### Decorators

A decorator in python is a function that receives another function as input and adds some functionality(decoration) to and it and returns it.

This can happen only because python functions are 1st class citizens.

There are 2 types of decorators available in python
- `Built in decorators` like `@staticmethod`, `@classmethod`, `@abstractmethod` and `@property` etc
- `User defined decorators` that we programmers can create according to our needs

## 📖 Functions are first-class citizens


- Python me functions = bhi values hain — pass, return, store sab

## 🔬 Deep Dive : First decorator — closure + wrapper

```python
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print('before call')
        result = func(*args, **kwargs)
        print('after call')
        return result
    return wrapper          # returns the wrapped function

@my_decorator
def hello():
    print('hi')
```

- `@decorator` — syntactic sugar = `func = decorator(func)`
- Decorator receives `func`, defines `wrapper`, returns wrapper
- wrapper extends behavior (before/after)
- Closure: wrapper holds func reference
- Calling hello() → actual wrapper runs → func()
- Return value passthrough: `result = func(...)` + `return result`
- **args/**kwargs forwarding — preserve signature
- Real usage: logging, timing, auth, caching
- Decorators = HOF applied at definition
- Behavior injection without touching original

````python
# simple example

def my_decorator(func):
  def wrapper():
    print('***********************')
    func()
    print('***********************')
  return wrapper

def hello():
  print('hello')

def display():
  print('hello nitish')

a = my_decorator(hello)
a()

b = my_decorator(display)
b()
````


- `@decorator` ke baad `hello.__name__` = 'wrapper' (original name khota)
- Debugging tools (`help`, name) galat show karte hain
- Fix: `functools.wraps(func)` — metadata copy
- Isliye production decorators me wraps zaroor lagaaiye

## 🔬 Deep Dive : Problem with plain wrapper

```python
def my_decorator(func):
    def wrapper(*args, **kwargs):
        ...
    return wrapper

def greet():
    '''docstring'''
    ...

```

- Decorated function loses metadata: `__name__`, `__doc__`, `__module__`
- introspection tools see wrapper, not original
- Fix: `functools.wraps`:

```python

def my_decorator(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        ...
    return wrapper
```
- `@wraps(func)` copies __name__, __doc__, signature-ish
- Best practice: ALWAYS use @wraps
- debug & trace clean
- othera — functools.wraps essential professional decorators
- Without — API/documentation broken

## 📖 sanity_check ka structure

```python
def sanity_check(data_type):          # layer 1: args
    def outer_wrapper(func):          # layer 2: function
        def inner_wrapper(*args):     # layer 3: real logic
            if type(*args) == data_type:
                print('number is integer')
            else:
                print('number is not integer')
        return inner_wrapper
    return outer_wrapper

@checkdt(int)   # == square = sanity_check(int)(square)
```

- 3 layers = args + func + args call
- `@name(args)` — me pehla call arguments ke saath
- Then decorator function milta hai — usi pe original function
- Layer counting yaad rakhna: args → func → wrapper logic

````python
def sanity_check(data_type):
  def outer_wrapper(func):
    def inner_wrapper(*args):
      if type(*args) == data_type:
        func(*args)
      else:
        raise TypeError('Ye datatype nai chalega')
    return inner_wrapper
  return outer_wrapper

@sanity_check(int)
def square(num):
  print(num**2)

@sanity_check(str)
def greet(name):
  print('hello',name)

square(2)
````
