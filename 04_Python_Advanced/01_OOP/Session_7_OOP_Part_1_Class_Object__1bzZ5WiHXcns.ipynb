## 📖 Function vs Method

- **Function** = khali ese code jo baahar hota hai — `len(L)`

## 🔬 Deep Dive : Class + `__init__` constructor

```python
class Atm:
    def __init__(self, pin, balance=1000):
        self.pin = pin
        self.balance = balance
```

- `class` keyword → blueprint define karta
- `__init__` = **constructor** — object bante hi khud call hota
- `self` = har method ka pehla param — CURRENT OBJECT ka reference
- `self.pin` — object ka attribute (state)
- Constructor parameterized — object customize kar sakte
- Bina constructor: default object with no data
- `Atm(1234)` → __init__ runs, sets pin
- Java/C++ jaisa hi concept, syntax different

````python
class Atm:

  # constructor(special function)->superpower -> 
  def __init__(self):
    print(id(self))
    self.pin = ''
    self.balance = 0
    #self.menu()

  def menu(self):
    user_input = input("""
    Hi how can I help you?
    1. Press 1 to create pin
    2. Press 2 to change pin
    3. Press 3 to check balance
    4. Press 4 to withdraw
    5. Anything else to exit
    """)

    if user_input == '1':
      self.create_pin()
    elif user_input == '2':
      self.change_pin()
    elif user_input == '3':
      self.check_balance()
    elif user_input == '4':
      self.withdraw()
    else:
      exit()

  def create_pin(self):
    user_pin = input('enter your pin')
    self.pin = user_pin

    user_balance = int(input('enter balance'))
    self.balance = user_balance

    print('pin created successfully')
    self.menu()

  def change_pin():
    old_pin = input('enter old pin')

    if old_pin == self.pin:
      # let him change the pin
      new_pin = input('enter new pin')
      self.pin = new_pin
      print('pin change successful')
      self.menu()
    else:
      print('nai karne de sakta re baba')
      self.menu()

  def check_balance(self):
    user_pin = input('enter your pin')
    if user_pin == self.pin:
      print('your balance is ',self.balance)
    else:
      print('chal nikal yahan se')

  def withdraw(self):
    user_pin = input('enter the pin')
    if user_pin == self.pin:
      # allow to withdraw
      amount = int(input('enter the amount'))
      if amount <= self.balance:
        self.balance = self.balance - amount
        print('withdrawl successful.balance is',self.balance)
      else:
        print('abe garib')
    else:
      print('sale chor')
    self.menu()


````

```python

```

- `len(L)` — function hai, class ke baahar; list ko andar bhejo

## 🔬 Deep Dive : len vs append — function vs method

```python
L = [1,2,3]
len(L)        # function — sequence pe kaam (outside class)
L.append(42)  # method — belongs to list class (inside)
```

- **Function** (`len`, `min`, `max`, `sorted`) — object ke BAHAR defined jo uspe kaam karta
- **Method** (`append`, `upper`, `sort`) — class ke ANDAR defined, object se attached
- len works on strings/lists/tuples — generic utility
- append only lists — type specific behavior
- Call style: function `func(obj)` vs method `obj.method()`
- Internally len → obj.__len__() magic method
- Design: shared generic operations = functions; type unique = methods

## 📖 Magic methods & operator overloading

```python
class Fraction:
    def __init__(self, x, y):     # parameterized constructor
        self.num = x
        self.den = y

    def __str__(self):            # print hone par kya dikhe
        return '{}/{}'.format(self.num, self.den)

    def __add__(self, other):     # + ko apne hisab se define
        new_num = self.num * other.den + other.num * self.den
        new_den = self.den * other.den
        return Fraction(new_num, new_den)
```

- `__init__(self, x, y)` — arguments ke saath constructor
- `__str__` — `print(fr1)` me readable output dene ke liye
- `__add__` / `__sub__` / `__mul__` / `__truediv__` — operator overloading
- `fr1 + fr2` internally `fr1.__add__(fr2)` call karta hai

## 🔬 Deep Dive : Magic methods — operator overloading

```python
class Fraction:
    def __init__(self, x, y):   # constructor
        self.n, self.d = x, y
    def convert_to_decimal(self):
        return self.n / self.d
    def __add__(self, other):    # MAGIC method → defines +
        return Fraction(self.n*other.d + other.n*self.d, self.d*other.d)
    def __sub__(self, o):
        ...
    def __mul__(self, o):
        ...
    def __truediv__(self, o):
        ...

fr1 = Fraction(3,4)
fr2 = Fraction(1,2)
fr1 + fr2      # uses __add__
fr1 - fr2      # __sub__
```

- **Magic methods** (dunder `__...__`) — python internally special ops ke liye call karta
- `__add__` → `+` | `__sub__` → `-` | `__mul__` → `*` | `__truediv__` → `/`
- Operator overloading = apne class pe operators define
- `len(L)` internally `L.__len__()` | str → `__str__`
- `print(fr1)` → uses `__str__`/`__repr__`
- Kab chahiye: custom math types (Fraction, Vector, Matrix)
- Formula for fraction add: cross-multiply + lcm shortcut
- Magic methods = Pythonka "speech interface" for operators
- Favorite interview topic: name few dunders

````python
class Fraction:

  # parameterized constructor
  def __init__(self,x,y):
    self.num = x
    self.den = y

  def __str__(self):
    return '{}/{}'.format(self.num,self.den)

  def __add__(self,other):
    new_num = self.num*other.den + other.num*self.den
    new_den = self.den*other.den

    return '{}/{}'.format(new_num,new_den)

  def __sub__(self,other):
    new_num = self.num*other.den - other.num*self.den
    new_den = self.den*other.den

    return '{}/{}'.format(new_num,new_den)

  def __mul__(self,other):
    new_num = self.num*other.num
    new_den = self.den*other.den

    return '{}/{}'.format(new_num,new_den)

  def __truediv__(self,other):
    new_num = self.num*other.den
    new_den = self.den*other.num

    return '{}/{}'.format(new_num,new_den)

  def convert_to_decimal(self):
    return self.num/self.den

    

  
````

````python
fr1 = Fraction(3,4)
fr2 = Fraction(1,2)
````

````python
fr1.convert_to_decimal()
# 3/4
````

````python
print(fr1 + fr2)
print(fr1 - fr2)
print(fr1 * fr2)
print(fr1 / fr2)
````

## 🔬 Deep Dive : Every type has its own operator set

```python
s1 = {1,2,3}
s2 = {3,4,5}
s1 + s2     # ❌ TypeError — sets support | not +
```

- Har TYPE apne operators define karta hai (via magic methods)
- set: `|` union, `&` intersection work, but `+` NOT defined
- list/str: `+` concatenation
- int/float: `+` addition
- Aap apne class me jo operators chahiye wo define karo
- So "operator" meaning = type-dependent
- Symbol same, behavior class decides (polymorphism)
- Error hamesha type mismatch / unsupported op
- Knowing supported ops per type = type fluency
