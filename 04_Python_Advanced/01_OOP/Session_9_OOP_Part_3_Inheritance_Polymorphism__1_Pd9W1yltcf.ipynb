## 🔬 Deep Dive : Method overriding

```python
class A:
    def __init__(self):
        self.var1 = 100
    def m1(self):
        print("Method in A")
class B(A):
    def m1(self):           # OVERRIDE — same name, own version
        print("Method in B")

b = B()
b.m1()       # prints "Method in B" — child's version wins
```

- Child apne same-named method se parent ko REPLACE karta
- Overriding = child kisi method ko re-define karta
- Call resolution: child class ka method pehle
- To call parent version inside: `super().m1()`
- Signature match recommended (overload same nama different args)
- Real: `Phone` parent price calc, camera-phone child override
- Foundation of polymorphism
- Overloading hi Python me default nahi (last def wins)

## 🔬 Deep Dive : Inheritance — Is-A relationship

## 🔬 Deep Dive : What gets inherited?

| Cheez | Inherit hota? |
|---|---|
| Constructor (`__init__`) | ❌ nahi — khud call karna `super().__init__()` |
| Non-private attributes/methods | ✅ |
| Private members (`__x`) | ❌ nahi |
| Class/static vars | ✅ |

```python
class Phone:
    def __init__(self, price, brand):     # parent constructor
        ...
class SmartPhone(Phone):
    def __init__(self, price, brand):
        super().__init__(price, brand)    # parent ka constructor call
```

- Child me apna __init__ def karo → parent wala **NOT auto-called**
- Explicit `super().__init__(...)` parent constructor chalao
- Attributes/methods (public) inherit automatically
- Private (`__num`) — child kos bhi access nahi — mangled to parent only
- Static/class variables shared
- Agar child me __init__ nahi → parent ka use by default (MRO)
- Python uses MRO (Method Resolution Order)
- Access rules: public ✓ protected(_) ✓ private ✗

## 📖 Method overriding

```python
class A:
    def __init__(self):
        self.var1 = 100

    def display1(self, var1):
        print(var1)

class B(A):
    def display1(self, var1):    # same name naya — PARENT wala override
        print('This is overriding the parent method!', var1)
```

- Child me wahi method name parent jaisa → **overriding**
- Runtime pe child ka hi version chalta hai (polymorphism ka hissa)
- Agar chahiye parent ka bhi, `super().method()` bula lo

````python
class A:
    def __init__(self):
        self.var1=100

    def display1(self,var1):
        print("class A :", self.var1)
class B(A):
  
    def display2(self,var1):
        print("class B :", self.var1)

obj=B()
obj.display1(200)
````

##### Inheritance in summary

- A class can inherit from another class.

- Inheritance improves code reuse

- Constructor, attributes, methods get inherited to the child class

- The parent has no access to the child class

- Private properties of parent are not accessible directly in child class

- Child class can override the attributes or methods. This is called method overriding

- super() is an inbuilt function which is used to invoke the parent class methods and constructor

````python
class Parent:

    def __init__(self,num):
      self.__num=num

    def get_num(self):
      return self.__num

class Child(Parent):
  
    def __init__(self,num,val):
      super().__init__(num)
      self.__val=val

    def get_val(self):
      return self.__val
      
son=Child(100,200)
print(son.get_num())
print(son.get_val())
````

## 🔬 Deep Dive : Diamond problem

```python
class A:
    def m1(self): return 20
class B(A): ...
class C(A): ...
class D(B, C): ...        # which m1? from A via B or C?
```

- Multiple inheritance me same ancestor ka method 2 path
- Conflict: D.m1() kaunsa?
- Python MRO (C3 linearization) — deterministic order solves
- `D.mro()` → order list check karo
- `class D(B, C)` — B pehle listing → B chain wins
- Ambiguity risk — same base twice in graph
- Fix: explicit MRO order, super() propagation
- Prefer composition over multiple inheritance (design advice)
- Interview classic — Python handles quietly, other langs nahi

````python
# the diamond problem
# https://stackoverflow.com/questions/56361048/what-is-the-diamond-problem-in-python-and-why-its-not-appear-in-python2
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class Product:
    def buy(self):
        print ("Product buy method")

# Method resolution order
class SmartPhone(Phone,Product):
    pass

s=SmartPhone(20000, "Apple", 12)

s.buy()
````

````python
class A:

    def m1(self):
        return 20

class B(A):

    def m1(self):
        return 30

    def m2(self):
        return 40

class C(B):
  
    def m2(self):
        return 20
obj1=A()
obj2=B()
obj3=C()
print(obj1.m1() + obj3.m1()+ obj3.m2())
````

## 📖 Diamond problem

```python
class A:
    def m1(self):
        return 20

class B(A):
    def m1(self):
        val = super().m1() + 30
        return val

class C(B):
    def m1(self):
        val = self.m1() + 20   # warning: recursion risk!
        return val
```

- Multiple inheritance me jab same method kai parents me ho — ambiguity
- Python ka MRO (Method Resolution Order) decide karta hai kaun chalega
- `super()` MRO ke hisab se parent wala method dhundta hai
- Recursion sai bacho — `self.m1()` andar se wahi method call karega

````python
class A:

    def m1(self):
        return 20

class B(A):

    def m1(self):
        val=super().m1()+30
        return val

class C(B):
  
    def m1(self):
        val=self.m1()+20
        return val
obj=C()
print(obj.m1())
````

### Abstraction

## 🔬 Deep Dive : Abstraction (ABC)

```python
from abc import ABC, abstractmethod

class BankApp(ABC):
    def database(self):
        print('connected')
    @abstractmethod
    def security(self):      # child ko FORCE karta define karna
        pass

class MobileApp(BankApp):
    def mobile_login(self):
        print('login')
    def security(self):      # required!
        print('secure login added')

mob = MobileApp()            # abstract child bhi abstract → error
obj = BankApp()              # ❌ cannot instantiate abstract class
```

- **Abstract class** — interface/contract define karta
- `@abstractmethod` — child ko IMPLEMENT karna zaroori
- Abstract class directly instantiate nahi hoti
- Child bhi abstract rehta jab tak sab abstract methods define na ho
- Baaki (non-abstract) methods inherit + share
- Design: policy defined, implementation child pe
- Guarantee konsa method exist karega — consumers safe
- vs Interface (Python no keyword) — ABC is the way
- Real: frameworks, base services

````python
from abc import ABC,abstractmethod
class BankApp(ABC):

  def database(self):
    print('connected to database')

  @abstractmethod
  def security(self):
    pass

  @abstractmethod
  def display(self):
    pass
````

## 📖 Abstraction (ABC)

```python
from abc import ABC, abstractmethod

class BankApp(ABC):              # abstract class
    @abstractmethod
    def security(self): ...      # abstract method: child pe maange

class MobileApp(BankApp):
    def mobile_login(self): ...
    def security(self): ...      # ye to implement karna hi hoga
```

- `ABC` + `@abstractmethod` → children ko method **compel** karta hai
- Abstract class ka object directly nahi ban sakta
- Benefits: template design — parent guarantee, child details

````python
class MobileApp(BankApp):

  def mobile_login(self):
    print('login into mobile')

  def security(self):
    print('mobile security')

  def display(self):
    print('display')
````
