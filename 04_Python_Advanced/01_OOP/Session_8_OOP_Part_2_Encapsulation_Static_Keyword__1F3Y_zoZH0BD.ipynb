## 📖 Real scenario: Point & Line classes

```python
class Point:
    def __init__(self, x, y):
        self.x_cod = x
        self.y_cod = y
```

- OOP ka real use — real-world cheezein (point, line, distance) ko classes
  se model karte hain
- `__str__` define karoge to `print(p1)` readable string deta hai
- Methods like `euclidean_distance(other)` — object ki helper cheezein

````python
class Point:

  def __init__(self,x,y):
    self.x_cod = x
    self.y_cod = y

  def __str__(self):
    return '<{},{}>'.format(self.x_cod,self.y_cod)

  def euclidean_distance(self,other):
    return ((self.x_cod - other.x_cod)**2 + (self.y_cod - other.y_cod)**2)**0.5

  def distance_from_origin(self):
    return (self.x_cod**2 + self.y_cod**2)**0.5
    # return self.euclidean_distance(Point(0,0))


class Line:

  def __init__(self,A,B,C):
    self.A = A
    self.B = B
    self.C = C

  def __str__(self):
    return '{}x + {}y + {} = 0'.format(self.A,self.B,self.C)

  def point_on_line(line,point):
    if line.A*point.x_cod + line.B*point.y_cod + line.C == 0:
      return "lies on the line"
    else:
      return "does not lie on the line"

  def shortest_distance(line,point):
    return abs(line.A*point.x_cod + line.B*point.y_cod + line.C)/(line.A**2 + line.B**2)**0.5
````

````python
l1 = Line(1,1,-2)
p1 = Point(1,10)
print(l1)
print(p1)

l1.shortest_distance(p1)
````

### Reference Variables

- Reference variables hold the objects
- We can create objects without reference variable as well
- An object can have multiple reference variables
- Assigning a new reference variable to an existing object does not create a new object

## 📖 Reference variables

```python
class Person:
    def __init__(self):
        self.name = 'nitish'
        self.gender = 'male'

p = Person()    # object is referenced by label `p`
q = p           # q bhi same object ko point karega!
```

- Object without reference banaya to wo memory me kho jata hai
- An object can have multiple reference variables
- `q = p` se FORMAT naya object nahi banta — same object, 2 labels
- Multiple ref → `print(id(p))`, `print(id(q))` same address

## 📖 Encapsulation — private attributes

```python
class Atm:
    def __init__(self):
        self.pin = ''          # public attribute
        self.__balance = 0      # PRIVATE: __ prefix

    def get_balance(self):
        return self.__balance

    def set_balance(self, value):
        self.__balance = value
```

- Encapsulation = internal data ko method ke piche chhipana
- `__name` (double underscore) = private — direct access se `AttributeError`
- Clean access methods `get_balance()`, `set_balance()` — gatekeepers
- Benefit: logic baad me change karo, caller ka code toote nahi

````python
class Atm:

  # constructor(special function)->superpower -> 
  def __init__(self):
    print(id(self))
    self.pin = ''
    self.__balance = 0
    #self.menu()

  def get_balance(self):
    return self.__balance

  def set_balance(self,new_value):
    if type(new_value) == int:
      self.__balance = new_value
    else:
      print('beta bahot maarenge')

  def __menu(self):
    user_input = input("""
    Hi how can I help you?
    1. Press 1 to create pin
    2. Press 2 to change pin
    3. Press 3 to check balance
    4. Press 4 to withdraw
    5. Anything else to exit
    """)

    if user_input == '1':
      self.create_pin()
    elif user_input == '2':
      self.change_pin()
    elif user_input == '3':
      self.check_balance()
    elif user_input == '4':
      self.withdraw()
    else:
      exit()

  def create_pin(self):
    user_pin = input('enter your pin')
    self.pin = user_pin

    user_balance = int(input('enter balance'))
    self.__balance = user_balance

    print('pin created successfully')

  def change_pin(self):
    old_pin = input('enter old pin')

    if old_pin == self.pin:
      # let him change the pin
      new_pin = input('enter new pin')
      self.pin = new_pin
      print('pin change successful')
    else:
      print('nai karne de sakta re baba')

  def check_balance(self):
    user_pin = input('enter your pin')
    if user_pin == self.pin:
      print('your balance is ',self.__balance)
    else:
      print('chal nikal yahan se')

  def withdraw(self):
    user_pin = input('enter the pin')
    if user_pin == self.pin:
      # allow to withdraw
      amount = int(input('enter the amount'))
      if amount <= self.__balance:
        self.__balance = self.__balance - amount
        print('withdrawl successful.balance is',self.__balance)
      else:
        print('abe garib')
    else:
      print('sale chor')
````

### Static Variables(Vs Instance variables)

## 📖 Static variables (vs instance)

```python
class Atm:
    __counter = 1            # static: class level, sab objects share

    def __init__(self):
        self.pin = ''        # instance: har object ki copy

```

- **Instance variable** — har object ki apni copy (`self.pin`)
- **Static variable** — class level (`Atm.__counter`), sab objects share
- Static access: `ClassName.variable` (object bina bhi chalega)
- Use-case: counter, config, shared global value

- static var values visit across instances shared
- `self.__counter` inside instance — good for counters/ids
- vs instance var: `self.pin` — per-object
- Use: auto-incrementing IDs, counts, constants

### Static methods

##### Points to remember about static

- Static attributes are created at class level.
- Static attributes are accessed using ClassName.
- Static attributes are object independent. We can access them without creating instance (object) of the class in which they are defined.
- The value stored in static attribute is shared between all instances(objects) of the class in which the static attribute is defined.

## 📖 Static methods & static reminders

```python
class Lion:
    __water_source = 'well in the circus'   # static private

    def __init__(self, name, gender):
        self.__name = name

    def drinks_water(self):
        print(self.__name, 'drinks from', Lion.__water_source)
```

- Static attributes class level create hote hain — `obj` koi nahi
- Static ko `ClassName.attr` se access karo
- Private static `__water_source` ko class ke andar se hi use karo
- Points to remember block ka matlab: object-independent, shared, class-scope

## 🔬 Deep Dive : Static methods

```python
class Lion:
    __water_source = "well in the circus"     # static var
    @staticmethod
    def drink():                             # static method
        print("Drinking from", Lion.__water_source)
```

- `@staticmethod` — method bina `self`
- Called via class: `Lion.drink()` — instance zaroori nahi
- Kab use: helper logic jo state pe dependent nahi
- Utility inside class namespace
- No access to self / instance data
- Can access static var via ClassName
- vs instance method — self required
- vs classmethod `@classmethod` (cls first param)
- Statics: pure functions grouped with class
- Point to remember: static attr created at class load time

````python
class Lion:
  __water_source="well in the circus"

  def __init__(self,name, gender):
      self.__name=name
      self.__gender=gender

  def drinks_water(self):
      print(self.__name,
      "drinks water from the",Lion.__water_source)

  @staticmethod
  def get_water_source():
      return Lion.__water_source

simba=Lion("Simba","Male")
simba.drinks_water()
print( "Water source of lions:",Lion.get_water_source())
````
