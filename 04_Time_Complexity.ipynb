### 1. What is Efficiency in Programming?

Efficiency in programming refers to **how well a program utilizes resources** (time and memory) to accomplish a task.

### Two Programs, Same Task:

**Task:** Find if a number exists in a list of 1 million numbers.

| Approach | Method | Time Taken |
|----------|--------|------------|
| **Inefficient** | Check every element one by one | ~10 seconds |
| **Efficient** | Use binary search | ~0.001 seconds |

Both produce the **same result**, but one is **10,000x faster**.

### What Makes Code Efficient?

| Factor | Description |
|--------|-------------|
| **Time** | How fast the program runs |
| **Memory** | How much RAM the program uses |
| **Readability** | How easy the code is to understand |
| **Scalability** | How performance changes with larger inputs |

> **Goal:** Write code that is both efficient AND readable.

### 2. Why Efficiency is Important?

| Reason | Real-World Impact |
|--------|-------------------|
| **User Experience** | Slow apps = frustrated users, lost customers |
| **Cost** | More time = more cloud computing bills |
| **Scalability** | What works for 100 users may fail for 1 million |
| **Competition** | Faster product = competitive advantage |
| **Battery Life** | Inefficient code drains mobile batteries |
| **Real-time Systems** | Self-driving cars, medical devices need instant response |

### Example: Why It Matters

```
If an algorithm takes:
- 1 second for n = 10
- 2 seconds for n = 20
- 100 seconds for n = 100  <- Might be acceptable
- 1 year for n = 1,000,000 <- UNACCEPTABLE!
```

This is why we study **Time Complexity** - to predict how our code will perform as input grows.

### 3. Types of Efficiency

## Space Efficiency vs Time Efficiency

| Type | What It Measures | Trade-off |
|------|------------------|-----------|
| **Time Efficiency** | How fast the program runs | Speed |
| **Space Efficiency** | How much memory the program uses | Memory |

### The Trade-off

Often, improving one reduces the other:

```
Example: Find duplicates in a list

Approach 1: Use nested loops (O(n^2) time, O(1) space)
  - Slow but uses minimal memory

Approach 2: Use a hash set (O(n) time, O(n) space)
  - Fast but uses more memory
```

## Our Focus: Time Efficiency

In this notebook, we focus on **Time Efficiency** because:
- It's easier to measure and compare
- Time is often the critical constraint
- Understanding time complexity helps predict scalability

### 4. Techniques to Measure Time Efficiency

There are **3 main approaches** to measure time efficiency:

| # | Method | Description | Pros | Cons |
|---|--------|-------------|------|------|
| 1 | **Measuring Execution Time** | Run code and measure actual time | Simple, real numbers | Depends on hardware |
| 2 | **Counting Operations** | Count how many operations execute | Precise | Tedious for complex code |
| 3 | **Orders of Growth (Big O)** | Abstract classification of growth rate | Universal, scalable | Loses exact details |

We'll explore each method in detail.

### 5. Method 1: Measuring Execution Time

The simplest approach: **Run the code and measure how long it takes**.

### Python Example:

````python
import time

# Method: Measure time using time module
start = time.time()

# Code to measure
total = 0
for i in range(1000000):
    total += i

end = time.time()

print(f'Time taken: {end - start:.6f} seconds')
````

### Alternative: Using timeit for More Accurate Results

````python
import timeit

# More accurate measurement (runs multiple times)
code_to_test = '''
total = 0
for i in range(1000000):
    total += i
'''

time_taken = timeit.timeit(code_to_test, number=100)
print(f'Average time per run: {time_taken/100:.6f} seconds')
````

### Problems with This Approach

| Problem | Explanation |
|---------|-------------|
| **Hardware dependent** | Same code runs faster on better hardware |
| **OS dependent** | Windows vs Linux vs Mac give different times |
| **Background processes** | Other programs affect timing |
| **Not scalable** | Can't test for all input sizes |
| **No mathematical proof** | Just empirical evidence |

### Example of Hardware Dependency:

```
Same code on different machines:
- Old laptop:    2.5 seconds
- New laptop:    0.8 seconds  (3x faster!)
- Gaming PC:     0.2 seconds  (12x faster!)
```

> **Conclusion:** Measuring time is useful for quick tests but NOT for formal analysis.

### 6. Method 2: Counting Operations

Instead of measuring actual time, **count how many basic operations** the code performs.

### Basic Operations Include:
- Arithmetic: `+`, `-`, `*`, `/`, `%`, `//`
- Comparisons: `>`, `<`, `==`, `!=`
- Assignments: `=`
- Array accesses: `A[i]`
- Function calls: `func()`

### Example: Digit Extraction

````python
n = 345

sum = 0
while n > 0:
    sum = sum + n%10
    n = n // 10

print(sum)
````

### Operation Count:

| Iteration | Operations |
|-----------|------------|
| Start | `n = 345` (1), `sum = 0` (1) |
| 1st | `n > 0` (1), `sum + n%10` (2), `n // 10` (1), assignments (2) |
| 2nd | `n > 0` (1), `sum + n%10` (2), `n // 10` (1), assignments (2) |
| 3rd | `n > 0` (1), `sum + n%10` (2), `n // 10` (1), assignments (2) |
| Exit | `n > 0` (1) |

**Total:** For a 3-digit number, approximately **15 operations**

For a **d-digit number**: Approximately **6d operations**

### Problems with Counting Operations

| Problem | Example |
|---------|---------|
| **Tedious** | Manually counting every operation is time-consuming |
| **Inconsistent** | Different ways to count give different numbers |
| **Misses the big picture** | `100n` vs `100n + 50` look different but scale the same |
| **Too detailed** | We care about growth rate, not exact count |

### What We Really Care About:

How does the number of operations **grow** as input grows?

```
Input:    10    100    1000    10000
Ops A:    10    100    1000    10000    -> Linear growth
Ops B:    100   100    100     100      -> Constant (stays same)
Ops C:    1     100    10000   1000000  -> Quadratic growth
```

> **This leads us to the next method: Orders of Growth**

### 7. Method 3: Orders of Growth (Big O Notation)

The **abstract notion of order of growth** describes how the running time of an algorithm grows as the input size grows.

### Big O Notation

Big O notation describes the **upper bound** of growth rate. It focuses on:

1. **Dominant term** - The term that grows fastest
2. **Constants dropped** - `2n` becomes `n`
3. **Lower order terms dropped** - `n + n^2` becomes `n^2`

### Why Drop Constants?

```
Algorithm A: 2n operations
Algorithm B: 100n operations

At n = 1000:
- A: 2,000 operations
- B: 100,000 operations (50x more!)

At n = 1,000,000:
- A: 2,000,000 operations
- B: 100,000,000 operations (50x more!)

Both grow LINEARLY (constant factor 50 doesn't change)
So both are O(n)
```

### 8. Types of Order of Growth

| Big O | Name | Growth Rate | Example |
|-------|------|-------------|--------|
| `O(1)` | Constant | Does not grow | Array access |
| `O(log n)` | Logarithmic | Very slow growth | Binary search |
| `O(n)` | Linear | Grows with input | Simple loop |
| `O(n log n)` | Linearithmic | Slightly more than linear | Merge sort |
| `O(n^2)` | Quadratic | Grows fast | Nested loops |
| `O(n^3)` | Cubic | Grows very fast | Triple nested loops |
| `O(2^n)` | Exponential | Doubles each time | Recursive Fibonacci |
| `O(n!)` | Factorial | Extremely fast | Permutations |

### Visual Growth Comparison (for n = 1000)

```
O(1)        = 1
O(log n)    = 10
O(n)        = 1,000
O(n log n)  = 10,000
O(n^2)      = 1,000,000
O(n^3)      = 1,000,000,000
O(2^n)      = 1.07 x 10^301 (universe would end first!)
O(n!)       = Even worse!
```

### 9. Code Examples with Complexity Analysis

Let's analyze real code examples to understand complexity.

### Example 1: Integer to String Conversion

````python
number = int(input('enter the number'))

digits = '0123456789'
result = ''
while number != 0:
    result = digits[number % 10] + result
    number = number // 10

print(result)
````

### Analysis:
- The loop runs **d times** where d = number of digits
- Number of digits in n = log₁₀(n)
- Each iteration: O(1) operations

**Time Complexity: O(log n)**

### Example 2: Sum of List

````python
L = [1, 2, 3, 4]

sum = 0
for i in L:
    sum = sum + i

print(sum)
````

### Analysis:
- Loop runs **n times** (n = length of list)
- Each iteration: O(1) operations

**Time Complexity: O(n)** - Linear

### Example 3: Product of List

````python
L = [1, 2, 3, 4]

product = 1
for i in L:
    product = product * i

print(product)
````

### Analysis:
- Same as Example 2 - single loop through list

**Time Complexity: O(n)** - Linear

### Example 4: Nested Loops (Two Arrays)

````python
A = [1, 2, 3, 4]
B = [5, 6, 7, 8]

for i in A:
    for j in B:
        print(i, j)
````

### Analysis:
- Outer loop runs **n times** (length of A)
- Inner loop runs **m times** (length of B) for EACH outer iteration
- Total: n x m operations
- If n = m: n x n = n²

**Time Complexity: O(n x m)** or **O(n²)** if same size

### Example 5: Triple Nested Loop with Constant Range

````python
A = [1, 2, 3, 4]
B = [5, 6, 7, 8]

for i in A:
    for j in B:
        for k in range(1000):
            print(i, j)
````

My PC Not able to handle that much of times
### Analysis:
- Outer loop: n times
- Middle loop: m times
- Inner loop: **1,000,000 times** (constant!)
- Total: n x m x 1,000,000
- But 1,000,000 is a **constant**, so we drop it

**Time Complexity: O(n x m)** or **O(n²)**

> **Key insight:** A large constant factor doesn't change the complexity class!

## Example 6: List Reversal (In-Place)

````python
L = [1, 2, 3, 4, 5]

for i in range(0, len(L)//2):
    other = len(L) - i - 1
    temp = L[i]
    L[i] = L[other]
    L[other] = temp

print(L)
````

### Analysis:
- Loop runs **n/2 times**
- Each iteration: O(1) operations
- n/2 is still O(n) (constants dropped)

**Time Complexity: O(n)** - Linear

## Example 7: Complex Nested Loop

````python
n = 10
k = 0
for i in range(n//2, n):
    for j in range(2, n, pow(2, 0)):
        k = k + n / 2

print(k)
````

### Analysis:
- Outer loop: from n/2 to n = **n/2 iterations**
- Inner loop: range(2, n, 1) = **n-2 iterations** (step=1)
- Total: (n/2) x (n-2) = approximately n²/2
- Drop constants: O(n²)

**Time Complexity: O(n²)** - Quadratic

## Example 8: Division Without Division Operator

````python
a = 10
b = 3

if b <= 0:
    print(-1)

div = a // b

print(div)
````

### Analysis:
- No loops
- Just constant operations: comparison, division, print

**Time Complexity: O(1)** - Constant

## Example 9: Sum of Digits

````python
n = 345

sum = 0
while n > 0:
    sum = sum + n%10
    n = n // 10

print(sum)
````

### Analysis:
- Loop runs **d times** (number of digits)
- d = log₁₀(n)
- Each iteration: O(1) operations

**Time Complexity: O(log n)** - Logarithmic

| n | Digits (log₁₀(n)) | Operations |
|---|-------------------|------------|
| 10 | 1 | 1 |
| 100 | 2 | 2 |
| 1,000 | 3 | 3 |
| 1,000,000 | 6 | 6 |

Even for **1 million**, only **6 iterations**! This is very efficient.

# 10. Recursive Complexity Analysis

Recursive functions require **recurrence relations** to analyze.

## Recursive Fibonacci

````python
def fib(n):
    if n == 1 or n == 0:
        return 1
    else:
        return fib(n-1) + fib(n-2)
````

### Recurrence Relation:

```
T(n) = T(n-1) + T(n-2) + 1,  if n > 0
T(n) = 1,                     otherwise
```

### Call Tree for fib(4):

```
                fib(4)
               /      \
          fib(3)      fib(2)
         /     \      /    \
      fib(2)  fib(1) fib(1) fib(0)
      /    \
   fib(1) fib(0)
```

- fib(4) makes 9 function calls
- fib(5) makes 15 function calls
- The number of calls **doubles** with each increment

**Time Complexity: O(2^n)** - Exponential!

| n | Operations | Time (rough) |
|---|------------|--------------|
| 10 | 177 | 0.001 sec |
| 20 | 13,463 | 0.1 sec |
| 30 | 1,346,269 | 10 sec |
| 40 | 134,626,931 | 1000 sec (17 min) |
| 50 | 13,462,693,271 | 140 days! |
| 60 | would take 31,710 years! | |

## Another Recurrence Example

```
T(n) = 3T(n-1) + 1,  if n > 0
T(n) = 1,            otherwise
```

This means:
- Each call makes **3 recursive calls**
- Plus 1 constant operation
- Base case returns 1

**Complexity: O(3^n)** - Even worse than O(2^n)!

## Subset Algorithm Recurrence

```
T(n) = 2T(n-1) - 1,  if n > 0
T(n) = 1,            otherwise
```

**Complexity: O(2^n)** - Generates all subsets

## Summary: Complexity Cheat Sheet

| Complexity | Name | Code Pattern | Example |
|------------|------|--------------|---------|
| `O(1)` | Constant | No loops, direct access | `arr[5]`, `a + b` |
| `O(log n)` | Logarithmic | Divide by 2 each time | Binary search, digit extraction |
| `O(n)` | Linear | Single loop | Sum of list, linear search |
| `O(n log n)` | Linearithmic | Divide + combine | Merge sort, quick sort (avg) |
| `O(n²)` | Quadratic | Nested loops | Bubble sort, selection sort |
| `O(n³)` | Cubic | Triple nested loops | Matrix multiplication |
| `O(2^n)` | Exponential | Two recursive calls | Naive Fibonacci |

### Which Ones are Acceptable?

```
For n = 1,000,000:

O(1)         -> 1 operation           ✅ GREAT
O(log n)     -> 20 operations         ✅ GREAT
O(n)         -> 1,000,000 operations  ✅ GOOD
O(n log n)   -> 20,000,000 operations ✅ OK
O(n^2)       -> 10^12 operations      ❌ TOO SLOW
O(2^n)       -> Would take millions of years ❌ IMPOSSIBLE
```
