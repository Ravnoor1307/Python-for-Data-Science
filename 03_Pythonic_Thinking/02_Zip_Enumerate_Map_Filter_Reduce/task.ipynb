## ⭐⭐ Task 3: Uneven Zipper + Guard 🚨

`a = [1, 2, 3, 4]`, `b = ["x", "y"]`

1. Normal `zip(a, b)` list print — kitne pairs?
2. `zip(a, b, strict=True)` try/except ValueError print
3. `zip(*[("n", 1), ("i", 2), ("c", 3)])` — unzip 2 lists print

💡 **Hint:** shortest-list rule! strict guard crash karega jaise hi size-mismatch mile.

```python
# TODO 1-3
```

<details><summary>✅ Solution</summary>

2 pairs (shortest 'b' won!); strict → ValueError "argument 2 is shorter than argument 1";
unzip → ('n','i','c')/'(1,2,3). Guard protection ON! 🛡️
</details>

```python
# ✅ SOLUTION — Task 3
a = [1, 2, 3, 4]
b = ["x", "y"]
print("normal  :", list(zip(a, b)))
try:
    list(zip(a, b, strict=True))
except ValueError as e:
    print("strict 🚨:", e)
pack = [("n", 1), ("i", 2), ("c", 3)]
letters, digits = zip(*pack)
print("unzip   :", letters, digits)
# Expected OUTPUT:
# normal  : [(1, 'x'), (2, 'y')]
# strict 🚨: zip() argument 2 is shorter than argument 1
# unzip   : ('n', 'i', 'c') (1, 2, 3)
```

## ⭐⭐ Task 4: Spray & Gate Demo 🎨✂️

`marks = [67, 82, 45, 91, 38, 74]`

1. `map` lambda se har marks +5 grace print list
2. `filter` lambde se grace-ke-BAAD ≥80 list
3. `sum(1 for ...)` se count ≥80 (list BINA!) print

💡 **Hint:** chain karo ya two-step — dono chale; compositions readable! step-wise try!

```python
# TODO 1-3
```

<details><summary>✅ Solution</summary>

Grace list [72,87,50,96,43,79]; ≥80 filter [87,96]; pythonic count=2.
</details>

```python
# ✅ SOLUTION — Task 4
marks = [67, 82, 45, 91, 38, 74]
grace = list(map(lambda m: m + 5, marks))
print("grace   :", grace)
passed = list(filter(lambda m: m >= 80, grace))
print("pass≥80 :", passed)
count = sum(1 for m in grace if m >= 80)
print("count   :", count)
# Expected OUTPUT:
# grace   : [72, 87, 50, 96, 43, 79]
# pass≥80 : [87, 96]
# count   : 2
```

## ⭐⭐⭐ Task 5: Blender End — Reduce 🌀

`runs = [12, 45, 7, 89, 23]`

1. `reduce` sum (batul sum() bhi print — pythonic-compare!)
2. `reduce` max via `a if a > b else b`
3. `reduce` PRODUCT — math.prod() se compare! (`import math`)
4. `reduce` pe EMPTY list + initial=100 print (survives?)

💡 **Hint:** initial arg reduce ka third parameter — empty-input ka dava!

```python
# TODO 1-4
```

<details><summary>✅ Solution</summary>

sum 176, max 89, prod 995,400 — built-ins se agree; empty+initial=100 (survived!)
</details>

```python
# ✅ SOLUTION — Task 5
import math
runs = [12, 45, 7, 89, 23]
print("reduce sum :", reduce(lambda a, b: a + b, runs), "| builtin:", sum(runs))
print("reduce max :", reduce(lambda a, b: a if a > b else b, runs), "| builtin:", max(runs))
print("reduce prod:", reduce(lambda a, b: a * b, runs), "| math.prod:", math.prod(runs))
print("empty+init :", reduce(lambda a, b: a + b, [], 100))
# Expected OUTPUT:
# reduce sum : 176 | builtin: 176
# reduce max : 89 | builtin: 89
# reduce prod: 995400 | math.prod: 995400
# empty+init : 100
```

1. **map + zip**: 18% GST prices first-3 ke names saath print
2. **filter**: prices > 10,000 (premium shelf!) — names with int-prices print
3. **reduce/len**: inventory TOTAL value (price × stock for all) print (₹ with commas!)

💡 **Hint:** `f"{value:,.0f}"` commas + no-decimals formatting!

```python
# TODO 1-3
```

<details><summary>✅ Solution</summary>

GST triplet correct; 4 premium items; inventory ₹16,629,090 — warehouse report card done! 🏪
</details>

```python
# ✅ SOLUTION — Task 6
gst_first3 = list(zip(pnames, map(lambda p: round(p * 1.18, 2), prices)))[:3]
print("GST first-3 :", gst_first3)

premium = [(n, int(p)) for n, p in zip(pnames, prices) if p > 10_000]
print("Premium     :", premium)

inv = reduce(lambda acc, pair: acc + pair[0] * pair[1], zip(prices, stocks), 0)
print(f"Inventory   : ₹{inv:,.0f}")
# Expected OUTPUT:
# GST first-3 : [('Laptop', 56640.0), ('Smartphone', 17700.0), ('Headphones', 2360.0)]
# Premium     : [('Laptop', 48000), ('Smartphone', 15000), ('Tablet', 22000), ('DSLR_Camera', 35000)]
# Inventory   : ₹16,629,090
```

## 🏆 Task 7 (💾 BOSS): CLASS REPORT CARD — Sab Machines Ek Saath! 🎓🏭

Principal-Madam ko ek perfect class-report chahiye — tumhari 4 machines
lagegi: leaderboard sorted(ZIP) + RANKS(enumerate) + club-filter(filter) + stats!

Karna kya hai:
1. **Leaderboard**: `sorted(zip(avgs, names), reverse=True)[:5]` + enumerate rank —
   `  #1 │ Manish_29  │ avg 83.60` format (avg 2 decimals!)
2. **Distinction club** (avg ≥ 80): filter — names print sorted-by-avg DESC
3. **Class stats card**: class-avg (reduce/len) + how-many-above-avg (genexpr count!)
4. **Subject topper**: '<SUBS>: toppers ka combo' — simple version:
   computer-subject me TOP SCORER name+score (max with zip!) print
5. **Sign-off**: "— Card generated via 4-machines assembly 🏭"

💡 **Hints:**
- `zip(avgs, names)` sorted reverse — first element avg decides rank
- above-avg count: `sum(1 for a in avgs if a > class_avg)` — pythonic!
- computer scores: `[int(s["computer"]) for s in students]` + zip(names,...) + max(key=t[1]!

**Challenge:** f-string `{avg:>5.2f}` alignment — columns STRAIGHT! 🖨️

```python
# TODO 1-5
```

<details><summary>✅ Solution — PRINCIPAL APPROVED! 📜</summary>

Leaderboard Manish_29 #1 (83.60) → Priya_44 #5; distinction=2 (⩾80); class-avg 65.69,
27 above; computer-topper Kavya_22 (perfect 100!). Report-card laminated! 🎓
</details>

```python
# ✅ SOLUTION — Task 7 (Report Card BOSS 🎓)
print("  🎓 CLASS OF 2024 — REPORT CARD")
print("  " + "─" * 34)

# 1) Leaderboard top-5 (zip + sorted + enumerate!)
board = sorted(zip(avgs, names), reverse=True)
for rank, (avg, nm) in enumerate(board[:5], start=1):
    print(f"  #{rank} │ {nm:<12} │ avg {avg:>5.2f}")
print("  " + "─" * 34)

# 2) Distinction club (≥80) — filter! (zip(avgs, names) — na[0]=avg, na[1]=name!)
dist = sorted(filter(lambda na: na[0] >= 80, zip(avgs, names)), reverse=True)
print("  🎖️ Distinction:", [f"{n} ({a:.1f})" for a, n in dist])

# 3) Class stats
class_avg = reduce(lambda a, b: a + b, avgs) / len(avgs)
above = sum(1 for a in avgs if a > class_avg)
print(f"  📊 class avg : {class_avg:.2f} | above-avg : {above}/{len(avgs)} students")

# 4) Subject topper (computer) — max with key!
cs = [int(s["computer"]) for s in students]
topper = max(zip(cs, names))
print(f"  💻 Computer topper: {topper[1]} ({topper[0]}/100)")

# 5) Sign-off
print("  — Card generated via 4-machines assembly 🏭")
# Expected OUTPUT:
#   🎓 CLASS OF 2024 — REPORT CARD
#   ──────────────────────────────────
#   #1 │ Manish_29    │ avg 83.60
#   #2 │ Ananya_20    │ avg  80.20
#   #3 │ Arjun_11     │ avg  79.40
#   #4 │ Sneha_24     │ avg  78.60
#   #5 │ Priya_44     │ avg  78.00
#   ──────────────────────────────────
#   🎖️ Distinction: ['Manish_29 (83.6)', 'Ananya_20 (80.2)']
#   📊 class avg : 65.69 | above-avg : 27/50 students
#   💻 Computer topper: Kavya_22 (100/100)
#   — Card generated via 4-machines assembly 🏭
```
