```python
# ⚙️ SETUP — lazy_amounts reader (nb2 wala hi, shortcut!)
import csv, sys
from itertools import islice

def lazy_amounts(top_n=None):
    for path in ["../../00_Resources/datasets/sales_data.csv",
                 "00_Python/00_Resources/datasets/sales_data.csv"]:
        try:
            with open(path, newline="", encoding="utf-8") as f:
                for i, row in enumerate(csv.DictReader(f)):
                    if top_n is not None and i >= top_n:
                        return
                    yield row["region"], int(row["quantity"]) * int(row["unit_price"])
            return
        except FileNotFoundError:
            continue
    raise FileNotFoundError("sales_data.csv nahi mila!")

print("Lazy reader ready! (maango → milega, ek-ek 🫖)")
# Expected OUTPUT:
# Lazy reader ready! (maango → milega, ek-ek 🫖)
```

## ⭐ Task 2: For-Loop ka Khud-ka Replica 🔧

`nums = [7, 14, 21]` ka **ARITHMETIC mean** print karo — par `for` ya `sum()`/`len()`
jaise shortcuts BINA; sirf manual `iter+next+while` se! (Yaad: StopIteration pe break)

💡 **Hint:** Running total + count dono while me badhao; mean = total/count.

```python
# TODO
```

<details><summary>✅ Solution</summary>

mean = 42/3 = 14.0 — internal-engine pro 😎.
</details>

```python
# ✅ SOLUTION — Task 2
nums = [7, 14, 21]
it = iter(nums)
total, count = 0, 0
while True:
    try:
        n = next(it)
    except StopIteration:
        break
    total += n
    count += 1
print("mean:", total / count)
# Expected OUTPUT:
# mean: 14.0
```

## ⭐⭐ Task 4: Apna Chhota Generator — `evens_until(n)` 🏭

Function likho jo 0 se `n` tak saare EVEN numbers yield kare (0 included):
1. Generator object print (type!)
2. `evens_until(10)` ka list print

💡 **Hint:** `range(0, n+1, 2)` loop inside; har x pe `yield x`. Yaad — return nahi!

```python
# TODO 1-2
```

<details><summary>✅ Solution</summary>

[0, 2, 4, 6, 8, 10] — factory ready ma'am 🏭!
</details>

```python
# ✅ SOLUTION — Task 4
def evens_until(n):
    for x in range(0, n + 1, 2):
        yield x

gen = evens_until(10)
print("type:", type(gen).__name__)
print("values:", list(gen))
# Expected OUTPUT:
# type: generator
# values: [0, 2, 4, 6, 8, 10]
```

3. Ratio compute (int division `//`)

```python
# TODO 1-3
```

<details><summary>✅ Solution</summary>

85,176 vs 200 bytes — 425× chhota! Yehi example PPT me daalo jab boss pooche

```python
# ✅ SOLUTION — Task 5
l10k = [x * x for x in range(10_000)]
g10k = (x * x for x in range(10_000))
print(f"list bytes: {sys.getsizeof(l10k):,}")
print(f"gen  bytes: {sys.getsizeof(g10k):,}")
print("ratio     :", sys.getsizeof(l10k) // sys.getsizeof(g10k), "× smaller!")
# Expected OUTPUT:
# list bytes: 85,176
# gen  bytes: 200
# ratio     : 425 × smaller!
```
