## 🌍 Real-World Hook: Chai-tapri ka coupon-book ☕

Tapri wale bhaiya coupons dete hain: 10-coupon ki gaddi 📒.
- **Coupon-book** = **ITERABLE** (shelf/item ka group — sab coupons hai!)
- **Ek-ek phaat-te (tear) jaana** = **ITERATOR** — jo coupon phat gaya,
  wapas nahi milta; sab khatm → "Coupon khatam bhaiya!" (StopIteration 🚉)

```text
List [a,b,c] (ITERABLE — shelf 📚)
   │ iter()
   ▼
iter-object (ITERATOR — reader 🕵️) → next()→ a → next()→ b → next()→ c → 🛑StopIteration
(reader ko reset karne = naya iter() chahiye — wapas index-0 NAHI hota!)
```

### 📖 Do ONE-LINE definitions (interview-grade!)

> **ITERABLE:** jis object pe `.iter()` chal jaye / `for` loop accept kare —
> List, str, dict, set, tuple, range... (shelf hai!)
>
> **ITERATOR:** jis pe `next()` chale — **state yaad rakhta** (kaun sa page
> abhi haath me!) — ek-directional stream. Exhaust ho jaye to STOP 🛑

````python
# 🐣 Manual iter/next — for-loop ka hathiyar directly!
fruits = ["apple", "mango", "kiwi"]
it = iter(fruits)                      # shelf se reader uthao!
print("Type fruits :", type(fruits).__name__)
print("Type iter() :", type(it).__name__, "← ek ALAG machine!")

print("next 1:", next(it))
print("next 2:", next(it))
print("next 3:", next(it))
try:
    next(it)
except StopIteration:
    print("next 4: 🛑 StopIteration — coupon khatam bhaiya!")
# Expected OUTPUT:
# Type fruits : list
# Type iter() : list_iterator ← ek ALAG machine!
# next 1: apple
# next 2: mango
# next 3: kiwi
# next 4: 🛑 StopIteration — coupon khatam bhaiya!
````

````python
# 🐣 LIST khud ITERATOR nahi hai! (protocol proof)
l = [10, 20]
try:
    next(l)
except TypeError as e:
    print("next(list) 🚨:", e, "\n→ pehle iter(l) chahiye!")

# iter(iterable) ≠ iterable; iter(iterator) = SAME object (self-iterable!)
it = iter(l)
print("\niter(l) is l      :", it is l, "← naya machine ✅")
print("iter(it) is it    :", iter(it) is it, "← reader khud iterate ho sakta!")
# Expected OUTPUT:
# next(list) 🚨: 'list' object is not an iterator
# → pehle iter(l) chahiye!
#
# iter(l) is l      : False ← naya machine ✅
# iter(it) is it    : True ← reader khud iterate ho sakta!
````

````python
# 🐥 For-loop ka UNDER-THE-HOOD (UA) reconstruction 🔧
nums = [10, 20, 30]
print("for-loop karega:")
for n in nums:
    print("  body:", n)

print("\nAsli engine (while+next manual):")
it = iter(nums)
while True:
    try:
        n = next(it)
    except StopIteration:
        break
    print("  body:", n)
# Expected OUTPUT:
# for-loop karega:
#   body: 10
#   body: 20
#   body: 30
#
# Asli engine (while+next manual):
#   body: 10
#   body: 20
#   body: 30
````

````python
# 🐥 Sab iterables pe iter() — ek protocol, saare shelves! 📚
print("str       :", list(iter("abc")))
print("tuple     :", list(iter((1, 2, 3))))
print("dict      :", list(iter({"name": "Ravi", "age": 25})), "← ITERATES KEYS!")
print("set       :", sorted(iter({3, 1, 2})), "← set unordered, sorted karke dikhaya")
print("range     :", list(iter(range(3))))
# Expected OUTPUT:
# str       : ['a', 'b', 'c']
# tuple     : [1, 2, 3]
# dict      : ['name', 'age'] ← ITERATES KEYS!
# set       : [1, 2, 3] ← set unordered, sorted karke dikhaya
# range     : [0, 1, 2]
````

````python
# 🐔 THE SILENT-EMPTY TRAP 🤫 — sabse sneaky bug!
# Iterator EK-BAARI ek-dish hai — dobaara khaane jaoge, khaali plate mile!!
data = iter([100, 200, 300])
print("first pass  :", sum(data), "← poora 600 ✅")
print("second pass :", sum(data), "← ZERO?! 😱 (iterator already exhausted!)")

# Yehi cheez list pe repeat-safe hai (bar-bar naya iter!) —
for_times = [1, 2, 3]
print("\nlist sum 1st:", sum(for_times))
print("list sum 2nd:", sum(for_times), "← list repeat-safe kyunki sum() khud iter() banata!")
# Expected OUTPUT:
# first pass  : 600 ← poora 600 ✅
# second pass : 0 ← ZERO?! 😱 (iterator already exhausted!)
#
# list sum 1st: 6
# list sum 2nd: 6 ← list repeat-safe kyunki sum() khud iter() banata!
````

### 🔍 WHY does this design exist?

Ek-baar-use stream super-USEFUL hai jab DATA BADI ho — poora load karne ki
zaroorat nahi, ek-ek process karte chalo (streaming 🌊!). Yehi Iterator ka
SUPER-ROADMAP — next notebook me `yield` wale **generators** se duniya chaap!

Sheet-style example: weather-station 24×365 readings — ek second me ek
reading process karna = perfect iterator-style workflow (nb2 memory proof!).

## ⚠️ 5 Common Mistakes (Iterator Special)

| # | ❌ Galti | ✅ Sahi |
|---|---|---|
| 1 | `next([1,2])` list pe direct | `next(iter([1,2]))` — list itrABLE hai, itrATOR nahi! |
| 2 | Iterator dobaara "rewind" expect | Naya `iter(...)` banao — reader reset nahi hota |
| 3 | `for x in data:` exhaust ke baad dobara use | `data` ITERATOR tha ya ITERABLE? pata rakho! (silent empty) |
| 4 | `list(iter({..}))` se VALUES expect | Dict iter = KEYS; values: `.values()` iter |
| 5 | Infinite stream pe `list(...)` wrap | Khatam hi nahi hoga — hang! Use `islice` |

> 🎯 **Rule-of-feel:** `iter()` = fridge se plate nikalna; `next()` = ek bite;
> `StopIteration` = plate khaali. Naya plate chahiye? naya `iter()`! 🍽️

## 📌 Recap — 30 Second Me

- Iterable = jis pe `iter()` chale (shelf); Iterator = jis pe `next()` chale (reader with state).
- `for x in obj:` internally = `iter(obj)` + loop of `next()` + `StopIteration` catch → loop end.
- Iterator **self-iterable** (`iter(it) is it` = True); lists NOT iterators ✔️.
- Dict iterates **keys**; str iterates chars; sets unordered (sorted karke dikhao).
- ⚠️ Exhausted iterator = silent-empty plate (second use gives 0/empty!) — sneaky!
