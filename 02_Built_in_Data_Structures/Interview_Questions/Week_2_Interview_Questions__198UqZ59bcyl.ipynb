## 📖 Aliasing — same object, 2 names

```python
a = 4
b = a
id(a) == id(b)   # True — dono ek hi object
```

- `b = a` koi naya object nahi banata — bas naya name
- `id(obj)` — object ka unique address
- Immutable (int/str) pe pehle se existing object bhi reused
- `sys.getrefcount(x)` — kitne refs hain
- `del a` — name hataya, object tab tak rehta jab tak refs
- Strings/intergers — same value, same id (interning)

## 📖 Aliasing vs mutation

```python
L = [1, 2, 3]
L.append(4)      # MUTATION — id SAME
T = (1, 2, 3)
T = T + (4,)     # tuple immutable → naya object, id DIFF
a = [1, 2, 3]; b = a
b.append(4)      # a bhi badal jayega (same list!)
```

- List mutable — `append` same object modify
- Tuple immutable — + se naya tuple
- Alias = risk: ek jagah change, sab jagah dikhe
- Function me list pass → `func(a[:])` slice copy bhejo

````python
L = [1,2,3]
print(id(L))

L.append(4)
print(L)
print(id(L))
````

````python
T = (1,2,3)
print(id(T))

T = T + (4,)

print(T)
print(id(T))
````

````python
a = [1,2,3]
b = a

b.append(4)
print(b)
````

````python
print(a)
````

````python
def func(data):
  data.append(4)

a = [1,2,3]
func(a[:])
print(a)
````

## 📖 Cloning — `[:]` and `.copy()`

```python
a = [1, 2, 3]
b = a[:]        # copy — nayi list
id(a) != id(b)  # True
b.append(4)     # a unaffected
d = {'name': 'nitish'}
e = d.copy()    # dict copy
```

- `a[:]`, `list(a)`, `a.copy()` — 1D list clone
- `.copy()` — dict clone too
- Top-level independent: append/change original ko nahi hilta
- Lekin **shallow** — nested jo change hota hai wo shared

````python
a = [1,2,3]
# cloning
b = a[:]
````

````python
id(a)
````

````python
id(b)
````

````python
b.append(4)
b
````

````python
a
````

````python
a = {'name':'nitish','age':33}
# aliasing->cloning
b = a.copy()

b['gender'] = 'male'

print(b)
print(a)
````

````python
a = [1,2,3]
# shallow
b = a.copy()

b.append(4)

print(a)
print(b)
````

## 📖 Shallow copy — nested trap

```python
a = [1, 2, 3, [4, 5]]
b = a.copy()        # SHALLOW
id(a[-1]) == id(b[-1])   # True — inner list SHARED
b[-1][0] = 400      # a[-1] bhi change ho jayegi!
```

- Shallow copy sirf outer layer copy karta
- Nested list same object share
- `a[:]` bhi shallow hai
- Fix: `copy.deepcopy(a)` (age capa)

````python
a = [1,2,3,[4,5]]
a
````

````python
print(id(a[-1]))
print(id(b[-1]))
````

````python
b = a.copy()
b
````

````python
print(id(a))
print(id(b))
````

````python
b[-1][0] = 400
print(b)
````

````python
a
````

## 📖 Deep copy

```python
import copy
b = copy.deepcopy(a)
id(a[-1]) != id(b[-1])   # True — inner bhi naya
b[-1][0] = 400           # a me koi farak nahi
```

- `deepcopy` — harr level copy hota hai
- Fully independent clone
- Lists/dicts nested ho to deepcopy use
- Static strings immutable — same object he ok

````python
import copy

a = [1,2,3,[4,5]]
a
````

````python
b = a[:]
b
````

````python
b[-1][0] = 400
b
````

````python
a
````

````python
print(id(a[-1]))
print(id(b[-1]))
````

````python
# [:] -> shallow copy
````
