## 📖 1. kth largest/smallest

```python
L = [12, 23, 1, 4, 56, 34, 22, 3]
L.sort()            # ascending
L[k - 1]            # kth smallest
L[-k]               # kth largest
```

- `sort()` pehle, phir index
- kth smallest = index k-1
- kth largest = index -k
- O(n log n) sort based — interview me ok

````python
# 1. Find the kth largest/smallest item from a list

L = [12,23,1,4,56,34,22,3]
k=3

L.sort(reverse=True)
print(L[k-1])
````

## 📖 2. Check array sorted

```python
L = [1, 2, 3, 4, 5]
flag = True
for i in range(len(L) - 1):
    if L[i] > L[i + 1]:
        flag = False
        break
```

- Ek loop, adjacent compare
- `L[i] > L[i+1]` — descending nahi
- jald hi break = efficient

````python
# 2. Check if an array is sorted
L = [1,2,3,4,5]

flag = True

for i in range(0,len(L)-1):
  if L[i] > L[i+1]:
    flag = False

if flag:
  print('sorted')
else:
  print('not sorted')
````

## 📖 3. Min / Max

L = [21, 1, 34, 23, 54, 11, 10]

## 📖 4. First element occurred k times

```python
d = {}            # count map
for i in L:
    d[i] = d.get(i, 0) + 1
    if d[i] == k:
        print(i)   # jaldi waala
        break
```

- Dict se frequency count
- `dict.get(k, default)` — safe counting
- First time count == k → wo answer

````python
# 4. Find the first element to occur k times in an array

L = [1,1,2,3,4,4,5,5]

k = 2

d = {}

for i in L:
  if i in d:
    d[i] = d[i] + 1
  else:
    d[i] = 1

for i in d:
  if d[i] == k:
    print(i)
    break
````

## 📖 5. Duplicates

```python
d = {}
for i in L:
    if i in d:
        print(i)      # duplicate
    else:
        d[i] = 1
```

- Seen set/dict tracking
- doosri baar dikha → duplicate
- O(n) time, O(n) space

````python
# 5. Find duplicates in an array
L = [1,1,2,3,4,4,5,5]

d = {}

for i in L:
  if i in d:
    d[i] = d[i] + 1
  else:
    d[i] = 1

for i in d:
  if d[i] > 1:
    print(i)


````

## 📖 6. Left rotate by d

```python
L = [1, 2, 3, 4, 5]; d = 2
for i in range(d):
    L.append(L.pop(0))     # pop front → push end
```

- Har rotate: first element hatay, end pe lagao
- `L[1:] + L[:1]` — slice trick bhi
- d rotations

````python
# 6. Rotate array to left d items

L = [1,2,3,4,5]
rotate = 2

for i in range(rotate):
  temp = L[0]
  for j in range(0,len(L)-1):
    L[j] = L[j+1]
  L[len(L)-1] = temp

print(L)
````

## 📖 7. Intersection of sorted arrays

```python
while i < len(a) and j < len(b):
    if a[i] == b[j]:
        print(a[i]); i += 1; j += 1
    elif a[i] < b[j]:
        i += 1
    else:
        j += 1
```

- Two pointers approach
- Dono sorted → compare-advance
- O(n+m) — sets se efficient

````python
a = [1,2,3,4,5,8]
b = [3,6,7,8]

i=j=0

while i<len(a) and j<len(b):
  if a[i] == b[j]:
    print(a[i])
    i+=1
    j+=1
  elif a[i] > b[j]:
    j+=1
  else:
    i+=1
````

## 📖 8. Subarray with given sum

```python
d = {0: -1}
curr_sum = 0
for i, x in enumerate(L):
    curr_sum += x
    if curr_sum - S in d:
        print(d[curr_sum - S] + 1, i)
    d[curr_sum] = i
```

- Prefix-sum map: agar curr-S pehle dikha to sum mila
- Sliding window ya prefix-sum

````python
# 8. Find continous subarray with a given sum(given non-negative numbers)
# return the starting and ending index of the subarray
# return 1st subarray in case of multiple

L = [1,22,13,7,9,11,10]
S = 16

for i in range(0,len(L)):
  subarray = []
  for j in range(i,len(L)):
    subarray.append(L[j])
    if sum(subarray) == S:
      print(subarray)
````

## 📖 9. Left smaller / right greater

```python
for i in range(1, len(L) - 1):
    if max(L[:i]) < L[i] < min(L[i+1:]):
        print(L[i])
```

- Check: left max < element < right min
- Precompute max_left/min_right arrays
- Naive slicing O(n²), precompute O(n)

````python
# 9. Find element with left side smaller/right side greater in an array

L = [3,1,2,5,8,7,9]

for i in range(1,len(L)-1):
  flag = True

  for j in range(0,i):
    if L[j] > L[i]:
      flag = False

  for k in range(i+1,len(L)):
    if L[k] < L[i]:
      flag = False

  if flag:
    print(L[i])

````

````python
L = [3,1,2,5,8,7,9]

for i in range(1,len(L)-1):
  if max(L[:i]) < L[i] < min(L[i+1:]):
    print(L[i])
````

````python
L = [3,1,2,5,8,7,9]

max_arr = []
min_arr = []

max_val = L[0]
min_val = L[-1]

for i in L:
  if i>max_val:
    max_val = i
  max_arr.append(max_val)

for i in range(len(L)-1,-1,-1):
  if L[i] < min_val:
    min_val = L[i]
  
  min_arr.insert(0,min_val)


for i in range(1,len(L)-1):
  if max_arr[i-1] < L[i] < min_arr[i+1]:
    print(L[i])
````

## 📖 10. Max subarray (Kadane)

```python
curr_sum = 0; best_sum = 0
for x in L:
    curr_sum = max(x, curr_sum + x)   # restart ya continue
    best_sum = max(best_sum, curr_sum)
```

- Negative hote hi nayi shuruaat
- Kadane's algorithm — O(n)
- Best sum tracking

````python
# 10. Maximum sum subarray
L = [-2,4,7,-1,6,-11,14,3,-1,-6]

d = {}

for i in range(0,len(L)):
  subarray = []
  for j in range(i,len(L)):
    subarray.append(L[j])
    d[sum(subarray)] = subarray

max_val = max(d.keys())

for i in d:
  if i == max_val:
    print(d[i])
````

````python

L = [-2,4,7,-1,6,-11,14,3,-1,-6]


curr_sum = 0
curr_seq = []
best_sum = L[0]
best_seq = []

for i in L:
  if i + curr_sum > i:
    curr_sum = curr_sum + i
    curr_seq.append(i)
  else:
    curr_sum = i
    curr_seq.clear()
    curr_seq.append(i)

  if curr_sum > best_sum:
    best_sum = curr_sum
    best_seq = curr_seq

print(best_sum,best_seq)


````
