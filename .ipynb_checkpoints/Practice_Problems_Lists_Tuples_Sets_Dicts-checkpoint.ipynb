# List Practice Problems

### Problem 1: Combine two lists index-wise

Write a program to add two lists index-wise. Create a new list that contains the 0th index item from both the list, then the 1st index item, and so on till the last element. Any leftover items will get added at the end of the new list.

**Given:**
```python
list1 = ["M", "na", "i", "Kh"]
list2 = ["y", "me", "s", "an"]
```

**Output:**
```python
[['M','y'], ['na', 'me'], ['i', 's'], ['Kh', 'an']]
```

````python
list1 = ["M", "na", "i", "Kh"]
list2 = ["y", "me", "s", "an"]

[[i,j] for (i,j) in zip(list1,list2)]
````

### Problem 5: Sum of elements greater than or equal to itself

You are given a list of integers. Make a list by adding all elements greater than and equal to itself.

**Input:**
```python
[2,4,6,10,1]
```

**Output:**
```python
[22,20,16,10,13]
```

For element `2` -> greater elements are `(4+6+10)` and `2` itself = `22`

````python
L = [2,4,6,10,1]
result = []

for i in L:
    sum = 0
    for j in L:
        if i <= j:
            sum = sum + j
    result.append(sum)

print(result)
````

### Problem 6: Find common unique items from two lists

Find list of common unique items from two lists and show in increasing order.

**Input:**
```python
num1 = [23,45,67,78,89,34]
num2 = [34,89,55,56,39,67]
```

**Output:**
```python
[34, 67, 89]
```

````python
num1 = [23,45,67,78,89,34,67]
num2 = [34,89,55,56,39,67,67]

common = []

for i in num1:
    if i in num2:
        if i not in common:
            common.append(i)

print(common)
````

### Problem 7: Sort alphanumeric strings by product of numeric characters

Sort a list of alphanumeric strings based on product value of numeric characters. If no numeric character, take product value as 1.

**Input:**
```python
['1ac21', '23fg', '456', '098d','1','kls']
```

**Output:**
```python
['456', '23fg', '1ac21', '1', 'kls', '098d']
```

````python
L = ['1ac21', '23fg', '456', '098d','1','kls']
prod_value = []

for item in L:
    product = 1
    for char in item:
        if char.isdigit():
            product = product * int(char)
    prod_value.append(product)

[i[1] for i in sorted(list(zip(prod_value,L)),reverse=True)]
````

### Problem 10: Add Space between Potential Words

**Input:**
```python
['campusxIs', 'bestFor', 'dataScientist']
```

**Output:**
```python
['campusx Is', 'best For', 'data Scientist']
```

````python
test_list = ["campusxIs", "bestFor", "dataScientist"]
res = []

for i in test_list:
    temp = [[]]
    for char in i:
        if char.isupper():
            temp.append([])
        temp[-1].append(char)
    
    temp_string = ""
    for item in temp:
        temp_string = temp_string + "".join(item) + " "
    res.append(temp_string[0:-1])

print(res)
````

### Problem 11: Union operation on 2 lists

**Input:**
```python
[1,2,3,4,5,1]
[2,3,5,7,8]
```

**Output:**
```python
[1,2,3,4,5,7,8]
```

````python
L1 = [1,2,3,4,5,1]
L2 = [2,3,5,7,8]

union = []

for i in L1:
    if i not in union:
        union.append(i)

for j in L2:
    if j not in union:
        union.append(j)

print(union)
````

# Tuple Practice Problems

### Q1: Join Tuples if similar initial element

While working with Python tuples, we can have a problem in which we need to perform concatenation of records from the similarity of initial element.

**Input:**
```python
test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]
```

**Output:**
```python
[(5, 6, 7, 8), (6, 10), (7, 13)]
```

````python
test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]

unique = []
for i in test_list:
    unique.append(i[0])
unique = set(unique)

result = []
for i in unique:
    result.append([i])
    for j in test_list:
        if j[0] == i:
            result[-1].append(j[1])

print(list(map(tuple,result)))
````

### Q2: Multiply Adjacent elements and take sum

Multiply adjacent elements (both sides) and take sum of right and left side multiplication result.

**Input:**
```python
(1, 5, 7, 8, 10)
```

**Output:**
```python
(5, 40, 91, 136, 80)
```

For element at index 1 (5): `5*1 + 5*7 = 5 + 35 = 40`

````python
t = (1, 5, 7, 8, 10)
L = []

L.append(t[0]*t[1])

for i in range(1, len(t)-1):
    L.append(t[i]*t[i-1] + t[i]*t[i+1])

L.append(t[-1]*t[-2])

print(tuple(L))
````

### Q3: Check if tuples are same or not

Two tuples would be same if both tuples have same element at same index.

**Input:**
```python
t1 = (1,2,3,0)
t2 = (0,1,2,3)
```

**Output:** `not same`

---

**Input:**
```python
t1 = (1,2,3,0)
t2 = (1,2,3,0)
```

**Output:** `same`

````python
t1 = (1,2,3,0)
t2 = (1,2,3,0)

flag = True
for i,j in zip(t1,t2):
    if i == j:
        continue
    else:
        flag = False
        break

if flag:
    print('same')
else:
    print('not same')
````

### Q4: Count tuples, lists and sets from a list

**Input:**
```python
list1 = [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi', 'bye'],['a', 'b']]
```

**Output:**
```
List-2
Set-2
Tuples-1
```

````python
L = [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi', 'bye'],['a', 'b']]
output = [0,0,0]

for i in L:
    if type(i) == list:
        output[0] = output[0] + 1
    elif type(i) == set:
        output[1] = output[1] + 1
    elif type(i) == tuple:
        output[2] = output[2] + 1

print('Lists-{}\nSets-{}\nTuples-{}'.format(output[0],output[1],output[2]))
````

### Q5: Shortlist Students for a Job role

Ask user to input students record and store in tuples. Then ask user to input three things: Primary Skill, Higher Education, Year of Graduation. Show every students record that matches all required criteria.

**Input:**
```
Enter No of records- 2
Enter Details of student-1
Enter Student name- Manohar
Enter Higher Education- B.Tech
Enter Primary Skill- Python
Enter Year of Graduation- 2022
Enter Details of student-2
Enter Student name- Ponian
Enter Higher Education- B.Sc.
Enter Primary Skill- C++
Enter Year of Graduation- 2020

Enter Job Role Requirement
Enter Skill- Python
Enter Higher Education- B.Tech
Enter Year of Graduation- 2022
```

**Output:**
```
('Manohar', 'B.Tech', 'Python', '2022')
```

````python
students = []
num = int(input('Enter the number of applicants: '))

for i in range(num):
    print('Enter details of', i+1, 'applicant:')
    name = input('Enter name: ')
    h_ed = input('Enter higher education: ')
    p_skill = input('Enter primary skill: ')
    yog = input('Enter year of graduation: ')
    students.append((name, h_ed, p_skill, yog))

required_skill = input('\nEnter required skill: ')
required_hed = input('Enter required higher education: ')
required_yog = input('Enter required year of graduation: ')

flag = False
for i in students:
    if i[1] == required_hed and i[2] == required_skill and i[3] == required_yog:
        print(i)
        flag = True

if flag == False:
    print('No such candidates')
````

# Set Practice Problems

### Q1: Find common elements in three lists using sets

**Input:**
```python
ar1 = [1, 5, 10, 20, 40, 80]
ar2 = [6, 7, 20, 80, 100]
ar3 = [3, 4, 15, 20, 30, 70, 80, 120]
```

**Output:**
```python
[80, 20]
```

````python
ar1 = [1, 5, 10, 20, 40, 80]
ar2 = [6, 7, 20, 80, 100]
ar3 = [3, 4, 15, 20, 30, 70, 80, 120]

s1 = set(ar1)
s2 = set(ar2)
s3 = set(ar3)

result = list((s1 & s2) & s3)
print(result)
````

### Q2: Count unique vowels using sets

Write a program to count unique number of vowels using sets. Lowercase and uppercase vowels will be taken as different.

**Input:**
```python
Str1 = "hands-on data science mentorship progrAm with live classes at affordable fee only on CampusX"
```

**Output:**
```
No of unique vowels-6
```

````python
vowels = set('aeiouAEIOU')

s = set("hands-on data science mentorship progrAm with live classes at affordable fee only on CampusX")

print('No of unique vowels-', len(s & vowels))
````

### Q3: Check if string is binary or not

A string is said to be binary if it consists of only two unique characters.

**Input:**
```
str = "01010101010" -> Yes
str = "1222211" -> Yes
str = "Campusx" -> No
```

````python
s = "01010101010"

if len(set(s)) == 2:
    print('binary')
else:
    print('not binary')
````

### Q4: Find union of n arrays

**Input:**
```python
[[1, 2, 2, 4, 3, 6],
 [5, 1, 3, 4],
 [9, 5, 7, 1],
 [2, 4, 1, 3]]
```

**Output:**
```python
[1, 2, 3, 4, 5, 6, 7, 9]
```

````python
L = [[1, 2, 2, 4, 3, 6],
     [5, 1, 3, 4],
     [9, 5, 7, 1],
     [2, 4, 1, 3]]

s = set()

for i in L:
    s.update(i)

print(s)
````

### Q5: Intersection of two lists using list comprehension

```python
lst2 = {9, 4, 5, 36, 47, 26, 10, 45, 87}
```

**Output:**
```python
[9, 10, 4, 5]
```

````python
lst1 = {15, 9, 10, 56, 23, 78, 5, 4, 9}
lst2 = {9, 4, 5, 36, 47, 26, 10, 45, 87}

[item for item in lst1 if item in lst2]
````

# Dictionary Practice Problems

### Q1: Key with maximum unique values

Given a dictionary with values list, extract key whose value has most unique values.

**Input:**
```python
test_dict = {"CampusX" : [5, 7, 7, 7, 7], "is" : [6, 7, 7, 7], "Best" : [9, 9, 6, 5, 5]}
```

**Output:**
```python
Best
```

````python
test_dict = {"CampusX" : [5, 7, 7, 7, 7], "is" : [6, 7, 7, 7], "Best" : [9, 9, 6, 5, 5]}

max_val = 0
max_key = ''

for i in test_dict:
    if max_val < len(set(test_dict[i])):
        max_val = len(set(test_dict[i]))
        max_key = i

print(max_key)
````

### Q2: Replace words from Dictionary

Given a string, replace its words from lookup dictionary.

**Input:**
```python
test_str = 'CampusX best for DS students.'
repl_dict = {"best" : "is the best channel", "DS" : "Data-Science"}
```

**Output:**
```python
CampusX is the best channel for Data-Science students.
```

````python
test_str = 'CampusX best for DS students.'
repl_dict = {"best" : "is the best channel", "DS" : "Data-Science"}

res = []
for i in test_str.split():
    if i in repl_dict:
        res.append(repl_dict[i])
    else:
        res.append(i)

print(" ".join(res))
````

### Q3: Convert List to List of dictionaries

Given list values and keys list, convert these values to key value pairs in form of list of dictionaries.

**Input:**
```python
test_list = ["CampusX", 10]
key_list = ["name", "id"]
```

**Output:**
```python
[{'name': 'CampusX', 'id': 10}]
```

````python
test_list = ["CampusX", 10]
key_list = ["name", "id"]

n = len(test_list)
res = []

for i in range(0, n, 2):
    res.append({key_list[0]: test_list[i], key_list[1]: test_list[i+1]})

print(res)
````

### Q4: Convert a list of Tuples into Dictionary

**Input:**
```python
L = [('A', 1), ('B', 2), ('C', 3)]
```

**Output:**
```python
{'A': [1], 'B': [2], 'C': [3]}
```

````python
L = [('A', 1), ('B', 2), ('C', 3)]
d = {}

for i,j in L:
    d[i] = [j]

print(d)
````

### Q5: Sort Dictionary key and values List

**Input:**
```python
{'c': [3], 'b': [12, 10], 'a': [19, 4]}
```

**Output:**
```python
{'a': [4, 19], 'b': [10, 12], 'c': [3]}
```

````python
d = {'c': [3], 'b': [12, 10], 'a': [19, 4]}
res = {}

for i in sorted(d):
    res[i] = sorted(d[i])

print(res)
````
