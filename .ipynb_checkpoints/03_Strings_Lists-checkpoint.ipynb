## 0.1 Warm-up Programs

### Program 1: Population Growth
The current population of a town is 10000. The population of the town is increasing at the rate of 10% per year. Find out the population at the end of each of the last 10 years.

````python
curr_pop = 10000
for i in range(10, 0, -1):
    print(i, curr_pop)
    curr_pop = curr_pop / 1.1
````

### Program 2: Sequence Sum
Calculate: `1/1! + 2/2! + 3/3! + ... + n/n!`

````python
n = int(input('Enter n: '))
result = 0
fact = 1
for i in range(1, n + 1):
    fact = fact * i
    result = result + i / fact
print(result)
````

## 1. Patterns

### Pattern 1: Right-Angled Triangle

````python
rows = int(input('Enter number of rows: '))
for i in range(1, rows + 1):
    for j in range(1, i + 1):
        print('*', end='')
    print()
````

### Pattern 2: Number Pyramid

````python
rows = int(input('Enter number of rows: '))
for i in range(1, rows + 1):
    for j in range(1, i + 1):
        print(j, end='')
    for k in range(i - 1, 0, -1):
        print(k, end='')
    print()
````

## 2. Loop Control Statements

| Statement | Description |
|-----------|-------------|
| `break` | Terminates the loop immediately |
| `continue` | Skips the current iteration, moves to next |
| `pass` | Does nothing, placeholder |

### Break Statement

````python
for i in range(1, 10):
    if i == 5:
        break
    print(i)
````

### Break with Nested Loop - Finding Prime Numbers

````python
lower = int(input('Enter lower range: '))
upper = int(input('Enter upper range: '))
for i in range(lower, upper + 1):
    for j in range(2, i):
        if i % j == 0:
            break
    else:
        print(i)
````

### Continue Statement

````python
for i in range(1, 10):
    if i == 5:
        continue
    print(i)
````

### Pass Statement

````python
for i in range(1, 10):
    pass
````

## 3. Strings - Creating, Accessing, Editing & Deleting

**Strings are sequences of characters.** In Python, strings are a sequence of Unicode characters.

### 3.1 Creating Strings

````python
s = 'hello'
s = "hello"
# multiline strings
s = '''hello'''
s = """hello"""
s = str('hello')
print(s)
````

````python
# Strings with quotes
print("it's raining outside")
print('it\'s raining outside')
````

### 3.2 Accessing Substrings from a String

````python
# Positive Indexing
s = 'hello world'
print(s[0])   # h
print(s[4])   # o
# print(s[41])  # IndexError - string index out of range
````

````python
# Negative Indexing
s = 'hello world'
print(s[-1])  # d
print(s[-3])  # r
````

````python
# Slicing
s = 'hello world'
print(s[6:0:-2])    # wol
print(s[::-1])      # dlrowolleh
print(s[-1:-6:-1])  # dlrow
````

### 3.3 Editing and Deleting in Strings

**Important:** Python strings are **immutable** - they cannot be changed after creation.

````python
# Strings are immutable - cannot edit
s = 'hello world'
# s[0] = 'H'  # TypeError: 'str' object does not support item assignment
````

````python
# Cannot delete parts of a string
s = 'hello world'
# del s  # This deletes the whole variable
# del s[-1:-5:2]  # TypeError: 'str' object does not support item deletion
````

## 4. Operations on Strings

| Category | Operations |
|----------|------------|
| **Arithmetic** | `+` (concatenation), `*` (repetition) |
| **Relational** | `==`, `!=`, `>`, `<`, `>=`, `<=` (lexicographic) |
| **Logical** | `and`, `or`, `not` |
| **Membership** | `in`, `not in` |

### Arithmetic Operations

````python
# Concatenation
print('delhi' + ' ' + 'mumbai')

# Repetition
print('delhi' * 5)
print('*' * 50)
````

### Relational Operations (Lexicographic Comparison)

````python
print('delhi' != 'delhi')    # False
print('mumbai' > 'pune')     # False (lexicographic)
print('Pune' > 'pune')       # False (uppercase < lowercase in ASCII)
````

### Logical Operations

````python
print('hello' and 'world')   # world
print('hello' or 'world')    # hello
print('' and 'world')        # (empty string is falsy)
print('' or 'world')         # world
print(not 'hello')           # False
````

### Loops on Strings

````python
for i in 'hello':
    print(i)
````

### Membership Operations

````python
print('D' in 'delhi')     # False (case-sensitive)
print('d' in 'delhi')     # True
print('x' not in 'delhi') # True
````

## 5. String Functions

### 5.1 len, max, min, sorted

````python
s = 'hello world'
print(len(s))                          # 11
print(max(s))                          # w
print(min(s))                          # (space)
print(sorted(s, reverse=True))         # ['w', 'r', 'o', 'o', 'l', 'l', 'l', 'h', 'e', 'd', ' ']
````

### 5.2 capitalize / title / upper / lower / swapcase

````python
s = 'hello world'
print(s.capitalize())    # Hello world (first letter uppercase)
print(s.title())         # Hello World (each word first letter uppercase)
print(s.upper())         # HELLO WORLD
print(s.lower())         # hello world
print('HeLlO WorLD'.swapcase())  # hElLo wORld
````

### 5.3 count / find / index

````python
s = 'my name is nitish'
print(s.count('i'))           # 3
print(s.find('x'))            # -1 (not found)
# s.index('x')               # ValueError - substring not found
````

### 5.4 endswith / startswith

````python
s = 'my name is nitish'
print(s.endswith('sh'))       # True
print(s.endswith('sho'))      # False
print(s.startswith('my'))     # True
print(s.startswith('1my'))    # False
````

### 5.5 format

````python
name = 'nitish'
gender = 'male'
'Hi my name is {1} and I am a {0}'.format(gender, name)
````

### 5.6 isalnum / isalpha / isdigit / isidentifier

````python
print('nitish1234%'.isalnum())    # False (has special char)
print('nitish'.isalpha())         # True (all alphabets)
print('123abc'.isdigit())         # False (has alphabets)
print('first-name'.isidentifier()) # False (has hyphen)
````

### 5.7 Split / Join

````python
# Split - string to list
print('hi my name is nitish'.split())  # ['hi', 'my', 'name', 'is', 'nitish']

# Join - list to string
print(' '.join(['hi', 'my', 'name', 'is', 'nitish']))  # 'hi my name is nitish'
````

### 5.8 Replace

````python
print('hi my name is nitish'.replace('nitish', 'campusx'))
````

### 5.9 Strip

````python
s = '   nitish   '
print(s.strip())    # 'nitish' (removes leading/trailing whitespace)
````

## 6. String Example Programs

### Example 1: Find length without len()

````python
s = input('Enter the string: ')
counter = 0
for i in s:
    counter += 1
print('Length of string is', counter)
````

### Example 2: Extract username from email

````python
s = input('Enter the email: ')
pos = s.index('@')
print(s[0:pos])
````

### Example 3: Count frequency of a character

````python
s = input('Enter the string: ')
term = input('What would you like to search for: ')
counter = 0
for i in s:
    if i == term:
        counter += 1
print('Frequency:', counter)
````

### Example 4: Remove a particular character from string

````python
s = input('Enter the string: ')
term = input('What would you like to remove: ')
result = ''
for i in s:
    if i != term:
        result = result + i
print(result)
````

### Example 5: Check Palindrome

````python
s = input('Enter the string: ')
flag = True
for i in range(0, len(s) // 2):
    if s[i] != s[len(s) - i - 1]:
        flag = False
        print('Not a Palindrome')
        break
if flag:
    print('Palindrome')
````

### Example 6: Count words without split()

````python
s = input('Enter the string: ')
L = []
temp = ''
for i in s:
    if i != ' ':
        temp = temp + i
    else:
        L.append(temp)
        temp = ''
L.append(temp)
print(L)
````

### Example 7: Convert to title case without title()

````python
s = input('Enter the string: ')
L = []
for i in s.split():
    L.append(i[0].upper() + i[1:].lower())
print(' '.join(L))
````

### Example 8: Convert integer to string

````python
number = int(input('Enter the number: '))
digits = '0123456789'
result = ''
while number != 0:
    result = digits[number % 10] + result
    number = number // 10
print(result)
print(type(result))
````
