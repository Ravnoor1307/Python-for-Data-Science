### Python Output (print function)

The `print()` function is used to display output on the screen. It is the most basic and frequently used function in Python.

**Syntax:**

```python
print(*objects, sep=' ', end='\n', file=sys.stdout, flush=False)
```

**Parameters:**

| Parameter | Description | Default |
|-----------|-------------|--------|
| `*objects` | One or more values to print (any data type) | - |
| `sep` | Separator printed **between** values | `' '` (single space) |
| `end` | Character printed **after** all values | `'\n'` (newline) |
| `file` | Where to write the output | `sys.stdout` (screen) |
| `flush` | Whether to flush the output stream | `False` |

> **Important:** Python is a **case-sensitive** language. `print()` works, but `Print()` or `PRINT()` will throw an error.

**Output:** `hello-world`

The `end='-'` replaced the newline with a dash, so `print('world')` continued on the same line.

````python
# end='' with no character - prints on same line with no gap
print('Loading', end='...')
print('Done!')
````

# Topic 2 - Python Data Types

Every value in Python has a **data type**. Understanding data types is fundamental because:
- Different types support different operations
- Data is stored differently in memory
- Mixing incompatible types causes errors

### Python's Built-in Data Types:

| Category | Type | Example | Mutable? |
|----------|------|---------|----------|
| **Numeric** | `int` | `8`, `100`, `-5` | No |
| | `float` | `8.55`, `3.14` | No |
| | `complex` | `5+6j` | No |
| **Boolean** | `bool` | `True`, `False` | No |
| **Text** | `str` | `'Hello'`, `"World"` | No |
| **Sequence** | `list` | `[1, 2, 3]` | Yes |
| | `tuple` | `(1, 2, 3)` | No |
| **Set** | `set` | `{1, 2, 3}` | Yes |
| **Mapping** | `dict` | `{'name': 'Nitish'}` | Yes |

> **Mutable** = can be changed after creation
> **Immutable** = cannot be changed after creation

### 2.1 Integer (`int`)

Whole numbers without any decimal point. Python supports **arbitrarily large** integers - there is no upper limit, unlike C/C++ or Java.

````python
# Integer
print(8)
# 1*10^308
print(1e309)
````

**Output:**
```
8
inf
```

`1e309` means `1 x 10^309`. When an integer exceeds Python's float range, it becomes `inf` (infinity).

### 2.2 Float (Decimal Numbers)

Numbers with decimal points. Stored using IEEE 754 double-precision format.

````python
# Decimal/Float
print(8.55)
print(1.7e309)
````

**Output:**
```
8.55
inf
```

`1.7e309` exceeds the maximum float value, so Python returns `inf`.

### 2.3 Boolean (`bool`)

Booleans represent one of two values: `True` or `False`. In Python, `bool` is actually a **subclass of `int`**:
- `True` has the value `1`
- `False` has the value `0`

````python
# Boolean
print(True)
print(False)
````

````python
# Booleans are subclass of int!
print(True + 1)   # 2 (True = 1)
print(False + 1)  # 1 (False = 0)
print(True * 5)   # 5
print(False * 5)  # 0
````

### 2.4 String (`str`)

Strings are sequences of characters enclosed in quotes. They are **immutable** - once created, you cannot change individual characters.

````python
# Text/String
print('Hello World')
````

````python
# Single and double quotes are equivalent
print('Hello')
print("Hello")
````

### 2.5 Complex Numbers

Complex numbers have a real and imaginary part. Python uses `j` (not `i` as in mathematics) for the imaginary unit.

````python
# complex
print(5+6j)
````

````python
# Access real and imaginary parts
c = 5+6j
print(c.real)   # 5.0
print(c.imag)   # 6.0
print(type(c))  # <class 'complex'>
````

### 2.6 List

Lists are **ordered, mutable** collections that can hold mixed data types. Think of them as flexible arrays (compared to fixed-type arrays in C).

````python
# List-> C-> Array
print([1,2,3,4,5])
````

````python
# Lists can hold mixed types!
print([1, 'hello', 3.14, True])
````

### 2.7 Tuple

Tuples are **ordered, immutable** collections. Once created, they cannot be modified. Use parentheses `( )`.

````python
# Tuple
print((1,2,3,4,5))
````

### 2.8 Set

Sets are **unordered** collections of **unique** elements. Duplicates are automatically removed.

````python
# Sets
print({1,2,3,4,5})
````

````python
# Duplicates are removed automatically
print({1,1,2,2,3,3})  # {1, 2, 3}
````

### 2.9 Dictionary (`dict`)

Dictionaries store data as **key-value pairs**. Like hash maps in other languages. Keys must be unique and immutable.

````python
# Dictionary
print({'name':'Nitish','gender':'Male','weight':70})
````

### 2.10 Using `type()` to Check Data Types

The `type()` function returns the data type of any value. Essential for debugging.

````python
# type
type([1,2,3])
````

````python
# Check type of every data type
print(type(8))         # <class 'int'>
print(type(8.55))      # <class 'float'>
print(type(True))      # <class 'bool'>
print(type('hello'))   # <class 'str'>
print(type(5+6j))      # <class 'complex'>
print(type([1,2]))     # <class 'list'>
print(type((1,2)))     # <class 'tuple'>
print(type({1,2}))     # <class 'set'>
print(type({'a':1}))   # <class 'dict'>
````

# Topic 3 - Variables

Variables are **named containers** that store data. Think of them as labeled boxes where you put values.

### 3.1 Static vs Dynamic Typing

| Feature | Static Typing (C/Java) | Dynamic Typing (Python) |
|---------|----------------------|------------------------|
| Type declaration | Required: `int a = 5` | Not needed: `a = 5` |
| Type flexibility | Fixed once declared | Can change anytime |
| Type errors | Caught at compile time | Caught at runtime |

````python
# Static Vs Dynamic Typing
# Static Vs Dynamic Binding
# stylish declaration techniques
````

````python
# Basic variable assignment
name = 'nitish'
print(name)

a = 5
b = 6
print(a + b)
````

**Output:**
```
nitish
11
```

In Python, no type declaration needed. Just use `=` to assign.

````python
# Dynamic Typing - Python infers the type automatically
a = 5
# Static Typing - In C/Java you would write: int a = 5
# Python doesn't need this!
````

### 3.2 Dynamic Binding

In Python, a variable can **change its type** during execution. This is called **dynamic binding**.

````python
# Dynamic Binding - variable changes type at runtime
a = 5          # a is an integer
print(a)       # 5
a = 'nitish'   # now a is a string!
print(a)       # nitish

# In C/Java (Static Binding), this would cause a compile-time error
# because 'a' was declared as int and can't become a string
````

### Interview Question: What is Dynamic Typing in Python?

**Answer:** Dynamic typing means:
1. You don't need to declare variable types before using them
2. A variable can hold any type of data
3. The type can change during program execution
4. Type checking happens at runtime, not compile time

```python
# Python (Dynamic Typing)
x = 5        # x is int
x = 'hello'  # x is now str - perfectly valid!

// Java (Static Typing)
int x = 5;
x = 'hello'; // ERROR! Can't assign string to int
```

### 3.3 Variable Assignment Techniques

````python
# Standard assignment - one variable at a time
a = 1
b = 2
c = 3
print(a,b,c)
````

````python
# Multiple assignment (tuple unpacking) - assign multiple variables in one line
a,b,c = 1,2,3
print(a,b,c)
````

````python
# Same value to multiple variables
a=b=c= 5
print(a,b,c)
````

### 3.4 Variable Naming Rules

| Rule | Valid | Invalid |
|------|-------|---------|
| Must start with letter or `_` | `_name`, `age` | `2name`, `my-name` |
| Can contain letters, digits, `_` | `var_2`, `MAX` | `my var`, `my-var` |
| Case-sensitive | `Name` != `name` | - |
| Cannot be a keyword | `my_class` | `class`, `if`, `for` |

**Best Practices:**
- Use `snake_case` for variables: `student_name`
- Use descriptive names: `age` not `a`

````python
# Good variable names (descriptive)
student_name = 'Nitish'
student_age = 25
course_fee = 9999.99

print(student_name)
print(student_age)
print(course_fee)
````

# Topic 4 - Comments in Python

Comments are notes in your code that are **ignored by Python**. They help you and others understand what the code does.

### Types of Comments:
- **Single-line:** Starts with `#` - everything after `#` on that line is ignored
- **Multi-line:** Use triple quotes `"""..."""` (technically docstrings)

> **Tip:** Write comments that explain *why*, not *what*. The code already shows *what*.

````python
# this is a comment
# second line
a = 4
b = 6 # like this
# second comment
print(a+b)
````

**Output:** `10`

Everything after `#` is ignored. Comments don't affect the output.

````python
# Multi-line comment using triple quotes
"""
This is a multi-line comment.
It can span multiple lines.
Useful for longer explanations.
"""
print('Comments do not affect output')
````

````python
# Commenting out code (temporarily disable code)
# print('This line will not execute')
# print('Neither will this')
print('Only this runs')
````

### 5.2 Identifiers

Identifiers are **names** given to variables, functions, classes, etc.

**Rules for Identifiers:**
1. Must start with a letter (`a-z`, `A-Z`) or underscore (`_`)
2. Can contain letters, digits, and underscores
3. **Cannot** start with a digit
4. **Cannot** be a Python keyword
5. Are case-sensitive (`name` and `Name` are different)

````python
# Identifiers
# You can't start with a digit
name1 = 'Nitish'
print(name1)
# You can use special chars -> _
_ = 'ntiish'
print(_)
# identiers can not be keyword
````

````python
# More valid identifiers
student_name = 'Nitish'    # snake_case (recommended)
MAX_SIZE = 100             # UPPER_CASE for constants
_private = 'hidden'        # leading underscore

print(student_name)
print(MAX_SIZE)
print(_private)
````

# Topic 6 - User Input

The `input()` function gets data from the user at runtime.

### Key Facts:
- `input()` **always returns a string** (`str`)
- To get numbers, you must convert using `int()` or `float()`
- Syntax: `variable = input('prompt message')`

> **Static vs Dynamic Input:** In static languages like C, you must declare the type before reading input. In Python, `input()` always returns a string.

````python
# Basic input - always returns a string
name = input('Enter your name: ')
print(f'Hello, {name}!')
````

````python
# input() returns a string - even if user types a number!
age = input('Enter your age: ')
print(type(age))  # <class 'str'> - NOT int!
````

````python
# To do math with input, you must convert to int/float
fnum = int(input('enter first number'))
snum = int(input('enter second number'))
#print(type(fnum),type(snum))
# add the 2 variables
result = fnum + snum
# print the result
print(result)
print(type(fnum))
````

**Output:**
```
enter first number56
enter second number67
123
<class 'int'>
```

Without `int()`, Python would concatenate strings: `'56' + '67' = '5667'` instead of adding: `56 + 67 = 123`.

````python
# Practical example: Temperature converter
fahrenheit = float(input('Enter temperature in Fahrenheit: '))
celsius = (fahrenheit - 32) * 5/9
print(f'{fahrenheit}F = {celsius:.2f}C')
````

# Topic 7 - Type Conversion (Type Casting)

Type conversion means changing a value from one data type to another.

### 7.1 Implicit Type Conversion
Python **automatically** converts one type to another when it's safe. This is done by Python itself.

### 7.2 Explicit Type Conversion
You **manually** convert types using built-in functions like `int()`, `float()`, `str()`, etc.

````python
# Implicit Vs Explicit
print(5+5.6)
print(type(5),type(5.6))
````

**Output:**
```
10.6
<class 'int'> <class 'float'>
```

Python automatically converted `5` (int) to `5.0` (float) before adding. The result is `10.6` (float).

````python
# This will cause an error - can't implicitly convert string to int
# print(4 + '4')  # TypeError!
print('Use explicit conversion: 4 + int("4") =', 4 + int('4'))
````

### 7.3 Explicit Type Conversion

````python
# Explicit
# str -> int
#int(4+5j)  # TypeError! Can't convert complex to int

# int to str
print(str(5))          # '5'
print(type(str(5)))    # <class 'str'>

# float
print(float(4))        # 4.0
print(type(float(4)))  # <class 'float'>
````

````python
# More explicit conversion examples

# str to int
print(int('10'))       # 10

# float to int (truncates, does NOT round!)
print(int(9.99))       # 9
print(int(9.5))        # 9
print(int(-9.99))      # -9

# str to float
print(float('3.14'))   # 3.14

# int/float to bool
print(bool(0))         # False
print(bool(1))         # True
print(bool(''))        # False (empty string)
print(bool('hello'))   # True (non-empty string)
````

### Type Conversion Summary

| Conversion | Function | Example | Result |
|-----------|----------|---------|--------|
| int to float | `float()` | `float(5)` | `5.0` |
| float to int | `int()` | `int(9.99)` | `9` (truncates!) |
| int to str | `str()` | `str(5)` | `'5'` |
| str to int | `int()` | `int('10')` | `10` |
| str to float | `float()` | `float('3.14')` | `3.14` |

**Common Errors:**
- `int('hello')` -> `ValueError` (not a valid number)
- `int('3.14')` -> `ValueError` (use `float('3.14')` instead)
- `int(5+6j)` -> `TypeError` (can't convert complex to int)

# Topic 8 - Literals

A literal is a **fixed value** written directly in code. Unlike variables, literals never change.

### Types of Literals:
1. **Numeric:** Integer, Float, Complex
2. **String:** Single, double, multi-line, raw, unicode
3. **Boolean:** True, False
4. **Special:** None

### 8.1 Numeric Literals

````python
a = 0b1010 #Binary Literals
b = 100 #Decimal Literal 
c = 0o310 #Octal Literal
d = 0x12c #Hexadecimal Literal

#Float Literal
float_1 = 10.5 
float_2 = 1.5e2 # 1.5 * 10^2
float_3 = 1.5e-3 # 1.5 * 10^-3

#Complex Literal 
x = 3.14j

print(a, b, c, d)
print(float_1, float_2,float_3)
print(x, x.imag, x.real)
````

**Output:**
```
10 100 200 300
10.5 150.0 0.0015
3.14j 3.14 0.0
```

**Number System Prefixes:**
| Prefix | System | Example | Decimal |
|--------|--------|---------|--------|
| `0b` | Binary | `0b1010` | 10 |
| (none) | Decimal | `100` | 100 |
| `0o` | Octal | `0o310` | 200 |
| `0x` | Hexadecimal | `0x12c` | 300 |

**Scientific Notation:**
- `1.5e2` = `1.5 x 10^2` = `150.0`
- `1.5e-3` = `1.5 x 10^-3` = `0.0015`

````python
# binary
x = 3.14j
print(x.imag)
````

**Output:** `3.14` - Access the imaginary part of a complex number.

### 8.2 String Literals

````python
string = 'This is Python'
strings = "This is Python"
char = "C"
multiline_str = """This is a multiline string with more than one line code."""
unicode = u"\U0001f600\U0001F606\U0001F923"
raw_str = r"raw \n string"

print(string)
print(strings)
print(char)
print(multiline_str)
print(unicode)
print(raw_str)
````

**Output:**
```
This is Python
This is Python
C
This is a multiline string with more than one line code.
(emoji characters)
raw \n string
```

**String Types Explained:**

| Type | Description | Example |
|------|-------------|--------|
| Single quotes | Same as double quotes | `'Hello'` |
| Double quotes | Same as single quotes | `"Hello"` |
| Multi-line `""""""` | Preserves newlines | `"""Line1\nLine2"""` |
| Raw `r""` | Backslashes are literal | `r"raw \n string"` -> prints `raw \n string` |
| Unicode `u""` | Special characters | `u"\U0001f600"` (emoji) |

### 8.3 Boolean Literals

````python
a = True + 4
b = False + 10

print("a:", a)
print("b:", b)
````

**Output:**
```
a: 5
b: 10
```

Since `True = 1` and `False = 0`:
- `True + 4` = `1 + 4` = `5`
- `False + 10` = `0 + 10` = `10`

### 8.4 The `None` Literal

`None` represents the **absence of a value**. It is Python's way of saying "nothing here".

- It is of type `NoneType`
- Always use `is None` (not `== None`) to check for it
- `None` is different from `0`, `''` (empty string), or `False`

````python
k = None
a = 5
b = 6
print('Program exe')
````

````python
# None is different from other falsy values
print(None is None)      # True (preferred way to check)
print(type(None))        # <class 'NoneType'>

print(None == 0)         # False
print(None == '')        # False
print(None == False)     # False

print(0 is None)         # False
print('' is None)        # False
print(False is None)     # False
````

### Problem 1: Simple Calculator
Write a program that takes two numbers and an operator (+, -, *, /) as input and prints the result.

````python
# Solution
num1 = float(input('Enter first number: '))
operator = input('Enter operator (+, -, *, /): ')
num2 = float(input('Enter second number: '))

if operator == '+':
    result = num1 + num2
elif operator == '-':
    result = num1 - num2
elif operator == '*':
    result = num1 * num2
elif operator == '/':
    if num2 != 0:
        result = num1 / num2
    else:
        result = 'Error: Division by zero!'
else:
    result = 'Error: Invalid operator!'

print(f'{num1} {operator} {num2} = {result}')
````

### Problem 2: Leap Year Check
Write a program to check if a year is a leap year.
- Divisible by 4 AND not a century year, OR
- Divisible by 400

````python
# Solution
year = int(input('Enter a year: '))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f'{year} is a LEAP year')
else:
    print(f'{year} is NOT a leap year')
````

### Problem 3: Type Identifier
Write a program that takes any input and identifies its most likely data type.

````python
# Solution
user_input = input('Enter anything: ')

try:
    if '.' in user_input:
        value = float(user_input)
        print(f'Type: float, Value: {value}')
    else:
        value = int(user_input)
        print(f'Type: int, Value: {value}')
except ValueError:
    if user_input.lower() in ['true', 'false']:
        print(f'Type: bool, Value: {user_input}')
    elif user_input == 'None':
        print(f'Type: NoneType, Value: None')
    else:
        print(f'Type: str, Value: {user_input}')
````
