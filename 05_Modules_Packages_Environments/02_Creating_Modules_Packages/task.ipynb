## ⭐⭐ Task 4: Grade-Utils Toolbelt 📚

Instantiate `grade_utils.py` in same tempbow (tempdir) with 3 fns:
- `average(lst)` round-1
- `grade_letter(s)` (90:A, 75:B, 60:C, 40:D, less:F)
- `topper(d)` max-by-value!
Import & check: average([50, 60, 70])→60.0, grade_letter(88)→B,
topper({"Sana": 92, "Amit": 85, "Priya": 96})→Priya!

💡 **Hint:** classdocstring lagana mat bhoolna! 📖

````python
# TODO
````

<details><summary>✅ Solution</summary>

60.0 / B / Priya — toolbelt ready! 📚
</details>

````python
# ✅ SOLUTION — Task 4
gr_utils = '''"""grade_utils — exams-ka helper-module! 📚"""

def average(marks):
    """List-of-marks ka round-1 avg!"""
    return round(sum(marks) / len(marks), 1)

def grade_letter(score):
    """Score → exam-card letter!"""
    if score >= 90: return "A"
    if score >= 75: return "B"
    if score >= 60: return "C"
    if score >= 40: return "D"
    return "F"

def topper(students):
    """{'name': marks} me best!"""
    return max(students, key=students.get)
'''

with open(os.path.join(tmp, "grade_utils.py"), "w") as f:
    f.write(gr_utils)

importlib.invalidate_caches()         # 🔄 fresh-file dikhe!
import grade_utils as gu
print("avg   :", gu.average([50, 60, 70]))
print("grade :", gu.grade_letter(88))
print("topper:", gu.topper({"Sana": 92, "Amit": 85, "Priya": 96}))
print("doc   :", gu.__doc__)
# Expected OUTPUT:
# avg   : 60.0
# grade : B
# topper: Priya
# doc   : grade_utils — exams-ka helper-module! 📚
````

````python
# ✅ SOLUTION — Task 7 (Kotini-Toolkit BOSS 📋🔧)
BASE2 = tempfile.mkdtemp()
PKG2 = os.path.join(BASE2, "kotini_toolkit")
os.mkdir(PKG2)

INIT2 = '''"""kotini_toolkit — finance-team ka audit-pack! 📋🔧"""
print("kotini-toolkit banner: audit-ready! 🛡️")

from .audits import tax, savings_rate
from .reports import card
'''

AUDITS2 = '''"""audits — tax + savings helpers! 💰"""

def tax(amount, rate=0.05):
    """percent-rate ka tax round-2!"""
    return round(amount * rate, 2)

def savings_rate(savings, income):
    """income-share as-% round-1!"""
    return round(savings / income * 100, 1)
'''

REPORTS2 = '''"""reports — audit-card print-public! 📋"""

def card(name, income, savings, tax_amt):
    """4-line sync-align audit-card!"""
    lines = [
        "┌─ KOTINI-AUDIT-CARD ────────────────┐",
        f"│ name   : {name}",
        f"│ income : ₹{income:,}",
        f"│ savings: ₹{savings:,} ({round(savings / income * 100, 1)}%)",
        f"│ tax    : ₹{tax_amt:,}",
        "└─────────────────────────────────────┘",
    ]
    return "\\n".join(lines)
'''

with open(os.path.join(PKG2, "__init__.py"), "w") as f:
    f.write(INIT2)
with open(os.path.join(PKG2, "audits.py"), "w") as f:
    f.write(AUDITS2)
with open(os.path.join(PKG2, "reports.py"), "w") as f:
    f.write(REPORTS2)

sys.path.insert(0, BASE2)
import kotini_toolkit as kt

print("tax           :", kt.tax(100000))
print("savings-rate  :", kt.savings_rate(25000, 100000))
print("card-print 👇")
print(kt.card("Priya Sharma", 100000, 25000, 5000.0))
print("package-toolkit complete — import = ready-army! 🛡️")
# Expected OUTPUT:
# kotini-toolkit banner: audit-ready! 🛡️
# tax           : 5000.0
# savings-rate  : 25.0
# card-print 👇
# ┌─ KOTINI-AUDIT-CARD ────────────────┐
# │ name   : Priya Sharma
# │ income : ₹100,000
# │ savings: ₹25,000 (25.0%)
# │ tax    : ₹5,000.0
# └─────────────────────────────────────┘
# package-toolkit complete — import = ready-army! 🛡️
````
