## ⭐⭐⭐ Task 6: Shadow-Trap Proof (tempdir-yaar!) 💀🪤

Most-memorable import-trap LIVE-pirof:
1. tempdir banao → andar fake `math.py` likho (with `BANNER = "LOCAL-FAKE-MATH 💀"`)
2. `sys.path.insert(0, tmpdir)` → `import math` → `math.BANNER` print (FAKE mila!)
3. Cleanup: sys.path.pop(0) + `del sys.modules["math"]` → phir `import math`
   → ab REAL stdlib! `math.sqrt(9)` = 3.0 print proof!

💡 **Hints:**
- `tempfile.TemporaryDirectory()` + open-write for the fake-file!
- Cleanup-ORDER farzi: path-pop → modules-del → re-import!

````python
# TODO
````

<details><summary>✅ Solution</summary>

FAKE-math pakda (LOCAL-FAKE-MATH 💀!) → cleanup → REAL sqrt=3.0!
Aage-se-friend: file-naam shadow kabhi-nahi! 🪤✌️
</details>

````python
# ✅ SOLUTION — Task 6
import sys, os, tempfile

with tempfile.TemporaryDirectory() as tmp:
    with open(os.path.join(tmp, "math.py"), "w") as f:
        f.write('BANNER = "LOCAL-FAKE-MATH 💀"\n')

    sys.modules.pop("math", None)                    # cached-real hatao (pehle-cells ka!) 🧹
    sys.path.insert(0, tmp)
    import math                                        # 💀 abhi FAKE pakda!
    print("shadow-math :", math.BANNER)

    sys.path.pop(0)                                    # cleanup-step-1: path
    del sys.modules["math"]                            # cleanup-step-2: cache!
    import math                                        # 🛡️ ab REAL stdlib!
    print("real-sqrt9 :", math.sqrt(9))
    print("lesson      : apni file kabhi stdlib-naam mat rakho! 🪤")
# Expected OUTPUT:
# shadow-math : LOCAL-FAKE-MATH 💀
# real-sqrt9 : 3.0
# lesson      : apni file kabhi stdlib-naam mat rakho! 🪤
````

````python
# ✅ SOLUTION — Task 7 (MCQ Lucky-Paper BOSS 🎓🎲)
import random, math, datetime as dt

qbank = {
    1: "Python me 2^10 ka quick-value kya hai?",
    2: "'list vs tuple' ka key-difference ek-line me?",
    3: "len('python') ka output kya hoga?",
    4: "list.append() vs list.extend() — kaun-sa kab?",
    5: "dict.get(key, default) ka fayda kya?",
    6: "for-loop me else-part kab chalta hai?",
    7: "*args aur **kwargs me farq batao?",
    8: "append-kyun-not: list1 = list2 ka trap batao!",
    9: "range() lazy kyun hota hai — ek-line!",
    10: "__name__ == '__main__' ka matlab kya?",
}

SEED_NUM = 7
random.seed(SEED_NUM)
picked = random.sample(range(1, 11), 5)

paper_date = dt.date(2026, 9, 17)
print("═" * 32)
print(f"═══ LUCKY-PAPER 📄 (seed-{SEED_NUM}) ═══")
print(f"date: {paper_date} ({paper_date.strftime('%A')})")
print(f"total-possible-sets: {math.comb(10, 5)} (10C5)")
print("─" * 32)
for i, qnum in enumerate(picked, 1):
    print(f"Q{i}. [bank#{qnum}] {qbank[qnum]}")
print("─" * 32)

random.seed(SEED_NUM)                                # reseed-replay!
proof = random.sample(range(1, 11), 5)
print("reseed-proof :", proof == picked, "(deterministic-wall 🧱)")
print("good-luck! (reseed → new-set 🎲)")
# Expected OUTPUT:
# ════════════════════════════════
# ═══ LUCKY-PAPER 📄 (seed-7) ═══
# date: 2026-09-17 (Thursday)
# total-possible-sets: 252 (10C5)
# ────────────────────────────────
# Q1. [bank#6] for-loop me else-part kab chalta hai?
# Q2. [bank#3] len('python') ka output kya hoga?
# Q3. [bank#7] *args aur **kwargs me farq batao?
# Q4. [bank#10] __name__ == '__main__' ka matlab kya?
# Q5. [bank#1] Python me 2^10 ka quick-value kya hai?
# ────────────────────────────────
# reseed-proof : True (deterministic-wall 🧱)
# good-luck! (reseed → new-set 🎲)
````
