## 🔬 Deep Dive : read_csv → Series, copy & views

```python
subs = pd.read_csv('/content/subs.csv', squeeze=True)   # 1 col → Series
```

- read exact data w/ meaningful index
- **copy vs view**:
```python
new = vk.head()          # VIEW — same underlying data!
new[1] = 1               # original vk bhi badal jayega!
new = vk.head().copy()   # separate copy — safe
new[1] = 100             # original safe
```
- Always `.copy()` when you plan to modify subset
- Chained assignment trap → SettingWithCopyWarning
- pandas copies vs views classic interview
- `.loc`/`.iloc` similar semantics
- Read first, verify, then structure


- `drop_duplicates` — keep param: last/first/False
