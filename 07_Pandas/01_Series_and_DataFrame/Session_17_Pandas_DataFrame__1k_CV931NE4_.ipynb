### Math Methods

## 📖 Math methods — axis

```python
students.sum(axis=0)   # column-wise (default)
students.mean(axis=1)  # row-wise
students.var()         # column variance
```

- `axis=0` — columns me (down)
- `axis=1` — rows me (across)
- Series me axis n/a; DataFrame me chunna padega
- `sum/mean/min/max/var/std` sab same axis rules

## 🔬 Deep Dive : Math methods — axis

```python
students.sum(axis=0)      # column-wise sums (each col total)
students.mean(axis=1)     # row-wise averages
students.var()            # variance per column
```

- **axis=0** → vertical (over ROWS) — per column result
- **axis=1** → horizontal (over COLUMNS) — per row result
- Memory trick: axis=0 drops rows (collapses rows)
- axis=1 drops columns (collapses columns)
- Default axis=0 — column stats
- result: Series indexed by columns
- numeric columns auto-selected for math
- NaN exclusion (skipna default)
- SQL-like column aggregates (sum col)
- EDA: sum/mean/var each feature

### Selecting rows from a DataFrame

- **iloc** - searches using index positions
- **loc** - searches using index labels 

## 📖 Selecting rows — iloc & loc

```python
movies.iloc[5]                  # 5th row (position)
movies.iloc[:5]                 # rows 0-4
movies.iloc[[0,4,5]]            # fancy positions
students.loc['nitish']          # label access
students.loc['nitish':'rishabh':2]
```

- `iloc` — **position** se (integer)
- `loc` — **label** se (index name)
- Slice `iloc[:5]` exclusion hi; `loc['a':'c']` inclusive
- Fancy lists dono me kaam

## 🔬 Deep Dive : iloc & loc — row access

```python
movies.iloc[5]             # row 5 by POSITION → Series
movies.iloc[:5]            # first 5 rows by slice
movies.iloc[[0,4,5]]       # fancy positions
students.loc['nitish']                    # row by INDEX LABEL
students.loc['nitish':'rishabh':2]        # label slice incl BOTH ends
students.loc[['nitish','ankita','rupesh']]  # label list
```

- **iloc** — integer POSITION (like numpy index)
- **loc** — INDEX LABEL (by name)
- Single → Series; slice/fancy → DataFrame
- loc slice is INCLUSIVE of stop (unlike iloc!)
- Position: 0-based | label: actual index values
- Use positional when index unknown; label when meaningful
- Never mix: chained call ambiguity
- lead tools: reading subset rows
- `.at`/`.iat` single fast access
- String index labels natural dictionary-like

- Optimization: read CSV headings columns ahead
