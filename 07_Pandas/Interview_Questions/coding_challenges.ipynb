# Pandas - coding challenges

10 hands-on problems. Each solution cell ends with an assertion so you can check yourself.

## Setup

```python
import numpy as np
import pandas as pd

pd.set_option('display.width', 120)
pd.set_option('display.max_columns', 30)
np.random.seed(0)
print('pandas', pd.__version__)
```

## Challenge 1

Read a CSV and report shape, dtypes and null counts.

```python
df = pd.DataFrame({'a': [1, None], 'b': ['x', 'y']})
assert df.shape == (2, 2)
print(df.dtypes.to_dict(), df.isna().sum().to_dict())
```

## Challenge 2

Find the top three rows by a column.

```python
df = pd.DataFrame({'v': [5, 9, 1, 7]})
top = df.nlargest(3, 'v')
assert top.iloc[0]['v'] == 9
print(top)
```

## Challenge 3

Group by one column and compute two aggregations.

```python
df = pd.DataFrame({'g': ['a', 'a', 'b'], 'v': [1, 3, 5]})
out = df.groupby('g').agg(n=('v', 'size'), total=('v', 'sum'))
assert out.loc['a', 'total'] == 4
print(out)
```

## Challenge 4

Fill missing values per group with the group mean.

```python
df = pd.DataFrame({'g': ['a', 'a', 'b'], 'v': [1.0, None, 4.0]})
df['v'] = df.groupby('g')['v'].transform(lambda s: s.fillna(s.mean()))
assert df.loc[1, 'v'] == 1.0
print(df)
```

## Challenge 5

Merge two frames and count unmatched rows.

```python
a = pd.DataFrame({'k': [1, 2, 3]})
b = pd.DataFrame({'k': [2], 'v': [9]})
m = a.merge(b, on='k', how='left', indicator=True)
assert (m['_merge'] == 'left_only').sum() == 2
print(m)
```

## Challenge 6

Create a pivot table with margins.

```python
df = pd.DataFrame({'r': ['x', 'y'], 'c': ['p', 'p'], 'v': [1, 2]})
pt = df.pivot_table(index='r', columns='c', values='v',
                    aggfunc='sum', margins=True)
assert pt.loc['All', 'All'] == 3
print(pt)
```

## Challenge 7

Compute a 3-period rolling mean.

```python
s = pd.Series([1, 2, 3, 4, 5], dtype=float)
r = s.rolling(3).mean()
assert r.iloc[2] == 2.0
print(r.tolist())
```

## Challenge 8

Rank rows within each group.

```python
df = pd.DataFrame({'g': ['a', 'a', 'b'], 'v': [1, 9, 4]})
df['r'] = df.groupby('g')['v'].rank(ascending=False)
assert df.loc[1, 'r'] == 1.0
print(df)
```

## Challenge 9

Convert a wide frame to long format.

```python
wide = pd.DataFrame({'id': [1], 'a': [10], 'b': [20]})
long = wide.melt(id_vars='id')
assert len(long) == 2
print(long)
```

## Challenge 10

Downcast a numeric column and confirm memory shrinks.

```python
df = pd.DataFrame({'v': range(1000)})
before = df.memory_usage(deep=True).sum()
df['v'] = pd.to_numeric(df['v'], downcast='unsigned')
after = df.memory_usage(deep=True).sum()
assert after < before
print(before, '->', after)
```
